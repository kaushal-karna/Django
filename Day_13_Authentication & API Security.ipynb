    <h2 style="color: #17a2b8;">🔑 Token Authentication</h2>
    <p><strong>Simple and secure authentication using tokens</strong></p>
</div>

```python
class TokenAuthDemo:
    """Demonstrate Token Authentication"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 22 + "🔑 TOKEN AUTHENTICATION" + " " * 22)
        print("═" * 70)
        print()
        
        print("🔹 Setup Token Authentication:")
        print("─" * 70)
        print('''
  # settings.py
  INSTALLED_APPS = [
      ...
      'rest_framework',
      'rest_framework.authtoken',
      ...
  ]
  
  REST_FRAMEWORK = {
      'DEFAULT_AUTHENTICATION_CLASSES': [
          'rest_framework.authentication.TokenAuthentication',
          'rest_framework.authentication.SessionAuthentication',
      ],
      'DEFAULT_PERMISSION_CLASSES': [
          'rest_framework.permissions.IsAuthenticated',
      ]
  }
  
  # Create token for users
  python manage.py migrate
  python manage.py shell
  >>> from rest_framework.authtoken.models import Token
  >>> user = User.objects.get(username='ram')
  >>> token = Token.objects.create(user=user)
  >>> print(token.key)
  → 'abc123def456...' ''')
        
        print("\n🔹 Token Authentication View:")
        print("─" * 70)
        print('''
  # views.py
  from rest_framework.views import APIView
  from rest_framework.response import Response
  from rest_framework import status
  from rest_framework.permissions import IsAuthenticated
  from rest_framework.authtoken.models import Token
  from django.contrib.auth import authenticate

  class ObtainTokenView(APIView):
      """Obtain authentication token"""
      permission_classes = []  # Public endpoint
      
      def post(self, request):
          username = request.data.get('username')
          password = request.data.get('password')
          
          if not username or not password:
              return Response({
                  'error': 'Username and password required'
              }, status=status.HTTP_400_BAD_REQUEST)
          
          user = authenticate(username=username, password=password)
          if not user:
              return Response({
                  'error': 'Invalid credentials'
              }, status=status.HTTP_401_UNAUTHORIZED)
          
          token, created = Token.objects.get_or_create(user=user)
          return Response({
              'token': token.key,
              'user_id': user.id,
              'username': user.username,
          })
  
  class ProtectedView(APIView):
      """Example of a protected view"""
      permission_classes = [IsAuthenticated]
      
      def get(self, request):
          return Response({
              'message': 'You are authenticated!',
              'user': request.user.username,
              'token': request.auth.key
          })''')
        
        print("\n🔹 Using Token in Requests:")
        print("─" * 70)
        print('''
  # In Postman/Headers
  Authorization: Token abc123def456...
  
  # In curl
  curl -H "Authorization: Token abc123def456..." \
       http://127.0.0.1:8000/api/protected/
  
  # In Python requests
  headers = {'Authorization': f'Token {token}'}
  response = requests.get('http://127.0.0.1:8000/api/protected/', headers=headers)
  
  # In JavaScript fetch
  fetch('/api/protected/', {
      headers: {
          'Authorization': f'Token {token}'
      }
  })''')
        
        print()
        print("═" * 70)

token_auth = TokenAuthDemo()
token_auth.display()
```

    <p><strong>JSON Web Tokens for modern API authentication</strong></p>

```python
class JWTConceptDemo:
    """Demonstrate JWT Concepts"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 25 + "🪙 JWT CONCEPTS" + " " * 25)
        print("═" * 70)
        print()
        
        print("🔹 What is JWT?")
        print("─" * 70)
        print('''
  JSON Web Token (JWT) is an open standard (RFC 7519) that defines
  a compact and self-contained way for securely transmitting
  information between parties as a JSON object.
  
  📌 JWT Structure:
  xxxxx.yyyyy.zzzzz
  │     │     │
  │     │     └── Signature
  │     └── Payload
  └── Header''')
        
        print("\n🔹 JWT Components:")
        print("─" * 70)
        print('''
  1️⃣ HEADER (Base64 Encoded)
  {
      "alg": "HS256",
      "typ": "JWT"
  }
  
  2️⃣ PAYLOAD (Base64 Encoded)
  {
      "user_id": 1,
      "username": "ram",
      "email": "ram@example.com",
      "exp": 1700000000,  // Expiration time
      "iat": 1700000000,  // Issued at
      "jti": "unique-id"  // JWT ID
  }
  
  3️⃣ SIGNATURE
  HMACSHA256(
      base64UrlEncode(header) + "." + base64UrlEncode(payload),
      secret
  )''')
        
        print("\n🔹 JWT Setup with DRF:")
        print("─" * 70)
        print('''
  # Install
  pip install djangorestframework-simplejwt
  
  # settings.py
  REST_FRAMEWORK = {
      'DEFAULT_AUTHENTICATION_CLASSES': [
          'rest_framework_simplejwt.authentication.JWTAuthentication',
      ]
  }
  
  from datetime import timedelta
  
  SIMPLE_JWT = {
      'ACCESS_TOKEN_LIFETIME': timedelta(minutes=60),
      'REFRESH_TOKEN_LIFETIME': timedelta(days=7),
      'ROTATE_REFRESH_TOKENS': False,
      'BLACKLIST_AFTER_ROTATION': True,
      'ALGORITHM': 'HS256',
      'SIGNING_KEY': SECRET_KEY,
      'AUTH_HEADER_TYPES': ('Bearer',),
  }
  
  # urls.py
  from rest_framework_simplejwt.views import (
      TokenObtainPairView,
      TokenRefreshView,
      TokenVerifyView,
  )
  
  urlpatterns = [
      path('api/token/', TokenObtainPairView.as_view(), name='token_obtain_pair'),
      path('api/token/refresh/', TokenRefreshView.as_view(), name='token_refresh'),
      path('api/token/verify/', TokenVerifyView.as_view(), name='token_verify'),
  ]''')
        
        print("\n🔹 JWT vs Token Authentication:")
        print("─" * 70)
        print("""
  ┌─────────────────────────────────────────────────────────────────────┐
  │  Feature         │  Token Auth        │  JWT                      │
  │  ────────────────────────────────────────────────────────────────── │
  │  Database Lookup │  Yes (each request)│  No (self-contained)     │
  │  Scalability     │  Limited           │  Excellent               │
  │  Revocation      │  Easy              │  Hard (until expiry)     │
  │  Information     │  User only         │  User + custom data      │
  │  Expiry          │  Manual            │  Built-in                │
  │  Refresh Tokens  │  No                │  Yes                     │
  │  Use Case        │  Simple APIs       │  Modern/Microservices   │
  └─────────────────────────────────────────────────────────────────────┘""")
        
        print()
        print("═" * 70)

jwt = JWTConceptDemo()
jwt.display()
```

```python
class PermissionsDemo:
    """Demonstrate DRF Permissions"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 25 + "📋 PERMISSIONS IN DRF" + " " * 25)
        print("═" * 70)
        print()
        
        print("🔹 Built-in Permission Classes:")
        print("─" * 70)
        print('''
  from rest_framework import permissions
  
  # 1. AllowAny - No restrictions
  permission_classes = [permissions.AllowAny]
  
  # 2. IsAuthenticated - User must be logged in
  permission_classes = [permissions.IsAuthenticated]
  
  # 3. IsAdminUser - User must be staff/admin
  permission_classes = [permissions.IsAdminUser]
  
  # 4. IsAuthenticatedOrReadOnly - Read-only for unauthenticated
  permission_classes = [permissions.IsAuthenticatedOrReadOnly]
  
  # 5. DjangoModelPermissions - Based on model permissions
  permission_classes = [permissions.DjangoModelPermissions]
  
  # 6. DjangoObjectPermissions - Object-level permissions
  permission_classes = [permissions.DjangoObjectPermissions]''')
        
        print("\n🔹 Custom Permission Class:")
        print("─" * 70)
        print('''
  from rest_framework import permissions

  class IsOwnerOrReadOnly(permissions.BasePermission):
      """
      Custom permission to only allow owners of an object to edit it.
      """
      
      def has_object_permission(self, request, view, obj):
          # Read permissions are allowed to any request
          if request.method in permissions.SAFE_METHODS:
              return True
          
          # Write permissions are only allowed to the owner
          return obj.author == request.user

  class IsStaffOrReadOnly(permissions.BasePermission):
      """
      Custom permission: Staff can edit, others can only view.
      """
      
      def has_permission(self, request, view):
          if request.method in permissions.SAFE_METHODS:
              return True
          return request.user and request.user.is_staff

  # Using custom permissions in views
  class StudentViewSet(viewsets.ModelViewSet):
      queryset = Student.objects.all()
      serializer_class = StudentSerializer
      permission_classes = [IsOwnerOrReadOnly]''')
        
        print("\n🔹 Combining Permissions:")
        print("─" * 70)
        print('''
  from rest_framework.permissions import IsAuthenticated, IsAdminUser
  from rest_framework import permissions
  
  class CombinedPermissions(permissions.BasePermission):
      """
      Combined permission: Admin has full access, users have limited.
      """
      
      def has_permission(self, request, view):
          # Admin can do anything
          if request.user and request.user.is_staff:
              return True
          
          # Regular users can only read
          if request.method in permissions.SAFE_METHODS:
              return True
          
          # Others need to be authenticated
          return request.user and request.user.is_authenticated
  
  # OR using logical OR/AND
  from rest_framework.permissions import IsAuthenticated, IsAdminUser
  from rest_framework.permissions import OR, AND
  
  permission_classes = [IsAuthenticated & IsAdminUser]  # AND
  permission_classes = [IsAuthenticated | IsAdminUser]  # OR''')
        
        print()
        print("═" * 70)

permissions = PermissionsDemo()
permissions.display()
```

    <h2 style="color: #007bff;">⚖️ Authentication vs Authorization</h2>
    <p><strong>Understanding the key difference</strong></p>
</div>

```python
class AuthVsAuthzAPI:
    """Compare Authentication and Authorization"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 18 + "⚖️ AUTHENTICATION vs AUTHORIZATION" + " " * 18)
        print("═" * 70)
        print()
        
        print("🔹 Key Difference:")
        print("─" * 70)
        print("""
  ┌─────────────────────────────────────────────────────────────────────┐
  │  🔹 Authentication              🔹 Authorization                  │
  │  ─────────────────────          ─────────────────────             │
  │                                                                   │
  │  "Who are you?"                 "What can you do?"               │
  │                                                                   │
  │  • Verifies identity             • Checks permissions             │
  │  • Usually first step            • Second step                    │
  │  • Login credentials             • Roles/Groups/Permissions       │
  │  • Always required               • May not be required            │
  │  • Token generation              • Access control                 │
  │                                                                   │
  │  Example:                                                        │
  │  • Username/Password             • Admin can delete posts         │
  │  • Token validation              • User can only read             │
  │                                                                   │
  └─────────────────────────────────────────────────────────────────────┘""")
        
        print("\n🔹 In DRF:")
        print("─" * 70)
        print('''
  # Authentication (Who are you?)
  REST_FRAMEWORK = {
      'DEFAULT_AUTHENTICATION_CLASSES': [
          'rest_framework.authentication.TokenAuthentication',
          'rest_framework.authentication.SessionAuthentication',
      ],
  }
  
  # Authorization (What can you do?)
  REST_FRAMEWORK = {
      'DEFAULT_PERMISSION_CLASSES': [
          'rest_framework.permissions.IsAuthenticated',
          'rest_framework.permissions.DjangoModelPermissions',
      ],
  }
  
  # Together in a view
  class StudentViewSet(viewsets.ModelViewSet):
      authentication_classes = [TokenAuthentication]  # Auth
      permission_classes = [IsAuthenticated]          # Authorization''')
        
        print("\n🔹 Authentication Flow:")
        print("─" * 70)
        print("""
  🌐 Request → 🔐 Authentication → ✅ Identity Verified
                    │
                    ▼
              📋 Authorization → 🔑 Permissions Check
                    │
                    ├── ✅ Granted Access → Process Request
                    │
                    └── ❌ Denied Access → 403 Forbidden""")
        
        print()
        print("═" * 70)

auth_vs_authz = AuthVsAuthzAPI()
auth_vs_authz.display()
```

    <p><strong>Cross-Origin Resource Sharing explained</strong></p>

```python
class CORSDemo:
    """Demonstrate CORS Concepts"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 25 + "🌐 CORS CONCEPTS" + " " * 25)
        print("═" * 70)
        print()
        
        print("🔹 What is CORS?")
        print("─" * 70)
        print('''
  Cross-Origin Resource Sharing (CORS) is a mechanism that allows
  restricted resources on a web page to be requested from another
  domain outside the domain from which the first resource was served.
  
  📌 Same-Origin Policy:
  • Browsers restrict cross-origin requests by default
  • Scripts can only access resources from the same origin
  • CORS allows controlled access to cross-origin resources
  
  🔹 Example:
  • Frontend: https://myapp.com
  • API: https://api.myapp.com
  • CORS is needed to allow requests from frontend to API''')
        
        print("\n🔹 CORS Headers:")
        print("─" * 70)
        print('''
  # Response Headers (Server → Client)
  Access-Control-Allow-Origin: https://frontend.com
  Access-Control-Allow-Methods: GET, POST, PUT, DELETE, OPTIONS
  Access-Control-Allow-Headers: Authorization, Content-Type
  Access-Control-Allow-Credentials: true
  Access-Control-Max-Age: 86400
  
  # Request Headers (Client → Server)
  Origin: https://frontend.com
  Access-Control-Request-Method: POST
  Access-Control-Request-Headers: Authorization''')
        
        print("\n🔹 CORS Setup in Django:")
        print("─" * 70)
        print('''
  # Install
  pip install django-cors-headers
  
  # settings.py
  INSTALLED_APPS = [
      ...
      'corsheaders',
      ...
  ]
  
  MIDDLEWARE = [
      'corsheaders.middleware.CorsMiddleware',  # Add this
      ...
  ]
  
  # CORS Configuration
  CORS_ALLOWED_ORIGINS = [
      'http://localhost:3000',
      'http://127.0.0.1:3000',
      'https://yourdomain.com',
  ]
  
  CORS_ALLOW_METHODS = [
      'DELETE',
      'GET',
      'OPTIONS',
      'PATCH',
      'POST',
      'PUT',
  ]
  
  CORS_ALLOW_HEADERS = [
      'accept',
      'accept-encoding',
      'authorization',
      'content-type',
      'dnt',
      'origin',
      'user-agent',
      'x-csrftoken',
      'x-requested-with',
  ]
  
  CORS_ALLOW_CREDENTIALS = True
  
  # For development only
  CORS_ALLOW_ALL_ORIGINS = True  # Not for production''')
        
        print("\n🔹 CORS Preflight:")
        print("─" * 70)
        print("""
  📤 Preflight Request (OPTIONS):
  • Sent before actual request
  • Checks if server allows cross-origin request
  • Must respond with appropriate CORS headers
  
  🔄 Flow:
  🌐 Frontend → 🚀 OPTIONS Request → 🖥️ Backend
                    │
                    ▼
  🌐 Frontend ← 📨 CORS Headers ← 🖥️ Backend
                    │
                    ▼
  🌐 Frontend → 📤 Actual Request → 🖥️ Backend""")
        
        print()
        print("═" * 70)

cors = CORSDemo()
cors.display()
```

```python
import json
import hashlib
import datetime
from typing import Dict, List, Optional, Any

# Simulated Authentication System
class User:
    def __init__(self, username, email, password, is_staff=False):
        self.id = None
        self.username = username
        self.email = email
        self.password_hash = self.hash_password(password)
        self.is_active = True
        self.is_staff = is_staff
        self.is_superuser = is_staff
        self.token = None
        self.groups = []
        self.user_permissions = []
        self.created_at = datetime.datetime.now()
    
    @staticmethod
    def hash_password(password):
        return hashlib.sha256(password.encode()).hexdigest()
    
    def check_password(self, password):
        return self.password_hash == self.hash_password(password)
    
    def has_perm(self, permission):
        if self.is_superuser:
            return True
        if permission in self.user_permissions:
            return True
        for group in self.groups:
            if permission in group.permissions:
                return True
        return False

class Group:
    def __init__(self, name):
        self.name = name
        self.permissions = []

class Token:
    def __init__(self, user, key):
        self.user = user
        self.key = key
        self.created_at = datetime.datetime.now()

class AuthenticationSystem:
    def __init__(self):
        self.users = []
        self.tokens = []
        self.groups = []
        self.next_id = 1
        self.current_user = None
        self.current_token = None
    
    def create_group(self, name, permissions=None):
        group = Group(name)
        group.permissions = permissions or []
        self.groups.append(group)
        return group
    
    def register_user(self, username, email, password, is_staff=False):
        errors = {}
        
        if not username or len(username) < 3:
            errors['username'] = 'Username must be at least 3 characters'
        elif any(u.username == username for u in self.users):
            errors['username'] = 'Username already exists'
        
        if not email or '@' not in email:
            errors['email'] = 'Invalid email address'
        elif any(u.email == email for u in self.users):
            errors['email'] = 'Email already registered'
        
        if not password or len(password) < 8:
            errors['password'] = 'Password must be at least 8 characters'
        
        if errors:
            return {'success': False, 'errors': errors}
        
        user = User(username, email, password, is_staff)
        user.id = self.next_id
        self.next_id += 1
        self.users.append(user)
        return {'success': True, 'user': user}
    
    def login(self, username, password):
        user = next((u for u in self.users if u.username == username), None)
        
        if not user:
            return {'success': False, 'message': 'Invalid credentials'}
        
        if not user.is_active:
            return {'success': False, 'message': 'Account is deactivated'}
        
        if not user.check_password(password):
            return {'success': False, 'message': 'Invalid credentials'}
        
        # Generate token
        token_key = hashlib.sha256(f"{user.username}{datetime.datetime.now()}".encode()).hexdigest()
        token = Token(user, token_key)
        self.tokens.append(token)
        user.token = token_key
        self.current_user = user
        self.current_token = token
        
        return {
            'success': True,
            'user': user,
            'token': token_key
        }
    
    def authenticate_request(self, token_key):
        token = next((t for t in self.tokens if t.key == token_key), None)
        if not token:
            return None
        # Check token expiry (simulated - 1 hour)
        if (datetime.datetime.now() - token.created_at).seconds > 3600:
            self.tokens.remove(token)
            return None
        return token.user
    
    def logout(self, token_key):
        token = next((t for t in self.tokens if t.key == token_key), None)
        if token:
            self.tokens.remove(token)
            if self.current_token == token:
                self.current_user = None
                self.current_token = None
            return True
        return False
    
    def assign_permission(self, user, permission):
        if permission not in user.user_permissions:
            user.user_permissions.append(permission)
        return True
    
    def assign_group(self, user, group):
        if group not in user.groups:
            user.groups.append(group)
        return True

class CompleteAuthDemo:
    """Complete authentication system demonstration"""
    
    def run_demo(self):
        print("\n" + "═" * 70)
        print(" " * 15 + "🔧 COMPLETE AUTHENTICATION SYSTEM DEMO" + " " * 15)
        print("═" * 70)
        print()
        
        auth = AuthenticationSystem()
        
        # 1. Create Groups
        print("🔹 1. Creating Groups:")
        print("─" * 70)
        admin_group = auth.create_group('Admin', ['manage_users', 'manage_content', 'view_all'])
        editor_group = auth.create_group('Editor', ['create_content', 'edit_content', 'view_content'])
        viewer_group = auth.create_group('Viewer', ['view_content'])
        print("  ✅ Created: Admin, Editor, Viewer groups")
        
        # 2. Register Users
        print("\n🔹 2. User Registration:")
        print("─" * 70)
        
        users_data = [
            ('admin', 'admin@example.com', 'AdminPass123', True),
            ('editor', 'editor@example.com', 'EditorPass123', False),
            ('viewer', 'viewer@example.com', 'ViewerPass123', False),
            ('user1', 'user1@example.com', 'UserPass1234', False),
        ]
        
        for username, email, password, is_staff in users_data:
            result = auth.register_user(username, email, password, is_staff)
            if result['success']:
                user = result['user']
                print(f"  ✅ Registered: {user.username} ({user.email})")
                
                # Assign to groups
                if username == 'admin':
                    auth.assign_group(user, admin_group)
                elif username == 'editor':
                    auth.assign_group(user, editor_group)
                else:
                    auth.assign_group(user, viewer_group)
            else:
                print(f"  ❌ Failed: {result['errors']}")
        
        # 3. Login
        print("\n🔹 3. User Login:")
        print("─" * 70)
        
        login_attempts = [
            ('admin', 'AdminPass123'),
            ('editor', 'EditorPass123'),
            ('viewer', 'ViewerPass123'),
            ('user1', 'UserPass1234'),
        ]
        
        tokens = {}
        for username, password in login_attempts:
            result = auth.login(username, password)
            if result['success']:
                tokens[username] = result['token']
                print(f"  ✅ {username} logged in (Token: {result['token'][:20]}...)")
            else:
                print(f"  ❌ {username} login failed: {result['message']}")
        
        # 4. Token Authentication
        print("\n🔹 4. Token Authentication:")
        print("─" * 70)
        
        for username, token in tokens.items():
            user = auth.authenticate_request(token)
            if user:
                print(f"  ✅ Token valid for: {user.username}")
            else:
                print(f"  ❌ Invalid token for: {username}")
        
        # 5. Permission Checking
        print("\n🔹 5. Permission Checking:")
        print("─" * 70)
        
        admin_user = auth.users[0]
        editor_user = auth.users[1]
        viewer_user = auth.users[2]
        
        permissions_to_check = ['manage_users', 'edit_content', 'view_content', 'delete_content']
        
        print("  Admin Permissions:")
        for perm in permissions_to_check:
            has_perm = admin_user.has_perm(perm)
            status = "✅" if has_perm else "❌"
            print(f"    {status} {perm}: {has_perm}")
        
        print("\n  Editor Permissions:")
        for perm in permissions_to_check:
            has_perm = editor_user.has_perm(perm)
            status = "✅" if has_perm else "❌"
            print(f"    {status} {perm}: {has_perm}")
        
        print("\n  Viewer Permissions:")
        for perm in permissions_to_check:
            has_perm = viewer_user.has_perm(perm)
            status = "✅" if has_perm else "❌"
            print(f"    {status} {perm}: {has_perm}")
        
        # 6. Protected Endpoint Simulation
        print("\n🔹 6. Protected Endpoint Access:")
        print("─" * 70)
        
        def simulate_api_request(user, token, endpoint, method='GET'):
            # Authenticate
            authenticated_user = auth.authenticate_request(token)
            if not authenticated_user:
                return {'status': 401, 'message': 'Unauthorized'}
            
            # Check permissions based on endpoint
            if endpoint == '/api/admin/' and not user.has_perm('manage_users'):
                return {'status': 403, 'message': 'Forbidden - Admin access required'}
            
            if endpoint == '/api/edit/' and not user.has_perm('edit_content'):
                return {'status': 403, 'message': 'Forbidden - Editor access required'}
            
            return {'status': 200, 'message': 'Access granted', 'user': user.username}
        
        test_cases = [
            (admin_user, tokens['admin'], '/api/admin/', 'Admin accessing admin'),
            (editor_user, tokens['editor'], '/api/admin/', 'Editor accessing admin'),
            (editor_user, tokens['editor'], '/api/edit/', 'Editor accessing edit'),
            (viewer_user, tokens['viewer'], '/api/edit/', 'Viewer accessing edit'),
            (viewer_user, 'invalid_token', '/api/view/', 'Invalid token'),
        ]
        
        for user, token, endpoint, description in test_cases:
            result = simulate_api_request(user, token, endpoint)
            status_icon = "✅" if result['status'] == 200 else "❌"
            print(f"  {status_icon} {description}: {result['message']} (Status: {result['status']})")
        
        # 7. Logout
        print("\n🔹 7. Logout:")
        print("─" * 70)
        
        for username, token in tokens.items():
            if auth.logout(token):
                print(f"  ✅ {username} logged out successfully")
            else:
                print(f"  ❌ {username} logout failed")
        
        print("\n" + "═" * 70)
        print("🎯 Complete Authentication System Demo Complete!")
        print("═" * 70)

demo = CompleteAuthDemo()
demo.run_demo()
```
