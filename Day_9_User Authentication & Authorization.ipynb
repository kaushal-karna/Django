```python
class RegistrationDemo:
    """Demonstrate user registration"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 22 + "📝 USER REGISTRATION" + " " * 22)
        print("═" * 70)
        print()
        
        print("🔹 forms.py - Registration Form:")
        print("─" * 70)
        print('''
  from django import forms
  from django.contrib.auth.models import User
  from django.contrib.auth.forms import UserCreationForm
  
  class UserRegistrationForm(UserCreationForm):
      email = forms.EmailField(required=True, help_text='Required')
      first_name = forms.CharField(max_length=30, required=True)
      last_name = forms.CharField(max_length=30, required=True)
      
      class Meta:
          model = User
          fields = ['username', 'email', 'first_name', 'last_name', 
                   'password1', 'password2']
      
      def clean_email(self):
          email = self.cleaned_data.get('email')
          if User.objects.filter(email=email).exists():
              raise forms.ValidationError('This email is already registered')
          return email
      
      def save(self, commit=True):
          user = super().save(commit=False)
          user.email = self.cleaned_data['email']
          user.first_name = self.cleaned_data['first_name']
          user.last_name = self.cleaned_data['last_name']
          if commit:
              user.save()
          return user''')
        
        print("\n🔹 views.py - Registration View:")
        print("─" * 70)
        print('''
  from django.shortcuts import render, redirect
  from django.contrib import messages
  from .forms import UserRegistrationForm
  
  def register(request):
      if request.method == 'POST':
          form = UserRegistrationForm(request.POST)
          if form.is_valid():
              user = form.save()
              messages.success(request, f'Account created for {user.username}!')
              return redirect('login')
      else:
          form = UserRegistrationForm()
      return render(request, 'register.html', {'form': form})''')
        
        print("\n🔹 urls.py - Registration URL:")
        print("─" * 70)
        print('''
  from django.urls import path
  from . import views
  
  urlpatterns = [
      path('register/', views.register, name='register'),
  ]''')
        
        print()
        print("═" * 70)

registration = RegistrationDemo()
registration.display()
```

    <h2 style="color: #007bff;">🔑 Login and Logout</h2>
    <p><strong>Authenticating users and managing sessions</strong></p>

```python
class LoginLogoutDemo:
    """Demonstrate login and logout"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 22 + "🔑 LOGIN AND LOGOUT" + " " * 22)
        print("═" * 70)
        print()
        
        print("🔹 views.py - Login View:")
        print("─" * 70)
        print('''
  from django.contrib.auth import authenticate, login, logout
  from django.contrib.auth.forms import AuthenticationForm
  from django.shortcuts import render, redirect
  from django.contrib import messages
  
  def login_user(request):
      if request.method == 'POST':
          form = AuthenticationForm(request, data=request.POST)
          if form.is_valid():
              username = form.cleaned_data.get('username')
              password = form.cleaned_data.get('password')
              user = authenticate(username=username, password=password)
              if user is not None:
                  login(request, user)
                  messages.success(request, f'Welcome back, {username}!')
                  # Redirect to next parameter or homepage
                  next_url = request.GET.get('next', 'home')
                  return redirect(next_url)
              else:
                  messages.error(request, 'Invalid username or password')
          else:
              messages.error(request, 'Invalid username or password')
      else:
          form = AuthenticationForm()
      return render(request, 'login.html', {'form': form})
  
  def logout_user(request):
      logout(request)
      messages.success(request, 'You have been logged out')
      return redirect('login')''')
        
        print("\n🔹 Login Flow:")
        print("─" * 70)
        print("""
  ┌─────────────────────────────────────────────────────────────┐
  │  1️⃣ User submits login form                                │
  │  2️⃣ authenticate() checks username and password           │
  │  3️⃣ If valid, login() creates session                     │
  │  4️⃣ User is redirected to dashboard                       │
  │  5️⃣ request.user is now available in all views            │
  └─────────────────────────────────────────────────────────────┘""")
        
        print("\n🔹 Using Django's Built-in Login/Logout:")
        print("─" * 70)
        print('''
  from django.contrib.auth import views as auth_views
  
  urlpatterns = [
      path('login/', auth_views.LoginView.as_view(), name='login'),
      path('logout/', auth_views.LogoutView.as_view(), name='logout'),
  ]
  
  # Add to settings.py
  LOGIN_URL = '/login/'
  LOGIN_REDIRECT_URL = '/dashboard/'
  LOGOUT_REDIRECT_URL = '/login/''')
        
        print()
        print("═" * 70)

login_logout = LoginLogoutDemo()
login_logout.display()
```

```python
class PageProtection:
    """Demonstrate page protection"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 25 + "🛡️ PROTECTING PAGES" + " " * 25)
        print("═" * 70)
        print()
        
        print("🔹 Method 1: @login_required Decorator:")
        print("─" * 70)
        print('''
  from django.contrib.auth.decorators import login_required
  from django.shortcuts import render
  
  @login_required
  def dashboard(request):
      return render(request, 'dashboard.html')
  
  @login_required(login_url='/custom-login/')
  def profile(request):
      return render(request, 'profile.html')
  
  # Custom decorator for multiple conditions
  def staff_required(view_func):
      @login_required
      def wrapper(request, *args, **kwargs):
          if not request.user.is_staff:
              return redirect('home')
          return view_func(request, *args, **kwargs)
      return wrapper''')
        
        print("\n🔹 Method 2: LoginRequiredMixin (Class-Based Views):")
        print("─" * 70)
        print('''
  from django.contrib.auth.mixins import LoginRequiredMixin
  from django.views.generic import TemplateView
  
  class DashboardView(LoginRequiredMixin, TemplateView):
      template_name = 'dashboard.html'
      login_url = '/login/'
      redirect_field_name = 'redirect_to'
  
  # Multiple mixins
  from django.contrib.auth.mixins import PermissionRequiredMixin
  
  class AdminView(LoginRequiredMixin, PermissionRequiredMixin, TemplateView):
      template_name = 'admin.html'
      permission_required = 'app.can_manage_users' ''')
        
        print("\n🔹 Method 3: In-View Checks:")
        print("─" * 70)
        print('''
  def user_profile(request, user_id):
      user = get_object_or_404(User, id=user_id)
      # Check if user is authenticated
      if not request.user.is_authenticated:
          return redirect('login')
      # Check if user has permission
      if not request.user.is_staff and request.user != user:
          return redirect('home')
      return render(request, 'profile.html', {'user': user})''')
        
        print("\n🔹 settings.py - Configuration:")
        print("─" * 70)
        print('''
  # Login URL (where users are redirected if not authenticated)
  LOGIN_URL = '/login/'
  
  # Redirect after successful login
  LOGIN_REDIRECT_URL = '/dashboard/'
  
  # Redirect after logout
  LOGOUT_REDIRECT_URL = '/login/'
  
  # Session settings
  SESSION_COOKIE_AGE = 3600  # 1 hour
  SESSION_EXPIRE_AT_BROWSER_CLOSE = True''')
        
        print()
        print("═" * 70)

protection = PageProtection()
protection.display()
```

    <h2 style="color: #fd7e14;">🔑 Permissions and Authorization</h2>
    <p><strong>Fine-grained access control</strong></p>

```python
class PermissionsDemo:
    """Demonstrate permissions and authorization"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 20 + "🔑 PERMISSIONS & AUTHORIZATION" + " " * 20)
        print("═" * 70)
        print()
        
        print("🔹 Built-in Permissions:")
        print("─" * 70)
        print('''
  # Django automatically creates these for each model
  from django.contrib.auth.models import Permission
  
  # Available permissions for a model 'Post'
  permissions = [
      'add_post',     # Can add posts
      'change_post',  # Can change posts
      'delete_post',  # Can delete posts
      'view_post',    # Can view posts
  ]''')
        
        print("\n🔹 Creating Custom Permissions:")
        print("─" * 70)
        print('''
  # models.py
  class Post(models.Model):
      title = models.CharField(max_length=200)
      content = models.TextField()
      author = models.ForeignKey(User, on_delete=models.CASCADE)
      
      class Meta:
          permissions = [
              ('can_publish_post', 'Can publish posts'),
              ('can_archive_post', 'Can archive posts'),
              ('can_approve_comments', 'Can approve comments'),
          ]''')
        
        print("\n🔹 Checking Permissions:")
        print("─" * 70)
        print('''
  # In views
  @permission_required('app.can_publish_post')
  def publish_post(request, post_id):
      post = get_object_or_404(Post, id=post_id)
      post.is_published = True
      post.save()
      return redirect('post_detail', post_id=post.id)
  
  # Conditional checks in templates
  {% if perms.app.can_publish_post %}
      <button>Publish</button>
  {% endif %}
  
  # Check in views manually
  if request.user.has_perm('app.can_publish_post'):
      # Allow publishing
  else:
      # Deny access
  
  # Check multiple permissions
  if request.user.has_perms(['app.can_publish_post', 'app.can_archive_post']):
      # Has both permissions''')
        
        print("\n🔹 Assigning Permissions:")
        print("─" * 70)
        print('''
  # To a user directly
  from django.contrib.auth.models import Permission
  
  permission = Permission.objects.get(codename='can_publish_post')
  user.user_permissions.add(permission)
  
  # To a group
  from django.contrib.auth.models import Group
  
  editors_group = Group.objects.create(name='Editors')
  editors_group.permissions.set([permission])
  user.groups.add(editors_group)
  
  # Remove permission
  user.user_permissions.remove(permission)''')
        
        print()
        print("═" * 70)

permissions = PermissionsDemo()
permissions.display()
```

<div style="background: #f8f9fa; padding: 20px; border-radius: 10px; border-left: 5px solid #e83e8c;">
    <h2 style="color: #e83e8c;">👥 Role-Based Access Control (RBAC)</h2>
    <p><strong>Managing access through user roles</strong></p>
</div>

```python
class RBACDemo:
    """Demonstrate Role-Based Access Control"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 18 + "👥 ROLE-BASED ACCESS CONTROL" + " " * 18)
        print("═" * 70)
        print()
        
        print("🔹 Creating Roles/Groups:")
        print("─" * 70)
        print('''
  from django.contrib.auth.models import Group, Permission
  
  # Create groups
  admin_group = Group.objects.create(name='Admin')
  editor_group = Group.objects.create(name='Editor')
  viewer_group = Group.objects.create(name='Viewer')
  
  # Assign permissions to groups
  admin_permissions = Permission.objects.filter(
      content_type__app_label='app'
  )
  admin_group.permissions.set(admin_permissions)
  
  editor_permissions = Permission.objects.filter(
      codename__in=['add_post', 'change_post', 'view_post']
  )
  editor_group.permissions.set(editor_permissions)
  
  viewer_permissions = Permission.objects.filter(
      codename__in=['view_post']
  )
  viewer_group.permissions.set(viewer_permissions)
  
  # Add users to groups
  user.groups.add(editor_group)''')
        
        print("\n🔹 Custom Role Check Decorator:")
        print("─" * 70)
        print('''
  from django.core.exceptions import PermissionDenied
  from functools import wraps
  
  def role_required(allowed_roles):
      def decorator(view_func):
          @wraps(view_func)
          def wrapper(request, *args, **kwargs):
              if not request.user.is_authenticated:
                  return redirect('login')
              user_groups = request.user.groups.values_list('name', flat=True)
              if any(role in user_groups for role in allowed_roles):
                  return view_func(request, *args, **kwargs)
              raise PermissionDenied("You don't have permission to access this page")
          return wrapper
      return decorator
  
  # Usage
  @role_required(['Admin', 'Editor'])
  def manage_posts(request):
      # Only Admins and Editors can access
      return render(request, 'manage_posts.html')
  
  @role_required(['Admin'])
  def manage_users(request):
      # Only Admins can access
      return render(request, 'manage_users.html')''')
        
        print("\n🔹 Role-Based Template Controls:")
        print("─" * 70)
        print('''
  <!-- Check if user belongs to a group -->
  {% if user.groups.all.0.name == 'Admin' %}
      <div class="admin-panel">
          <h2>Admin Controls</h2>
          <button>Delete User</button>
          <button>Manage All Posts</button>
      </div>
  {% endif %}
  
  {% if user.groups.all.0.name == 'Editor' %}
      <div class="editor-panel">
          <h2>Editor Controls</h2>
          <button>Create Post</button>
          <button>Edit Post</button>
      </div>
  {% endif %}
  
  <!-- Check multiple groups -->
  {% if user.groups.all.0.name in 'Admin Editor' %}
      <button>Advanced Features</button>
  {% endif %}''')
        
        print()
        print("═" * 70)

rbac = RBACDemo()
rbac.display()
```

```python
import datetime
import hashlib
import re
from typing import Dict, List, Optional

# Simulated Django Authentication System
class User:
    """Simulated User model"""
    
    def __init__(self, username, email, password):
        self.id = None
        self.username = username
        self.email = email
        self.password_hash = self.hash_password(password)
        self.first_name = ''
        self.last_name = ''
        self.is_active = True
        self.is_staff = False
        self.is_superuser = False
        self.groups = []
        self.user_permissions = []
        self.last_login = None
        self.date_joined = datetime.datetime.now()
    
    @staticmethod
    def hash_password(password):
        return hashlib.sha256(password.encode()).hexdigest()
    
    def check_password(self, password):
        return self.password_hash == self.hash_password(password)
    
    def has_perm(self, permission):
        # Check direct permissions
        if permission in self.user_permissions:
            return True
        # Check group permissions
        for group in self.groups:
            if permission in group.permissions:
                return True
        # Superuser has all permissions
        return self.is_superuser
    
    def __str__(self):
        return self.username

class Group:
    """Simulated Group model"""
    
    def __init__(self, name):
        self.name = name
        self.permissions = []
    
    def __str__(self):
        return self.name

class AuthenticationSystem:
    """Complete authentication system"""
    
    def __init__(self):
        self.users = []
        self.groups = []
        self.next_id = 1
        self.current_user = None
    
    def create_group(self, name, permissions=None):
        group = Group(name)
        group.permissions = permissions or []
        self.groups.append(group)
        return group
    
    def register_user(self, username, email, password, confirm_password, **kwargs):
        """Register a new user"""
        errors = {}
        
        # Validation
        if not username or len(username) < 3:
            errors['username'] = 'Username must be at least 3 characters'
        elif any(u.username == username for u in self.users):
            errors['username'] = 'Username already exists'
        
        if not email or '@' not in email:
            errors['email'] = 'Invalid email address'
        elif any(u.email == email for u in self.users):
            errors['email'] = 'Email already registered'
        
        if not password or len(password) < 8:
            errors['password'] = 'Password must be at least 8 characters'
        elif not any(c.isupper() for c in password):
            errors['password'] = 'Password must contain at least one uppercase letter'
        elif not any(c.islower() for c in password):
            errors['password'] = 'Password must contain at least one lowercase letter'
        elif not any(c.isdigit() for c in password):
            errors['password'] = 'Password must contain at least one number'
        
        if password != confirm_password:
            errors['confirm_password'] = 'Passwords do not match'
        
        if errors:
            return {'success': False, 'errors': errors}
        
        # Create user
        user = User(username, email, password)
        user.id = self.next_id
        user.first_name = kwargs.get('first_name', '')
        user.last_name = kwargs.get('last_name', '')
        self.next_id += 1
        self.users.append(user)
        
        return {'success': True, 'user': user}
    
    def login_user(self, username, password):
        """Authenticate and login user"""
        user = next((u for u in self.users if u.username == username), None)
        
        if not user:
            return {'success': False, 'message': 'User not found'}
        
        if not user.is_active:
            return {'success': False, 'message': 'Account is deactivated'}
        
        if not user.check_password(password):
            return {'success': False, 'message': 'Invalid password'}
        
        user.last_login = datetime.datetime.now()
        self.current_user = user
        return {'success': True, 'user': user}
    
    def logout_user(self):
        """Logout current user"""
        self.current_user = None
        return {'success': True, 'message': 'Logged out successfully'}
    
    def change_password(self, user, old_password, new_password, confirm_password):
        """Change user password"""
        errors = {}
        
        if not user.check_password(old_password):
            errors['old_password'] = 'Current password is incorrect'
        
        if len(new_password) < 8:
            errors['new_password'] = 'Password must be at least 8 characters'
        
        if new_password != confirm_password:
            errors['confirm_password'] = 'Passwords do not match'
        
        if errors:
            return {'success': False, 'errors': errors}
        
        user.password_hash = User.hash_password(new_password)
        return {'success': True, 'message': 'Password changed successfully'}
    
    def assign_permission(self, user, permission):
        """Assign permission to user"""
        if permission not in user.user_permissions:
            user.user_permissions.append(permission)
        return True
    
    def assign_group(self, user, group):
        """Assign user to group"""
        if group not in user.groups:
            user.groups.append(group)
        return True

class AuthenticationDemo:
    """Run authentication demonstration"""
    
    def run_demo(self):
        print("\n" + "═" * 70)
        print(" " * 15 + "🔐 COMPLETE AUTHENTICATION SYSTEM DEMO" + " " * 15)
        print("═" * 70)
        print()
        
        auth = AuthenticationSystem()
        
        # 1. Create groups
        print("🔹 1. Creating Groups:")
        print("─" * 70)
        admin_group = auth.create_group('Admin', ['manage_users', 'manage_posts', 'view_all'])
        editor_group = auth.create_group('Editor', ['create_post', 'edit_post', 'view_post'])
        viewer_group = auth.create_group('Viewer', ['view_post'])
        print(f"  ✅ Created groups: Admin, Editor, Viewer")
        
        # 2. Register users
        print("\n🔹 2. User Registration:")
        print("─" * 70)
        
        users_data = [
            ('admin', 'admin@example.com', 'AdminPass123', 'AdminPass123', 
             {'first_name': 'Admin', 'last_name': 'User'}),
            ('editor', 'editor@example.com', 'EditorPass123', 'EditorPass123',
             {'first_name': 'Editor', 'last_name': 'User'}),
            ('viewer', 'viewer@example.com', 'ViewerPass123', 'ViewerPass123',
             {'first_name': 'Viewer', 'last_name': 'User'}),
        ]
        
        for username, email, password, confirm, extra in users_data:
            result = auth.register_user(username, email, password, confirm, **extra)
            if result['success']:
                user = result['user']
                print(f"  ✅ Registered: {user.username} ({user.email})")
                # Assign to appropriate group
                if username == 'admin':
                    auth.assign_group(user, admin_group)
                    user.is_superuser = True
                    user.is_staff = True
                elif username == 'editor':
                    auth.assign_group(user, editor_group)
                else:
                    auth.assign_group(user, viewer_group)
            else:
                print(f"  ❌ Failed to register {username}: {result['errors']}")
        
        # 3. Login attempts
        print("\n🔹 3. Login Attempts:")
        print("─" * 70)
        
        login_tests = [
            ('admin', 'AdminPass123', '✅ Correct credentials'),
            ('admin', 'WrongPass', '❌ Wrong password'),
            ('unknown', 'password', '❌ User not found'),
        ]
        
        for username, password, description in login_tests:
            result = auth.login_user(username, password)
            if result['success']:
                print(f"  {description}: {username} logged in successfully")
                auth.logout_user()  # Logout for next test
            else:
                print(f"  {description}: {result['message']}")
        
        # 4. Login as admin and check permissions
        print("\n🔹 4. Permission Checking:")
        print("─" * 70)
        auth.login_user('admin', 'AdminPass123')
        admin = auth.current_user
        print(f"  ✅ Logged in as: {admin.username}")
        
        permissions_to_check = ['manage_users', 'manage_posts', 'view_all', 'delete_post']
        for perm in permissions_to_check:
            has_perm = admin.has_perm(perm)
            status = "✅" if has_perm else "❌"
            print(f"  {status} {perm}: {has_perm}")
        
        # 5. Login as editor and check permissions
        auth.logout_user()
        auth.login_user('editor', 'EditorPass123')
        editor = auth.current_user
        print(f"\n  ✅ Logged in as: {editor.username}")
        
        for perm in ['create_post', 'edit_post', 'view_post', 'manage_users']:
            has_perm = editor.has_perm(perm)
            status = "✅" if has_perm else "❌"
            print(f"  {status} {perm}: {has_perm}")
        
        # 6. Change password
        print("\n🔹 5. Password Change:")
        print("─" * 70)
        result = auth.change_password(editor, 'EditorPass123', 'NewPass456', 'NewPass456')
        if result['success']:
            print(f"  ✅ Password changed successfully")
        else:
            print(f"  ❌ Failed: {result['errors']}")
        
        # 7. Logout
        print("\n🔹 6. Logout:")
        print("─" * 70)
        result = auth.logout_user()
        print(f"  ✅ {result['message']}")
        print(f"  Current user: {auth.current_user}")
        
        print("\n" + "═" * 70)
        print("🎯 Authentication System Demo Complete!")
        print("═" * 70)

demo = AuthenticationDemo()
demo.run_demo()
```
