```python
class PythonSyntax:
    """Demonstrate Python syntax rules"""
    
    def display(self):
        print("\n" + "╔" + "═" * 70 + "╗")
        print("║" + " " * 22 + "📝 PYTHON SYNTAX RULES" + " " * 23 + "║")
        print("╚" + "═" * 70 + "╝")
        print()
        
        print("🔹 Python Syntax Rules:")
        print("─" * 72)
        print("  ✅ Indentation is mandatory (4 spaces recommended)")
        print("  ✅ No semicolons (;) required at the end of lines")
        print("  ✅ Case-sensitive language (name != Name)")
        print("  ✅ Comments start with #")
        print("  ✅ Multi-line strings use ''' or \"\"\"")
        
        print("\n🔹 Example:")
        print("─" * 72)
        print('''  # This is a comment
  name = "Ram"  # Variable assignment
  
  def greet():  # Function definition
      print("Hello, Django!")  # Indented block
  
  greet()  # Function call''')
        
        # Execute the example
        print("\n💡 Let's write our first Python program!")
        print("─" * 72)
        
        name = "Ram"
        
        def greet():
            print("  Hello, Django!")
            print("  Welcome to Python Fundamentals!")
        
        greet()
        
        print("\n" + "═" * 70)

syntax = PythonSyntax()
syntax.display()
```

    <p>Conditional statements allow you to make decisions in your code.</p>

```python
class ConditionalDemo:
    """Demonstrate conditional statements"""
    
    def display(self):
        print("\n" + "╔" + "═" * 70 + "╗")
        print("║" + " " * 22 + "🔀 CONDITIONAL STATEMENTS" + " " * 23 + "║")
        print("╚" + "═" * 70 + "╝")
        
        # Simple if
        print("\n🔹 if Statement:")
        print("─" * 72)
        age = 20
        print(f"  Age: {age}")
        if age >= 18:
            print("  ✅ Adult")
        else:
            print("  ❌ Minor")
        
        # if-elif-else
        print("\n🔹 if-elif-else Statement:")
        print("─" * 72)
        score = 85
        print(f"  Score: {score}")
        if score >= 90:
            grade = 'A'
        elif score >= 80:
            grade = 'B'
        elif score >= 70:
            grade = 'C'
        else:
            grade = 'F'
        print(f"  Grade: {grade}")
        
        # Nested conditions
        print("\n🔹 Nested Conditions:")
        print("─" * 72)
        age = 22
        course = "CSIT"
        print(f"  Age: {age}, Course: {course}")
        if age >= 18:
            if course == "CSIT":
                print("  ✅ Eligible for scholarship")
            else:
                print("  ❌ Scholarship only for CSIT students")
        else:
            print("  ❌ Must be 18 or older")
        
        # Django backend logic
        print("\n🔹 Django Backend Logic Simulation:")
        print("─" * 72)
        username = "ram"
        password = "secret123"
        print(f"  Username: {username}, Password: {password}")
        
        if username == "ram" and password == "secret123":
            print("  ✅ Login successful! Redirecting to dashboard...")
        else:
            print("  ❌ Invalid credentials! Show error message...")
        
        # Multiple conditions
        print("\n🔹 Multiple Conditions:")
        print("─" * 72)
        user_role = "admin"
        is_authenticated = True
        print(f"  User: {user_role}")
        print(f"  Role: {user_role}")
        
        if is_authenticated and user_role == "admin":
            print("  Access granted to admin panel")
        elif is_authenticated and user_role == "student":
            print("  Access granted to student dashboard")
        else:
            print("  Access denied")
        
        print("\n💡 Conditional statements are used EVERYWHERE in Django!")
        print()
        print("═" * 70)

conditionals = ConditionalDemo()
conditionals.display()
```

    <p>Loops are used to repeat a block of code multiple times.</p>

```python
class LoopsDemo:
    """Demonstrate loops in Python"""
    
    def display(self):
        print("\n" + "╔" + "═" * 70 + "╗")
        print("║" + " " * 28 + "🔄 LOOPS" + " " * 29 + "║")
        print("╚" + "═" * 70 + "╝")
        
        # For loop with range
        print("\n🔹 For Loop - Range:")
        print("─" * 72)
        print("  for i in range(5):")
        for i in range(5):
            print(f"    i = {i}")
        
        # For loop with list
        print("\n🔹 For Loop - List Iteration:")
        print("─" * 72)
        students = ["Ram", "Sita", "Hari"]
        print(f"  Students list: {students}")
        for student in students:
            print(f"  Student: {student}")
        
        # While loop
        print("\n🔹 While Loop:")
        print("─" * 72)
        count = 0
        while count < 3:
            print(f"  count = {count}")
            count += 1
        
        # Break and Continue
        print("\n🔹 Loop Control - Break and Continue:")
        print("─" * 72)
        print("  break statement:")
        students = ["Ram", "Sita", "Hari"]
        for i, student in enumerate(students):
            if student == "Sita":
                print(f"  Found 'Sita' at index {i}")
                break
        
        print("\n  continue statement (skip even numbers):")
        for i in range(10):
            if i % 2 == 0:
                continue
            print(f"  {i}")
        
        # Nested loops
        print("\n🔹 Nested Loops - Multiplication Table:")
        print("─" * 72)
        for i in range(1, 6):
            print(f"  2 x {i} = {2 * i}")
        
        print("\n💡 Loops in Django:")
        print("─" * 72)
        print("  • Displaying lists of data in templates")
        print("  • Processing form data")
        print("  • Iterating through database records")
        print("  • Generating dynamic content")
        
        print()
        print("═" * 70)

loops = LoopsDemo()
loops.display()
```

```python
class FunctionsDemo:
    """Demonstrate functions in Python"""
    
    def display(self):
        print("\n" + "╔" + "═" * 70 + "╗")
        print("║" + " " * 28 + "📦 FUNCTIONS" + " " * 29 + "║")
        print("╚" + "═" * 70 + "╝")
        
        # Basic function
        print("\n🔹 Basic Function:")
        print("─" * 72)
        def greet(name):
            return f"  Hello {name}!"
        print(greet("Ram"))
        
        # Function with parameters
        print("\n🔹 Function with Parameters:")
        print("─" * 72)
        def introduce(name, age):
            return f"  {name} - {age} years old"
        print(introduce("Ram", 21))
        print(introduce("Sita", 22))
        
        # Function with return value
        print("\n🔹 Function with Return Value:")
        print("─" * 72)
        def calculate_grade(name, score):
            if score >= 90:
                grade = 'A'
            elif score >= 80:
                grade = 'B'
            elif score >= 70:
                grade = 'C'
            else:
                grade = 'F'
            return f"  {name}: {grade}"
        
        print(calculate_grade("Ram", 92))
        print(calculate_grade("Sita", 78))
        print(calculate_grade("Hari", 55))
        
        # Function with default parameters
        print("\n🔹 Function with Default Parameters:")
        print("─" * 72)
        def create_student(name, age=18):
            return f"  Student: {name}, Age: {age}"
        print(create_student("Ram"))
        print(create_student("Sita", 20))
        
        # Django view simulation
        print("\n🔹 Django View Function Example:")
        print("─" * 72)
        def student_detail(student_id):
            # Simulating database lookup
            students = {
                1: {"name": "Ram", "course": "CSIT"},
                2: {"name": "Sita", "course": "BBA"},
            }
            if student_id in students:
                student = students[student_id]
                print(f"  Student ID: {student_id}")
                print(f"  Student Name: {student['name']}")
                print(f"  Student Course: {student['course']}")
            else:
                print(f"  Student not found")
        
        student_detail(1)
        
        print("\n💡 Functions in Django:")
        print("─" * 72)
        print("  • Views are functions that handle HTTP requests")
        print("  • Helper functions for reusable logic")
        print("  • Model methods for database operations")
        print("  • Custom template filters")
        
        print()
        print("═" * 70)

functions = FunctionsDemo()
functions.display()
```

    <p>Write a program that calculates the factorial of a number.</p>
</div>

```python
class ExerciseFactorial:
    """Exercise: Calculate factorial"""
    
    def factorial(self, n):
        if n == 0:
            return 1
        result = 1
        for i in range(1, n + 1):
            result *= i
        return result
    
    def run_exercise(self):
        print("\n" + "═" * 70)
        print("📊 EXERCISE 3: FACTORIAL CALCULATOR")
        print("═" * 70)
        
        test_numbers = [1, 3, 5, 7, 10]
        
        print("\n🔹 Test Cases:")
        print("─" * 72)
        for num in test_numbers:
            fact = self.factorial(num)
            print(f"  {num}! = {fact}")
        
        print("\n💡 How it works:")
        print("─" * 72)
        print("  • n! = n × (n-1) × (n-2) × ... × 1")
        print("  • Uses a loop to multiply numbers")
        print("  • Special case: 0! = 1")
        
        print("\n✅ Your Turn! Try calculating factorial of larger numbers.")
        print()
        print("═" * 70)

exercise3 = ExerciseFactorial()
exercise3.run_exercise()
```

```python
# 📊 Exercise 3: Factorial Calculator
# Write a program that calculates the factorial of a number.

num = int(input("Enter a number to calculate its factorial: "))

def factorial(n):
    if n ==0:
        return 1
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result


fact = factorial(num)
print(f"The factorial of {num} is: {fact}")
```

```python
class ExerciseCalculator:
    """Exercise: Simple calculator"""
    
    def calculate(self, a, b, operation):
        if operation == '+':
            return a + b
        elif operation == '-':
            return a - b
        elif operation == '*':
            return a * b
        elif operation == '/':
            if b != 0:
                return a / b
            else:
                return "Error: Division by zero"
        elif operation == '%':
            return a % b
        else:
            return "Invalid operation"
    
    def run_exercise(self):
        print("\n" + "═" * 70)
        print("🧮 EXERCISE 4: SIMPLE CALCULATOR")
        print("═" * 70)
        
        test_cases = [
            (10, 5, '+'),
            (10, 5, '-'),
            (10, 5, '*'),
            (10, 5, '/'),
            (10, 0, '/'),
            (10, 5, '%')
        ]
        
        print("\n🔹 Test Cases:")
        print("─" * 72)
        for a, b, op in test_cases:
            result = self.calculate(a, b, op)
            print(f"  {a} {op} {b} = {result}")
        
        print("\n💡 How it works:")
        print("─" * 72)
        print("  • Takes two numbers and an operation")
        print("  • Performs the requested operation")
        print("  • Handles division by zero")
        print("  • Returns the result")
        
        print("\n✅ Your Turn! Add more operations (power, square root, etc.).")
        print()
        print("═" * 70)

exercise4 = ExerciseCalculator()
exercise4.run_exercise()
```

    <h3 style="color: #e83e8c; margin-top: 0;">📋 Exercise 5: Student Management System</h3>
    <p>Build a simple student management system with functions.</p>
</div>

```python
class ExerciseStudentSystem:
    """Exercise: Student management system"""
    
    def __init__(self):
        self.students = []
    
    def add_student(self, name, age, course):
        student = {
            "name": name,
            "age": age,
            "course": course
        }
        self.students.append(student)
        return f"📝 {name} added successfully"
              
              
    def list_students(self):
        return self.students
    
    def find_student(self, name):
        for student in self.students:
            if student["name"].lower() == name.lower():
                return student
        return None
    
    def count_students(self):
        return len(self.students)
    
    def update_student(self, name, age, course):
            for student in self.students:
                if student["name"].lower() == name.lower():
                    student.update({
                        "age": age,
                        "course": course
                    })
                    return f"✏️ {name} updated successfully to Age: {age}, Course: {course}"
            return f"❌ {name} not found"
        
    def delete_student(self, name):
        for student in self.students:
            if student["name"].lower() == name.lower():
                self.students.remove(student)
                return f"🗑️ {name} deleted successfully"
        return f"❌ {name} not found"
    
    def search_by_age(self, age):
        results = [student for student in self.students if student["age"] == age]
        return results if results else None 
    
    
    def run_exercise(self):
        print("\n" + "═" * 70)
        print("📋 EXERCISE 5: STUDENT MANAGEMENT SYSTEM")
        print("═" * 70)
        
        # Add students
        print("\n🔹 Test Cases:")
        print("─" * 72)
        
        print("\n✅ Adding students:")
        print("─" * 72)
        students_data = [
            ("Ram Sharma", 21, "CSIT"),
            ("Sita Patel", 22, "BBA"),
            ("Hari Gurung", 20, "BSc IT")
        ]
        
        for name, age, course in students_data:
            result = self.add_student(name, age, course)
            print(f"  {result}")
        
        # List all students
        print("\n👥 All Students:")
        print("─" * 72)
        for student in self.list_students():
            print(f"  📚 {student['name']} - {student['age']} - {student['course']}")
        
        # Find a student
        print("\n🔍 Finding students:")
        print("─" * 72)
        found = self.find_student("Ram Sharma")
        if found:
            print(f"  Found: {found['name']} - {found['age']} - {found['course']}")
        
        # Count students
        print(f"\n📊 Total students: {self.count_students()}")
        
        print("\n💡 How it works:")
        print("─" * 72)
        print("  • Uses a list to store students")
        print("  • Functions for adding, listing, finding")
        print("  • Dictionaries for student data")
        
        # Updating students data
        print("\n Updating students:")
        print("_" * 72)
        update = self.update_student("Ram Sharma", 22, "BSc CSIT")
        print(f"  {update}")
        
        # Deleting a student
        print("\n Deleting a student:")
        print("_" * 72)
        delete = self.delete_student("Sita Patel")
        print(f"  {delete}")
        
        # Searching by age
        print("\n Searching by age:")
        print("_" * 72)
        search_results = self.search_by_age(22)
        if search_results:
            for student in search_results:
                print(f"  📚 {student['name']} - {student['age']} - {student['course']}")
        else:
            print("  ❌ No students found with that age.")
        
        print("\n✅ Your Turn! Add update, delete, and search by age functions.")
        print()
        print("═" * 70)

exercise5 = ExerciseStudentSystem()
exercise5.run_exercise()
```
