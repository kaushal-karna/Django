<div style="background: #f8f9fa; padding: 20px; border-radius: 10px; border-left: 5px solid #ffc107;">
    <h2 style="color: #ffc107;">🔧 Practical Demo: Creating Models</h2>
    <p>Let's build a complete Student Management System</p>
</div>

```python
import datetime

class StudentModel:
    """Simulating Django Model"""
    
    def __init__(self, name, age, course, email, is_active=True):
        self.name = name
        self.age = age
        self.course = course
        self.email = email
        self.is_active = is_active
        self.enrolled_date = datetime.datetime.now()
        self.id = None
    
    def __str__(self):
        return f"{self.name} ({self.course})"

class StudentModelManager:
    """Simulating Django's Model Manager"""
    
    def __init__(self):
        self.students = []
        self.next_id = 1
    
    def create(self, name, age, course, email):
        """Create and save a new student"""
        student = StudentModel(name, age, course, email)
        student.id = self.next_id
        self.students.append(student)
        self.next_id += 1
        return student
    
    def all(self):
        """Get all students"""
        return self.students
    
    def get(self, student_id):
        """Get a student by ID"""
        for student in self.students:
            if student.id == student_id:
                return student
        return None
    
    def filter(self, **kwargs):
        """Filter students by conditions"""
        results = self.students
        for key, value in kwargs.items():
            if key == 'course':
                results = [s for s in results if s.course == value]
            elif key == 'is_active':
                results = [s for s in results if s.is_active == value]
            elif key == 'age__gte':
                results = [s for s in results if s.age >= value]
            elif key == 'age__lte':
                results = [s for s in results if s.age <= value]
        return results
    
    def update(self, student_id, **kwargs):
        """Update a student"""
        student = self.get(student_id)
        if student:
            for key, value in kwargs.items():
                setattr(student, key, value)
            return student
        return None
    
    def delete(self, student_id):
        """Delete a student"""
        student = self.get(student_id)
        if student:
            self.students.remove(student)
            return True
        return False

class ModelDemo:
    """Demonstrate model operations"""
    
    def run_demo(self):
        print("\n" + "═" * 70)
        print(" " * 20 + "🔧 PRACTICAL MODEL DEMO" + " " * 20)
        print("═" * 70)
        print()
        
        # Initialize manager
        manager = StudentModelManager()
        
        print("📝 CREATING STUDENTS:")
        print("─" * 70)
        
        # Create students
        students_data = [
            ("Ram Sharma", 21, "CSIT", "ram@example.com"),
            ("Sita Patel", 22, "BBA", "sita@example.com"),
            ("Hari Gurung", 23, "CSIT", "hari@example.com"),
            ("Gita Adhikari", 20, "BSc IT", "gita@example.com"),
        ]
        
        for name, age, course, email in students_data:
            student = manager.create(name, age, course, email)
            print(f"  ✅ Created: {student} (ID: {student.id})")
        
        print("\n📋 ALL STUDENTS:")
        print("─" * 70)
        for student in manager.all():
            print(f"  #{student.id}: {student.name} | {student.age} | {student.course} | {student.email}")
        
        print("\n🔍 FILTERING STUDENTS:")
        print("─" * 70)
        
        # Filter by course
        csit_students = manager.filter(course="CSIT")
        print(f"  CSIT Students: {len(csit_students)}")
        for student in csit_students:
            print(f"    • {student.name}")
        
        # Filter by age
        adult_students = manager.filter(age__gte=22)
        print(f"\n  Students 22+: {len(adult_students)}")
        for student in adult_students:
            print(f"    • {student.name} ({student.age})")
        
        print("\n✏️ UPDATING STUDENT:")
        print("─" * 70)
        updated = manager.update(1, course="BSc CS")
        if updated:
            print(f"  ✅ Updated: {updated} (New Course: {updated.course})")
        
        print("\n🗑️ DELETING STUDENT:")
        print("─" * 70)
        if manager.delete(4):
            print("  ✅ Deleted student with ID 4")
        
        print("\n📋 FINAL STUDENT LIST:")
        print("─" * 70)
        for student in manager.all():
            print(f"  #{student.id}: {student.name} | {student.age} | {student.course}")
        
        print()
        print("═" * 70)
        print("🎯 CRUD Operations Complete!")
        print("═" * 70)

demo = ModelDemo()
demo.run_demo()
```

```python
class ModelRelationships:
    """Explain Django model relationships"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 20 + "🔗 MODEL RELATIONSHIPS" + " " * 20)
        print("═" * 70)
        print()
        
        print("🔹 Types of Relationships:")
        print("─" * 70)
        
        print('''
  1️⃣ One-to-Many (ForeignKey)
     ────────────────────────────────
     class Course(models.Model):
         name = models.CharField(max_length=50)
     
     class Student(models.Model):
         name = models.CharField(max_length=100)
         course = models.ForeignKey(Course, on_delete=models.CASCADE)
     
     • One course has many students
     • Each student belongs to one course
  
  2️⃣ Many-to-Many (ManyToManyField)
     ────────────────────────────────
     class Student(models.Model):
         name = models.CharField(max_length=100)
     
     class Subject(models.Model):
         name = models.CharField(max_length=50)
         students = models.ManyToManyField(Student)
     
     • Students can take many subjects
     • Subjects can have many students
  
  3️⃣ One-to-One (OneToOneField)
     ────────────────────────────────
     class User(models.Model):
         username = models.CharField(max_length=100)
     
     class Profile(models.Model):
         user = models.OneToOneField(User, on_delete=models.CASCADE)
         bio = models.TextField()
     
     • One user has one profile
     • One profile belongs to one user''')
        
        print("\n🔹 Relationship Options:")
        print("─" * 70)
        print('''
  on_delete options:
  • CASCADE        → Delete related objects
  • PROTECT        → Prevent deletion if related objects exist
  • SET_NULL       → Set foreign key to NULL
  • SET_DEFAULT    → Set to default value
  • DO_NOTHING     → Take no action

  related_name:
  • Custom reverse relationship name
  • student.course_set → student.courses''')
        
        print()
        print("═" * 70)

relationships = ModelRelationships()
relationships.display()
```

    <h2 style="text-align: center;">📝 Complete Model Example</h2>
    <p style="text-align: center;">A real-world Django model for a Blog System</p>
</div>

```python
class BlogModelDemo:
    """Complete blog system model"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 20 + "📝 COMPLETE BLOG SYSTEM MODEL" + " " * 20)
        print("═" * 70)
        print()
        
        print("🔹 models.py:")
        print("─" * 70)
        print('''
  from django.db import models
  from django.contrib.auth.models import User
  from django.utils import timezone

  class Category(models.Model):
      name = models.CharField(max_length=100, unique=True)
      slug = models.SlugField(unique=True)
      description = models.TextField(blank=True)
      created_at = models.DateTimeField(auto_now_add=True)
      
      def __str__(self):
          return self.name

      class Meta:
          verbose_name_plural = "Categories"

  class Post(models.Model):
      title = models.CharField(max_length=200)
      slug = models.SlugField(unique=True)
      content = models.TextField()
      author = models.ForeignKey(User, on_delete=models.CASCADE, related_name='posts')
      categories = models.ManyToManyField(Category, related_name='posts')
      created_at = models.DateTimeField(auto_now_add=True)
      updated_at = models.DateTimeField(auto_now=True)
      published_at = models.DateTimeField(null=True, blank=True)
      is_published = models.BooleanField(default=False)
      views = models.IntegerField(default=0)
      
      def __str__(self):
          return self.title
      
      def publish(self):
          self.is_published = True
          self.published_at = timezone.now()
          self.save()
      
      class Meta:
          ordering = ['-created_at']

  class Comment(models.Model):
      post = models.ForeignKey(Post, on_delete=models.CASCADE, related_name='comments')
      author = models.CharField(max_length=100)
      email = models.EmailField()
      content = models.TextField()
      created_at = models.DateTimeField(auto_now_add=True)
      is_approved = models.BooleanField(default=False)
      
      def __str__(self):
          return f"Comment by {self.author} on {self.post}"''')
        
        print()
        print("═" * 70)

blog = BlogModelDemo()
blog.display()
```
