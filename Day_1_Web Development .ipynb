<div style="background: #f8f9fa; padding: 20px; border-radius: 10px; border-left: 5px solid #e83e8c;">
    <h2 style="color: #e83e8c;">🔐 Backend Authentication Process</h2>
    <p>Simulating what happens when a user logs in</p>
</div>

```python
import json

class AuthenticationSystem:
    """Simulates backend authentication process"""
    
    def __init__(self):
        self.users = {
            "ram": {"password": "12345", "role": "student"},
            "sita": {"password": "abc123", "role": "teacher"},
        }
    
    def authenticate(self, username, password):
        print(f"\n   👤 Username: {username}")
        print(f"   🔑 Password: {'*' * len(password)}")
        print(f"   \n   📤 Frontend → Backend")
        
        if username in self.users:
            print(f"   ✅ User '{username}' found in database")
            if self.users[username]["password"] == password:
                print(f"   ✅ Password verified")
                print(f"   🎉 Login successful!")
                return {"success": True, "user": username, "role": self.users[username]["role"]}
            else:
                print(f"   ❌ Invalid password")
                return {"success": False, "message": "Invalid password"}
        else:
            print(f"   ❌ User '{username}' not found")
            return {"success": False, "message": "User not found"}

class AuthenticationDemo:
    """Run authentication demonstrations"""
    
    def __init__(self):
        self.auth = AuthenticationSystem()
    
    def run_demo(self):
        print("─" * 70)
        print("🔐 AUTHENTICATION PROCESS DEMONSTRATION")
        print("─" * 70)
        print("\n🔄 Testing Authentication:")
        
        test_cases = [
            ("sita", "abc123", "Valid Login"),
            ("ram", "12345", "Valid Login"),
            ("ram", "wrongpass", "Invalid Password"),
            ("unknown", "pass", "User Not Found")
        ]
        
        for username, password, test_name in test_cases:
            print(f"\n📝 {test_name}")
            print("─" * 39)
            result = self.auth.authenticate(username, password)
            print(f"   \n📥 Response: {result}")
        
        print("\n" + "═" * 70)
        print("🔐 Backend authentication is critical for web application security!")
        print("═" * 70)

demo = AuthenticationDemo()
demo.run_demo()
```

    <h2 style="color: #20c997;">🍽️ Real-World Analogy: The Restaurant</h2>
    <p>Understanding client-server through a familiar concept</p>
</div>

```python
class RestaurantAnalogy:
    """Demonstrate client-server using restaurant analogy"""
    
    def __init__(self):
        self.menu = {
            "Chicken Momo": 150,
            "Fried Rice": 200,
            "Chowmein": 180
        }
    
    def take_order(self, customer, item):
        print(f"\n📋 Customer: {customer}")
        print(f"   Orders: {item}")
        print("\n👨‍🍳 Waiter → Kitchen")
        print("   🔍 Checking availability...")
        
        if item in self.menu:
            print(f"   ✅ Item available: {item}")
            print(f"   💰 Price: Rs. {self.menu[item]}")
            print("   🔥 Cooking in progress...")
            return f"🍽️ {item} served!"
        else:
            return f"❌ {item} not available"
    
    def display_analogy(self):
        print("\n" + "═" * 70)
        print(" " * 20 + "🍽️ THE RESTAURANT ANALOGY" + " " * 20)
        print("═" * 70)
        
        result = self.take_order("Ram", "Chicken Momo")
        print(f"\n📥 Response: {result}")
        
        print("\n🔗 Analogy Mapping:")
        print("─" * 70)
        print("  Customer  =  🔵 Client (Browser)")
        print("  Waiter    =  📡 Communication Layer (HTTP)")
        print("  Kitchen   =  🟢 Server (Django Application)")
        print("  Storage   =  🗄️ Database")
        print()
        print("═" * 70)

analogy = RestaurantAnalogy()
analogy.display_analogy()
```

```python
class HTTPProtocol:
    """Explain HTTP protocol"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 15 + "📡 HTTP: HYPER TEXT TRANSFER PROTOCOL" + " " * 15)
        print("═" * 70)
        print()
        print("HTTP defines rules for how the browser and server communicate.")
        print()
        print("📤 REQUEST:")
        print("─" * 70)
        print()
        print("  🌐 Browser")
        print("     │  GET / HTTP/1.1")
        print("     │  Host: example.com")
        print("     ▼")
        print("  🖥️ Server")
        print()
        print("📥 RESPONSE:")
        print("─" * 70)
        print()
        print("  🖥️ Server")
        print("     │  HTTP/1.1 200 OK")
        print("     │  Content-Type: text/html")
        print("     ▼")
        print("  🌐 Browser")
        print()
        print("📊 Key Characteristics:")
        print("─" * 70)
        print("  🔹 Stateless protocol (each request is independent)")
        print("  🔹 Request-Response model")
        print("  🔹 Text-based protocol")
        print("  🔹 Uses TCP/IP")
        print()
        print("─" * 70)
        print("   TCP = Makes sure the data arrives correctly.")
        print("   IP = Makes sure the data knows where to go")
        print("═" * 70)

http = HTTPProtocol()
http.display()
```

```python
class HTTPvsHTTPS:
    """Compare HTTP and HTTPS"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 25 + "🔐 HTTP vs HTTPS" + " " * 25)
        print("═" * 70)
        print()
        print("🔴 HTTP (Insecure)")
        print("─" * 70)
        print("  Data is NOT encrypted in transit")
        print("  Sensitive information is visible")
        print("  Vulnerable to man-in-the-middle attacks")
        print()
        print("🟢 HTTPS (Secure)")
        print("─" * 70)
        print("  HTTP + TLS(Transport Layer Security) Encryption")
        print("  All data is encrypted")
        print("  Protected from eavesdropping")
        print("  Server identity is verified")
        print()
        print("📊 Comparison:")
        print("─" * 70)
        print('  HTTP  →  🔓 Plain text transmission (Insecure)')
        print('  HTTPS →  🔒 Encrypted transmission (Secure)')
        print()
        print("⚠️ NEVER send sensitive information (passwords, credit cards, etc.)")
        print("   over unsecured HTTP!")
        print()
        print("═" * 70)
        print("🔐 Always use HTTPS in production for security!")
        print("═" * 70)

security = HTTPvsHTTPS()
security.display()
```

<div style="background: linear-gradient(135deg, #fd7e14 0%, #dc3545 100%); padding: 20px; border-radius: 10px; color: white;">
    <h2 style="text-align: center;">🎓 Student Registration System</h2>
    <p style="text-align: center;">A complete real-world example</p>
</div>

```python
class StudentRegistrationSystem:
    """Complete student registration system"""
    
    def __init__(self):
        self.students = []
        self.next_id = 1
    
    def register_student(self, name, age, course, email):
        if not name or len(name) < 2:
            return {"success": False, "message": "Name must be at least 2 characters"}
        if age < 18 or age > 60:
            return {"success": False, "message": "Age must be between 18 and 60"}
        if not email or '@' not in email:
            return {"success": False, "message": "Invalid email address"}
        
        student = {
            "id": self.next_id,
            "name": name,
            "age": age,
            "course": course,
            "email": email,
            "status": "Active"
        }
        self.students.append(student)
        self.next_id += 1
        return {"success": True, "message": f"Student {name} registered successfully!", "student": student}

    def run_demo(self):
        print("\n" + "═" * 70)
        print(" " * 20 + "🎓 STUDENT REGISTRATION SYSTEM" + " " * 20)
        print("═" * 70)
        print()
        print("📝 Registration Attempts:")
        print("─" * 70)
        
        # Test cases
        test_cases = [
            ("Ram Sharma", 21, "CSIT", "ram@example.com"),
            ("Sita Patel", 16, "BBA", "sita@example.com"),
            ("R", 20, "CSIT", "r@example.com"),
            ("Hari Gurung", 22, "BSc IT", "harigmail")
        ]
        
        for name, age, course, email in test_cases:
            result = self.register_student(name, age, course, email)
            if result["success"]:
                print(f"\n✅ {result['message']}")
                print(f"   📊 Student ID: {result['student']['id']}")
            else:
                print(f"\n❌ {result['message']}")
        
        print("\n📋 Registered Students:")
        print("─" * 70)
        if self.students:
            for student in self.students:
                print(f"  #{student['id']}  {student['name']:15s}  {student['age']:2d}  {student['course']:12s}  {student['status']}")
        else:
            print("  No students registered yet.")
        print()
        print("═" * 70)

system = StudentRegistrationSystem()
system.run_demo()
```
