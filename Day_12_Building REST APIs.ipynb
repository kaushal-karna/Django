```python
class ModelSerializerSetup:
    """Demonstrate Model and Serializer setup"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 20 + "📋 MODEL AND SERIALIZER SETUP" + " " * 20)
        print("═" * 70)
        print()
        
        print("🔹 models.py - Complete Model:")
        print("─" * 70)
        print('''
  from django.db import models
  from django.contrib.auth.models import User

  class Student(models.Model):
      """Student model for API"""
      user = models.OneToOneField(User, on_delete=models.CASCADE, null=True, blank=True)
      name = models.CharField(max_length=100)
      age = models.IntegerField()
      email = models.EmailField(unique=True)
      course = models.CharField(max_length=50)
      phone = models.CharField(max_length=15, blank=True)
      address = models.TextField(blank=True)
      is_active = models.BooleanField(default=True)
      created_at = models.DateTimeField(auto_now_add=True)
      updated_at = models.DateTimeField(auto_now=True)
      
      def __str__(self):
          return self.name
      
      class Meta:
          ordering = ['name']
          verbose_name_plural = 'Students'

  class Course(models.Model):
      name = models.CharField(max_length=100)
      code = models.CharField(max_length=10, unique=True)
      credits = models.IntegerField()
      instructor = models.CharField(max_length=100)
      students = models.ManyToManyField(Student, related_name='courses', blank=True)
      created_at = models.DateTimeField(auto_now_add=True)
      
      def __str__(self):
          return f"{self.name} ({self.code})"''')
        
        print("\n🔹 serializers.py - Complete Serializers:")
        print("─" * 70)
        print('''
  from rest_framework import serializers
  from .models import Student, Course

  class CourseSerializer(serializers.ModelSerializer):
      class Meta:
          model = Course
          fields = ['id', 'name', 'code', 'credits', 'instructor']

  class StudentSerializer(serializers.ModelSerializer):
      courses = CourseSerializer(many=True, read_only=True)
      course_ids = serializers.ListField(
          child=serializers.IntegerField(),
          write_only=True,
          required=False
      )
      
      class Meta:
          model = Student
          fields = [
              'id', 'name', 'age', 'email', 'course', 'phone',
              'address', 'is_active', 'created_at', 'updated_at',
              'courses', 'course_ids'
          ]
          read_only_fields = ['id', 'created_at', 'updated_at', 'courses']
      
      def validate_age(self, value):
          if value < 18 or value > 60:
              raise serializers.ValidationError("Age must be between 18 and 60")
          return value
      
      def validate_email(self, value):
          # Check if email exists (excluding current instance)
          if self.instance:
              if Student.objects.exclude(pk=self.instance.pk).filter(email=value).exists():
                  raise serializers.ValidationError("Email already exists")
          else:
              if Student.objects.filter(email=value).exists():
                  raise serializers.ValidationError("Email already exists")
          return value
      
      def create(self, validated_data):
          course_ids = validated_data.pop('course_ids', [])
          student = Student.objects.create(**validated_data)
          if course_ids:
              courses = Course.objects.filter(id__in=course_ids)
              student.courses.set(courses)
          return student''')
        
        print()
        print("═" * 70)

setup = ModelSerializerSetup()
setup.display()
```

    <p><strong>Complete control over API endpoints</strong></p>

```python
class APIViewDemo:
    """Demonstrate APIView"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 27 + "🔧 APIVIEW" + " " * 27)
        print("═" * 70)
        print()
        
        print("🔹 APIView - Full Control:")
        print("─" * 70)
        print('''
  # views.py
  from rest_framework.views import APIView
  from rest_framework.response import Response
  from rest_framework import status
  from rest_framework.permissions import IsAuthenticated, IsAdminUser
  from django.shortcuts import get_object_or_404
  from .models import Student
  from .serializers import StudentSerializer

  class StudentListAPIView(APIView):
      """List all students or create a new student"""
      permission_classes = [IsAuthenticated]
      
      def get(self, request):
          students = Student.objects.all()
          serializer = StudentSerializer(students, many=True)
          return Response({
              'status': 'success',
              'count': len(serializer.data),
              'data': serializer.data
          }, status=status.HTTP_200_OK)
      
      def post(self, request):
          serializer = StudentSerializer(data=request.data)
          if serializer.is_valid():
              serializer.save()
              return Response({
                  'status': 'success',
                  'message': 'Student created successfully',
                  'data': serializer.data
              }, status=status.HTTP_201_CREATED)
          return Response({
              'status': 'error',
              'errors': serializer.errors
          }, status=status.HTTP_400_BAD_REQUEST)

  class StudentDetailAPIView(APIView):
      """Retrieve, update or delete a student"""
      permission_classes = [IsAuthenticated]
      
      def get_object(self, pk):
          return get_object_or_404(Student, pk=pk)
      
      def get(self, request, pk):
          student = self.get_object(pk)
          serializer = StudentSerializer(student)
          return Response({
              'status': 'success',
              'data': serializer.data
          }, status=status.HTTP_200_OK)
      
      def put(self, request, pk):
          student = self.get_object(pk)
          serializer = StudentSerializer(student, data=request.data)
          if serializer.is_valid():
              serializer.save()
              return Response({
                  'status': 'success',
                  'message': 'Student updated successfully',
                  'data': serializer.data
              }, status=status.HTTP_200_OK)
          return Response({
              'status': 'error',
              'errors': serializer.errors
          }, status=status.HTTP_400_BAD_REQUEST)
      
      def delete(self, request, pk):
          student = self.get_object(pk)
          student.delete()
          return Response({
              'status': 'success',
              'message': 'Student deleted successfully'
          }, status=status.HTTP_204_NO_CONTENT)''')
        
        print()
        print("═" * 70)

apiview = APIViewDemo()
apiview.display()
```

```python
class GenericViewsDemo:
    """Demonstrate Generic API Views"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 22 + "⚡ GENERIC API VIEWS" + " " * 22)
        print("═" * 70)
        print()
        
        print("🔹 Generic Views - Minimal Code:")
        print("─" * 70)
        print('''
  # views.py
  from rest_framework import generics, filters
  from rest_framework.permissions import IsAuthenticated
  from django_filters.rest_framework import DjangoFilterBackend
  from .models import Student
  from .serializers import StudentSerializer

  class StudentListCreateView(generics.ListCreateAPIView):
      """
      GET: List all students
      POST: Create a new student
      """
      queryset = Student.objects.all()
      serializer_class = StudentSerializer
      permission_classes = [IsAuthenticated]
      filter_backends = [DjangoFilterBackend, filters.SearchFilter, filters.OrderingFilter]
      filterset_fields = ['course', 'is_active']
      search_fields = ['name', 'email', 'course']
      ordering_fields = ['name', 'age', 'created_at']

  class StudentRetrieveUpdateDestroyView(generics.RetrieveUpdateDestroyAPIView):
      """
      GET: Retrieve a student
      PUT: Update a student
      PATCH: Partial update a student
      DELETE: Delete a student
      """
      queryset = Student.objects.all()
      serializer_class = StudentSerializer
      permission_classes = [IsAuthenticated]
      lookup_field = 'pk'  # Default is 'pk'
  
  # URLs - urls.py
  urlpatterns = [
      path('students/', StudentListCreateView.as_view(), name='student-list-create'),
      path('students/<int:pk>/', StudentRetrieveUpdateDestroyView.as_view(), name='student-detail'),
  ]''')
        
        print("\n🔹 Generic View Types:")
        print("─" * 70)
        print('''
  ┌─────────────────────────────────────────────────────────────────────┐
  │  Generic View                    │  Methods Provided               │
  │  ────────────────────────────────────────────────────────────────── │
  │  ListAPIView                     │  GET (List)                     │
  │  CreateAPIView                   │  POST (Create)                  │
  │  ListCreateAPIView               │  GET, POST                      │
  │  RetrieveAPIView                 │  GET (Detail)                   │
  │  UpdateAPIView                   │  PUT, PATCH                     │
  │  DestroyAPIView                  │  DELETE                         │
  │  RetrieveUpdateAPIView           │  GET, PUT, PATCH                │
  │  RetrieveDestroyAPIView          │  GET, DELETE                    │
  │  RetrieveUpdateDestroyAPIView    │  GET, PUT, PATCH, DELETE       │
  └─────────────────────────────────────────────────────────────────────┘''')
        
        print()
        print("═" * 70)

generic = GenericViewsDemo()
generic.display()
```

```python
class ViewSetRouterDemo:
    """Demonstrate ViewSets and Routers"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 20 + "🔄 VIEWSETS AND ROUTERS" + " " * 20)
        print("═" * 70)
        print()
        
        print("🔹 ModelViewSet - Full CRUD:")
        print("─" * 70)
        print('''
  # views.py
  from rest_framework import viewsets, status
  from rest_framework.decorators import action
  from rest_framework.response import Response
  from rest_framework.permissions import IsAuthenticated, IsAdminUser
  from .models import Student
  from .serializers import StudentSerializer

  class StudentViewSet(viewsets.ModelViewSet):
      """
      A ViewSet for Student CRUD operations.
      
      Provides:
      - list: GET /students/
      - create: POST /students/
      - retrieve: GET /students/{id}/
      - update: PUT /students/{id}/
      - partial_update: PATCH /students/{id}/
      - destroy: DELETE /students/{id}/
      """
      queryset = Student.objects.all()
      serializer_class = StudentSerializer
      permission_classes = [IsAuthenticated]
      
      def get_queryset(self):
          """Custom queryset with filtering"""
          queryset = super().get_queryset()
          
          # Filter by course
          course = self.request.query_params.get('course')
          if course:
              queryset = queryset.filter(course=course)
          
          # Filter by active status
          is_active = self.request.query_params.get('is_active')
          if is_active is not None:
              queryset = queryset.filter(is_active=is_active.lower() == 'true')
          
          return queryset
      
      @action(detail=False, methods=['get'])
      def stats(self, request):
          """Custom action: Get student statistics"""
          total = Student.objects.count()
          active = Student.objects.filter(is_active=True).count()
          inactive = Student.objects.filter(is_active=False).count()
          
          # Course distribution
          courses = {}
          for student in Student.objects.all():
              courses[student.course] = courses.get(student.course, 0) + 1
          
          return Response({
              'total': total,
              'active': active,
              'inactive': inactive,
              'course_distribution': courses
          })
      
      @action(detail=True, methods=['post'])
      def enroll(self, request, pk=None):
          """Custom action: Enroll student in a course"""
          student = self.get_object()
          course_id = request.data.get('course_id')
          
          if not course_id:
              return Response(
                  {'error': 'course_id is required'},
                  status=status.HTTP_400_BAD_REQUEST
              )
          
          try:
              course = Course.objects.get(id=course_id)
              student.courses.add(course)
              return Response({
                  'status': 'success',
                  'message': f'Enrolled in {course.name}'
              })
          except Course.DoesNotExist:
              return Response(
                  {'error': 'Course not found'},
                  status=status.HTTP_404_NOT_FOUND
              )

  # urls.py
  from rest_framework.routers import DefaultRouter
  from .views import StudentViewSet

  router = DefaultRouter()
  router.register(r'students', StudentViewSet, basename='student')

  urlpatterns = [
      path('api/', include(router.urls)),
  ]''')
        
        print("\n🔹 Generated Endpoints:")
        print("─" * 70)
        print('''
  ┌─────────────────────────────────────────────────────────────────────┐
  │  Method   │  URL                           │  Action              │
  │  ────────────────────────────────────────────────────────────────── │
  │  GET      │  /api/students/                │  list                │
  │  POST     │  /api/students/                │  create              │
  │  GET      │  /api/students/{id}/           │  retrieve            │
  │  PUT      │  /api/students/{id}/           │  update              │
  │  PATCH    │  /api/students/{id}/           │  partial_update      │
  │  DELETE   │  /api/students/{id}/           │  destroy             │
  │  GET      │  /api/students/stats/          │  stats (custom)      │
  │  POST     │  /api/students/{id}/enroll/    │  enroll (custom)     │
  └─────────────────────────────────────────────────────────────────────┘''')
        
        print()
        print("═" * 70)

viewset = ViewSetRouterDemo()
viewset.display()
```

```python
import json
from typing import Dict, List, Any, Optional

# Simulated CRUD API Implementation
class Response:
    def __init__(self, data, status=200):
        self.data = data
        self.status = status

class Status:
    HTTP_200_OK = 200
    HTTP_201_CREATED = 201
    HTTP_204_NO_CONTENT = 204
    HTTP_400_BAD_REQUEST = 400
    HTTP_404_NOT_FOUND = 404

class Model:
    pass

class Student(Model):
    def __init__(self, name, age, email, course, phone='', address=''):
        self.id = None
        self.name = name
        self.age = age
        self.email = email
        self.course = course
        self.phone = phone
        self.address = address
        self.is_active = True
        self.created_at = None
        self.updated_at = None

class StudentManager:
    def __init__(self):
        self.objects = []
        self.next_id = 1
    
    def all(self):
        return self.objects
    
    def get(self, pk):
        for obj in self.objects:
            if obj.id == pk:
                return obj
        return None
    
    def filter(self, **kwargs):
        results = self.objects
        for key, value in kwargs.items():
            results = [obj for obj in results if getattr(obj, key, None) == value]
        return results
    
    def create(self, **kwargs):
        student = Student(
            kwargs.get('name'),
            kwargs.get('age'),
            kwargs.get('email'),
            kwargs.get('course'),
            kwargs.get('phone', ''),
            kwargs.get('address', '')
        )
        student.id = self.next_id
        self.next_id += 1
        import datetime
        student.created_at = datetime.datetime.now()
        student.updated_at = datetime.datetime.now()
        self.objects.append(student)
        return student
    
    def update(self, pk, **kwargs):
        student = self.get(pk)
        if student:
            for key, value in kwargs.items():
                setattr(student, key, value)
            import datetime
            student.updated_at = datetime.datetime.now()
        return student
    
    def delete(self, pk):
        student = self.get(pk)
        if student:
            self.objects.remove(student)
            return True
        return False

class StudentSerializer:
    def __init__(self, instance=None, data=None, many=False):
        self.instance = instance
        self.data = data
        self.many = many
        self.errors = {}
        self.validated_data = None
    
    def is_valid(self):
        if self.data is None:
            return True
        
        self.errors = {}
        self.validated_data = {}
        
        # Validate name
        name = self.data.get('name', '').strip()
        if not name or len(name) < 2:
            self.errors['name'] = 'Name must be at least 2 characters'
        else:
            self.validated_data['name'] = name
        
        # Validate age
        age = self.data.get('age')
        if age is None:
            self.errors['age'] = 'Age is required'
        elif not isinstance(age, int) or age < 18 or age > 60:
            self.errors['age'] = 'Age must be between 18 and 60'
        else:
            self.validated_data['age'] = age
        
        # Validate email
        email = self.data.get('email', '').strip()
        if not email or '@' not in email:
            self.errors['email'] = 'Invalid email address'
        else:
            self.validated_data['email'] = email
        
        # Validate course
        course = self.data.get('course', '').strip()
        if not course:
            self.errors['course'] = 'Course is required'
        else:
            self.validated_data['course'] = course
        
        # Optional fields
        self.validated_data['phone'] = self.data.get('phone', '')
        self.validated_data['address'] = self.data.get('address', '')
        
        return len(self.errors) == 0
    
    def save(self):
        if self.instance:
            for key, value in self.validated_data.items():
                setattr(self.instance, key, value)
            return self.instance
        return Student(
            self.validated_data['name'],
            self.validated_data['age'],
            self.validated_data['email'],
            self.validated_data['course'],
            self.validated_data.get('phone', ''),
            self.validated_data.get('address', '')
        )
    
    @property
    def data(self):
        if self.many:
            return [self._serialize(obj) for obj in self.instance]
        if self.instance:
            return self._serialize(self.instance)
        return {}
    
    def _serialize(self, obj):
        return {
            'id': obj.id,
            'name': obj.name,
            'age': obj.age,
            'email': obj.email,
            'course': obj.course,
            'phone': obj.phone,
            'address': obj.address,
            'is_active': obj.is_active,
            'created_at': str(obj.created_at) if obj.created_at else None,
            'updated_at': str(obj.updated_at) if obj.updated_at else None
        }

# API Views
class APIView:
    pass

class GenericAPIView:
    def __init__(self, manager, serializer_class):
        self.manager = manager
        self.serializer_class = serializer_class

class ListCreateAPIView(GenericAPIView):
    def get(self, request):
        queryset = self.manager.all()
        serializer = self.serializer_class(instance=queryset, many=True)
        return Response({
            'status': 'success',
            'count': len(serializer.data),
            'data': serializer.data
        }, Status.HTTP_200_OK)
    
    def post(self, request):
        serializer = self.serializer_class(data=request.data)
        if serializer.is_valid():
            student = serializer.save()
            student.id = self.manager.next_id
            self.manager.next_id += 1
            import datetime
            student.created_at = datetime.datetime.now()
            student.updated_at = datetime.datetime.now()
            self.manager.objects.append(student)
            response_serializer = self.serializer_class(instance=student)
            return Response({
                'status': 'success',
                'message': 'Student created successfully',
                'data': response_serializer.data
            }, Status.HTTP_201_CREATED)
        return Response({
            'status': 'error',
            'errors': serializer.errors
        }, Status.HTTP_400_BAD_REQUEST)

class RetrieveUpdateDestroyAPIView(GenericAPIView):
    def get(self, request, pk):
        student = self.manager.get(pk)
        if not student:
            return Response({'error': 'Student not found'}, Status.HTTP_404_NOT_FOUND)
        serializer = self.serializer_class(instance=student)
        return Response({
            'status': 'success',
            'data': serializer.data
        }, Status.HTTP_200_OK)
    
    def put(self, request, pk):
        student = self.manager.get(pk)
        if not student:
            return Response({'error': 'Student not found'}, Status.HTTP_404_NOT_FOUND)
        
        serializer = self.serializer_class(instance=student, data=request.data)
        if serializer.is_valid():
            updated = serializer.save()
            import datetime
            updated.updated_at = datetime.datetime.now()
            # Update in manager
            for key, value in serializer.validated_data.items():
                setattr(student, key, value)
            response_serializer = self.serializer_class(instance=student)
            return Response({
                'status': 'success',
                'message': 'Student updated successfully',
                'data': response_serializer.data
            }, Status.HTTP_200_OK)
        return Response({
            'status': 'error',
            'errors': serializer.errors
        }, Status.HTTP_400_BAD_REQUEST)
    
    def delete(self, request, pk):
        if self.manager.delete(pk):
            return Response({
                'status': 'success',
                'message': 'Student deleted successfully'
            }, Status.HTTP_204_NO_CONTENT)
        return Response({'error': 'Student not found'}, Status.HTTP_404_NOT_FOUND)

class Request:
    def __init__(self, data=None, method='GET', user=None):
        self.data = data
        self.method = method
        self.user = user

class CRUDAPIDemo:
    """Complete CRUD API demonstration"""
    
    def run_demo(self):
        print("\n" + "═" * 70)
        print(" " * 18 + "🔧 COMPLETE CRUD API DEMONSTRATION" + " " * 18)
        print("═" * 70)
        print()
        
        manager = StudentManager()
        
        # Create API endpoints
        student_list = ListCreateAPIView(manager, StudentSerializer)
        student_detail = RetrieveUpdateDestroyAPIView(manager, StudentSerializer)
        
        print("📝 API Endpoints:")
        print("─" * 70)
        print("  🔹 GET    /api/students/     → List all students")
        print("  🔹 POST   /api/students/     → Create a new student")
        print("  🔹 GET    /api/students/1/   → Get student details")
        print("  🔹 PUT    /api/students/1/   → Update student")
        print("  🔹 DELETE /api/students/1/   → Delete student")
        print()
        
        # Test Data
        test_students = [
            {'name': 'Ram Sharma', 'age': 21, 'email': 'ram@example.com', 
             'course': 'CSIT', 'phone': '9841234567'},
            {'name': 'Sita Patel', 'age': 22, 'email': 'sita@example.com',
             'course': 'BBA', 'phone': '9812345678'},
            {'name': 'Hari Gurung', 'age': 23, 'email': 'hari@example.com',
             'course': 'CSIT', 'phone': '9854321098'},
            {'name': 'Gita Adhikari', 'age': 20, 'email': 'gita@example.com',
             'course': 'BSc IT', 'phone': '9801234567'},
        ]
        
        # 1. CREATE - POST
        print("🔹 1. CREATE OPERATIONS (POST):")
        print("─" * 70)
        created_ids = []
        
        for data in test_students:
            request = Request(data=data, method='POST')
            response = student_list.post(request)
            if response.status == Status.HTTP_201_CREATED:
                created_ids.append(response.data['data']['id'])
                print(f"  ✅ Created: {response.data['data']['name']} (ID: {response.data['data']['id']})")
            else:
                print(f"  ❌ Error: {response.data}")
        
        # 2. LIST - GET
        print("\n🔹 2. LIST OPERATIONS (GET):")
        print("─" * 70)
        request = Request(method='GET')
        response = student_list.get(request)
        print(f"  📊 Total Students: {response.data['count']}")
        for student in response.data['data']:
            print(f"    #{student['id']}: {student['name']} | {student['age']} | {student['course']}")
        
        # 3. RETRIEVE - GET /1/
        print("\n🔹 3. RETRIEVE OPERATIONS (GET /1/):")
        print("─" * 70)
        if created_ids:
            request = Request(method='GET')
            response = student_detail.get(request, pk=created_ids[0])
            if response.status == Status.HTTP_200_OK:
                student = response.data['data']
                print(f"  📄 Student Details:")
                print(f"    Name: {student['name']}")
                print(f"    Age: {student['age']}")
                print(f"    Email: {student['email']}")
                print(f"    Course: {student['course']}")
                print(f"    Phone: {student['phone']}")
        
        # 4. UPDATE - PUT /2/
        print("\n🔹 4. UPDATE OPERATIONS (PUT /2/):")
        print("─" * 70)
        if len(created_ids) >= 2:
            update_data = {
                'name': 'Sita Patel',
                'age': 23,
                'email': 'sita@example.com',
                'course': 'MBA',
                'phone': '9812345678'
            }
            request = Request(data=update_data, method='PUT')
            response = student_detail.put(request, pk=created_ids[1])
            if response.status == Status.HTTP_200_OK:
                print(f"  ✅ Updated: {response.data['data']['name']}")
                print(f"    New Course: {response.data['data']['course']}")
                print(f"    New Age: {response.data['data']['age']}")
        
        # 5. DELETE - DELETE /3/
        print("\n🔹 5. DELETE OPERATIONS (DELETE /3/):")
        print("─" * 70)
        if len(created_ids) >= 3:
            request = Request(method='DELETE')
            response = student_detail.delete(request, pk=created_ids[2])
            if response.status == Status.HTTP_204_NO_CONTENT:
                print(f"  ✅ Deleted student with ID: {created_ids[2]}")
        
        # 6. Final List
        print("\n🔹 6. FINAL STUDENT LIST:")
        print("─" * 70)
        request = Request(method='GET')
        response = student_list.get(request)
        print(f"  📊 Remaining Students: {response.data['count']}")
        for student in response.data['data']:
            print(f"    #{student['id']}: {student['name']} | {student['course']}")
        
        print("\n" + "═" * 70)
        print("🎯 CRUD API Demo Complete!")
        print("═" * 70)

demo = CRUDAPIDemo()
demo.run_demo()
```
