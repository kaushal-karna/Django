```python
class HTTPMethodsAPI:
    """Demonstrate HTTP methods in REST API"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 20 + "📤 HTTP METHODS IN REST API" + " " * 20)
        print("═" * 70)
        print()
        
        print("🔹 RESTful URL Patterns:")
        print("─" * 70)
        print("""
  ┌─────────────────────────────────────────────────────────────────────┐
  │  Method  │  URL                │  Action     │  Equivalent CRUD   │
  │  ────────────────────────────────────────────────────────────────── │
  │  GET     │  /api/students/     │  List       │  READ (All)       │
  │  GET     │  /api/students/1/   │  Retrieve   │  READ (One)       │
  │  POST    │  /api/students/     │  Create     │  CREATE            │
  │  PUT     │  /api/students/1/   │  Update     │  UPDATE (Full)    │
  │  PATCH   │  /api/students/1/   │  Update     │  UPDATE (Partial) │
  │  DELETE  │  /api/students/1/   │  Delete     │  DELETE            │
  └─────────────────────────────────────────────────────────────────────┘""")
        
        print("\n🔹 HTTP Methods Explained:")
        print("─" * 70)
        print('''
  📖 GET - Read data from the server
  • Safe operation (doesn't modify data)
  • Can be cached
  • Example: GET /api/students/
  
  📤 POST - Create new data on the server
  • Not safe (modifies data)
  • Not idempotent (multiple requests create multiple resources)
  • Example: POST /api/students/ with student data
  
  🔄 PUT - Update entire resource
  • Not safe (modifies data)
  • Idempotent (same request gives same result)
  • Requires complete resource data
  • Example: PUT /api/students/1/ with complete student data
  
  ✏️ PATCH - Update part of a resource
  • Not safe (modifies data)
  • Not necessarily idempotent
  • Only sends fields to update
  • Example: PATCH /api/students/1/ with {age: 22}
  
  🗑️ DELETE - Remove data from the server
  • Not safe (modifies data)
  • Idempotent (same request gives same result)
  • Example: DELETE /api/students/1/''')
        
        print()
        print("═" * 70)

http_methods = HTTPMethodsAPI()
http_methods.display()
```

```python
class SerializersDemo:
    """Demonstrate Django REST Framework Serializers"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 25 + "📋 SERIALIZERS IN DRF" + " " * 25)
        print("═" * 70)
        print()
        
        print("🔹 What is a Serializer?")
        print("─" * 70)
        print("""
  A Serializer converts:
  • Python objects → JSON (Serialization)
  • JSON → Python objects (Deserialization)
  
  📌 Uses:
  • Validate incoming data
  • Transform complex data types
  • Handle nested relationships""")
        
        print("\n🔹 Basic Serializer Example:")
        print("─" * 70)
        print('''
  # serializers.py
  from rest_framework import serializers
  from .models import Student

  class StudentSerializer(serializers.ModelSerializer):
      """Serializer for Student model"""
      
      class Meta:
          model = Student
          fields = ['id', 'name', 'age', 'email', 'course', 'enrolled_date']
          read_only_fields = ['id', 'enrolled_date']
      
      # Custom validation
      def validate_age(self, value):
          if value < 18 or value > 60:
              raise serializers.ValidationError("Age must be between 18 and 60")
          return value
      
      def validate_email(self, value):
          if Student.objects.filter(email=value).exists():
              raise serializers.ValidationError("Email already exists")
          return value''')
        
        print("\n🔹 ModelSerializer vs Serializer:")
        print("─" * 70)
        print('''
  # Regular Serializer (Manual)
  class StudentSerializer(serializers.Serializer):
      id = serializers.IntegerField(read_only=True)
      name = serializers.CharField(max_length=100)
      age = serializers.IntegerField()
      email = serializers.EmailField()
      
      def create(self, validated_data):
          return Student.objects.create(**validated_data)
      
      def update(self, instance, validated_data):
          instance.name = validated_data.get('name', instance.name)
          instance.age = validated_data.get('age', instance.age)
          instance.email = validated_data.get('email', instance.email)
          instance.save()
          return instance
  
  # ModelSerializer (Automatic)
  class StudentSerializer(serializers.ModelSerializer):
      class Meta:
          model = Student
          fields = '__all__'
  
  ✅ ModelSerializer is recommended for model-based APIs!''')
        
        print("\n🔹 Nested Serializers:")
        print("─" * 70)
        print('''
  # For related models
  class CourseSerializer(serializers.ModelSerializer):
      class Meta:
          model = Course
          fields = ['id', 'name', 'code', 'credits']
  
  class StudentSerializer(serializers.ModelSerializer):
      courses = CourseSerializer(many=True, read_only=True)
      
      class Meta:
          model = Student
          fields = ['id', 'name', 'age', 'email', 'courses']
  
  # Nested output:
  # {
  #     "id": 1,
  #     "name": "Ram Sharma",
  #     "age": 21,
  #     "email": "ram@example.com",
  #     "courses": [
  #         {"id": 1, "name": "Python", "code": "PY101", "credits": 3},
  #         {"id": 2, "name": "Django", "code": "DJ101", "credits": 4}
  #     ]
  # }''')
        
        print()
        print("═" * 70)

serializers = SerializersDemo()
serializers.display()
```

    <h2 style="color: #fd7e14;">👁️ API Views in DRF</h2>
    <p><strong>Building views for REST APIs</strong></p>

```python
class APIViewsDemo:
    """Demonstrate API Views"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 27 + "👁️ API VIEWS" + " " * 27)
        print("═" * 70)
        print()
        
        print("🔹 Method 1: Function-Based Views (FBV):")
        print("─" * 70)
        print('''
  # views.py
  from rest_framework.decorators import api_view
  from rest_framework.response import Response
  from rest_framework import status
  from .models import Student
  from .serializers import StudentSerializer

  @api_view(['GET', 'POST'])
  def student_list(request):
      if request.method == 'GET':
          students = Student.objects.all()
          serializer = StudentSerializer(students, many=True)
          return Response(serializer.data)
      
      elif request.method == 'POST':
          serializer = StudentSerializer(data=request.data)
          if serializer.is_valid():
              serializer.save()
              return Response(serializer.data, status=status.HTTP_201_CREATED)
          return Response(serializer.errors, status=status.HTTP_400_BAD_REQUEST)

  @api_view(['GET', 'PUT', 'DELETE'])
  def student_detail(request, pk):
      try:
          student = Student.objects.get(pk=pk)
      except Student.DoesNotExist:
          return Response(status=status.HTTP_404_NOT_FOUND)
      
      if request.method == 'GET':
          serializer = StudentSerializer(student)
          return Response(serializer.data)
      
      elif request.method == 'PUT':
          serializer = StudentSerializer(student, data=request.data)
          if serializer.is_valid():
              serializer.save()
              return Response(serializer.data)
          return Response(serializer.errors, status=status.HTTP_400_BAD_REQUEST)
      
      elif request.method == 'DELETE':
          student.delete()
          return Response(status=status.HTTP_204_NO_CONTENT)''')
        
        print("\n🔹 Method 2: Class-Based Views (CBV):")
        print("─" * 70)
        print('''
  # views.py
  from rest_framework.views import APIView
  from rest_framework.response import Response
  from rest_framework import status
  from django.http import Http404
  from .models import Student
  from .serializers import StudentSerializer

  class StudentList(APIView):
      def get(self, request):
          students = Student.objects.all()
          serializer = StudentSerializer(students, many=True)
          return Response(serializer.data)
      
      def post(self, request):
          serializer = StudentSerializer(data=request.data)
          if serializer.is_valid():
              serializer.save()
              return Response(serializer.data, status=status.HTTP_201_CREATED)
          return Response(serializer.errors, status=status.HTTP_400_BAD_REQUEST)

  class StudentDetail(APIView):
      def get_object(self, pk):
          try:
              return Student.objects.get(pk=pk)
          except Student.DoesNotExist:
              raise Http404
      
      def get(self, request, pk):
          student = self.get_object(pk)
          serializer = StudentSerializer(student)
          return Response(serializer.data)
      
      def put(self, request, pk):
          student = self.get_object(pk)
          serializer = StudentSerializer(student, data=request.data)
          if serializer.is_valid():
              serializer.save()
              return Response(serializer.data)
          return Response(serializer.errors, status=status.HTTP_400_BAD_REQUEST)
      
      def delete(self, request, pk):
          student = self.get_object(pk)
          student.delete()
          return Response(status=status.HTTP_204_NO_CONTENT)''')
        
        print("\n🔹 Method 3: Generic Views (Recommended):")
        print("─" * 70)
        print('''
  # views.py
  from rest_framework import generics
  from .models import Student
  from .serializers import StudentSerializer

  class StudentListCreate(generics.ListCreateAPIView):
      queryset = Student.objects.all()
      serializer_class = StudentSerializer

  class StudentRetrieveUpdateDestroy(generics.RetrieveUpdateDestroyAPIView):
      queryset = Student.objects.all()
      serializer_class = StudentSerializer
      lookup_field = 'pk'
  
  # This gives us full CRUD with minimal code!''')
        
        print()
        print("═" * 70)

api_views = APIViewsDemo()
api_views.display()
```

    <h2 style="color: #20c997;">🔧 Practical: Building a Complete REST API</h2>
    <p><strong>Step-by-step API implementation</strong></p>
</div>

```python
import json
from typing import Dict, List, Any

# Simulated Django REST Framework
class Response:
    """Simulated DRF Response"""
    def __init__(self, data, status=200):
        self.data = data
        self.status = status

class Status:
    """Simulated HTTP Status Codes"""
    HTTP_200_OK = 200
    HTTP_201_CREATED = 201
    HTTP_204_NO_CONTENT = 204
    HTTP_400_BAD_REQUEST = 400
    HTTP_404_NOT_FOUND = 404

# Simulated Models
class Student:
    def __init__(self, name, age, email, course):
        self.id = None
        self.name = name
        self.age = age
        self.email = email
        self.course = course

class StudentManager:
    def __init__(self):
        self.objects = []
        self.next_id = 1
    
    def all(self):
        return self.objects
    
    def get(self, pk):
        for obj in self.objects:
            if obj.id == pk:
                return obj
        return None
    
    def create(self, **kwargs):
        student = Student(
            kwargs.get('name'),
            kwargs.get('age'),
            kwargs.get('email'),
            kwargs.get('course')
        )
        student.id = self.next_id
        self.next_id += 1
        self.objects.append(student)
        return student
    
    def update(self, pk, **kwargs):
        student = self.get(pk)
        if student:
            for key, value in kwargs.items():
                setattr(student, key, value)
        return student
    
    def delete(self, pk):
        student = self.get(pk)
        if student:
            self.objects.remove(student)
            return True
        return False

# Simulated Serializer
class StudentSerializer:
    def __init__(self, instance=None, data=None, many=False):
        self.instance = instance
        self.data = data
        self.many = many
        self.errors = {}
        self.validated_data = None
    
    def is_valid(self):
        if self.data is None:
            return True
        
        self.errors = {}
        self.validated_data = {}
        
        # Validate name
        name = self.data.get('name', '').strip()
        if not name or len(name) < 2:
            self.errors['name'] = 'Name must be at least 2 characters'
        else:
            self.validated_data['name'] = name
        
        # Validate age
        age = self.data.get('age')
        if age is None:
            self.errors['age'] = 'Age is required'
        elif age < 18 or age > 60:
            self.errors['age'] = 'Age must be between 18 and 60'
        else:
            self.validated_data['age'] = age
        
        # Validate email
        email = self.data.get('email', '').strip()
        if not email or '@' not in email:
            self.errors['email'] = 'Invalid email address'
        else:
            self.validated_data['email'] = email
        
        # Validate course
        course = self.data.get('course', '').strip()
        if not course:
            self.errors['course'] = 'Course is required'
        else:
            self.validated_data['course'] = course
        
        return len(self.errors) == 0
    
    def save(self):
        if self.instance:
            # Update
            for key, value in self.validated_data.items():
                setattr(self.instance, key, value)
            return self.instance
        else:
            # Create
            return Student(
                self.validated_data['name'],
                self.validated_data['age'],
                self.validated_data['email'],
                self.validated_data['course']
            )
    
    def to_json(self):
        if self.many:
            return [self._serialize(obj) for obj in self.instance]
        if self.instance:
            return self._serialize(self.instance)
        return {}
    
    def _serialize(self, obj):
        return {
            'id': obj.id,
            'name': obj.name,
            'age': obj.age,
            'email': obj.email,
            'course': obj.course
        }

# Simulated API Views
class APIView:
    pass

class ListCreateAPIView:
    def __init__(self, manager, serializer_class):
        self.manager = manager
        self.serializer_class = serializer_class
    
    def get(self, request):
        queryset = self.manager.all()
        serializer = self.serializer_class(instance=queryset, many=True)
        return Response(serializer.to_json(), Status.HTTP_200_OK)
    
    def post(self, request):
        serializer = self.serializer_class(data=request.data)
        if serializer.is_valid():
            student = serializer.save()
            student.id = self.manager.next_id
            self.manager.next_id += 1
            self.manager.objects.append(student)
            response_serializer = self.serializer_class(instance=student)
            return Response(response_serializer.to_json(), Status.HTTP_201_CREATED)
        return Response(serializer.errors, Status.HTTP_400_BAD_REQUEST)

class RetrieveUpdateDestroyAPIView:
    def __init__(self, manager, serializer_class):
        self.manager = manager
        self.serializer_class = serializer_class
    
    def get(self, request, pk):
        student = self.manager.get(pk)
        if not student:
            return Response({'error': 'Not found'}, Status.HTTP_404_NOT_FOUND)
        serializer = self.serializer_class(instance=student)
        return Response(serializer.to_json(), Status.HTTP_200_OK)
    
    def put(self, request, pk):
        student = self.manager.get(pk)
        if not student:
            return Response({'error': 'Not found'}, Status.HTTP_404_NOT_FOUND)
        
        serializer = self.serializer_class(instance=student, data=request.data)
        if serializer.is_valid():
            updated = serializer.save()
            # Update in manager
            for key, value in serializer.validated_data.items():
                setattr(student, key, value)
            response_serializer = self.serializer_class(instance=student)
            return Response(response_serializer.to_json(), Status.HTTP_200_OK)
        return Response(serializer.errors, Status.HTTP_400_BAD_REQUEST)
    
    def delete(self, request, pk):
        if self.manager.delete(pk):
            return Response({}, Status.HTTP_204_NO_CONTENT)
        return Response({'error': 'Not found'}, Status.HTTP_404_NOT_FOUND)

class Status:
    HTTP_200_OK = 200
    HTTP_201_CREATED = 201
    HTTP_204_NO_CONTENT = 204
    HTTP_400_BAD_REQUEST = 400
    HTTP_404_NOT_FOUND = 404

class Request:
    def __init__(self, data=None, method='GET'):
        self.data = data
        self.method = method

class RESTAPIDemo:
    """Complete REST API demonstration"""
    
    def run_demo(self):
        print("\n" + "═" * 70)
        print(" " * 18 + "🔧 COMPLETE REST API DEMONSTRATION" + " " * 18)
        print("═" * 70)
        print()
        
        manager = StudentManager()
        
        # Create API endpoints
        student_list = ListCreateAPIView(manager, StudentSerializer)
        student_detail = RetrieveUpdateDestroyAPIView(manager, StudentSerializer)
        
        print("📝 API Endpoints:")
        print("─" * 70)
        print("  GET    /api/students/     → List all students")
        print("  POST   /api/students/     → Create a new student")
        print("  GET    /api/students/1/   → Get student details")
        print("  PUT    /api/students/1/   → Update student")
        print("  DELETE /api/students/1/   → Delete student")
        print()
        
        # 1. CREATE - POST /api/students/
        print("🔹 1. CREATE (POST /api/students/):")
        print("─" * 70)
        
        students_data = [
            {'name': 'Ram Sharma', 'age': 21, 'email': 'ram@example.com', 'course': 'CSIT'},
            {'name': 'Sita Patel', 'age': 22, 'email': 'sita@example.com', 'course': 'BBA'},
            {'name': 'Hari Gurung', 'age': 23, 'email': 'hari@example.com', 'course': 'CSIT'},
        ]
        
        created_students = []
        for data in students_data:
            request = Request(data=data, method='POST')
            response = student_list.post(request)
            if response.status == Status.HTTP_201_CREATED:
                created_students.append(response.data)
                print(f"  ✅ Created: {response.data['name']} (ID: {response.data['id']})")
            else:
                print(f"  ❌ Error: {response.data}")
        
        # 2. LIST - GET /api/students/
        print("\n🔹 2. LIST (GET /api/students/):")
        print("─" * 70)
        request = Request(method='GET')
        response = student_list.get(request)
        print(f"  📊 Response (JSON):")
        print(f"  {json.dumps(response.data, indent=2)}")
        
        # 3. RETRIEVE - GET /api/students/1/
        print("\n🔹 3. RETRIEVE (GET /api/students/1/):")
        print("─" * 70)
        request = Request(method='GET')
        response = student_detail.get(request, pk=1)
        if response.status == Status.HTTP_200_OK:
            print(f"  📊 Student Details:")
            print(f"  {json.dumps(response.data, indent=2)}")
        
        # 4. UPDATE - PUT /api/students/2/
        print("\n🔹 4. UPDATE (PUT /api/students/2/):")
        print("─" * 70)
        update_data = {'name': 'Sita Patel', 'age': 23, 'email': 'sita@example.com', 'course': 'MBA'}
        request = Request(data=update_data, method='PUT')
        response = student_detail.put(request, pk=2)
        if response.status == Status.HTTP_200_OK:
            print(f"  ✅ Updated Student:")
            print(f"  {json.dumps(response.data, indent=2)}")
        
        # 5. DELETE - DELETE /api/students/3/
        print("\n🔹 5. DELETE (DELETE /api/students/3/):")
        print("─" * 70)
        request = Request(method='DELETE')
        response = student_detail.delete(request, pk=3)
        if response.status == Status.HTTP_204_NO_CONTENT:
            print("  ✅ Student deleted successfully")
        
        # 6. FINAL LIST
        print("\n🔹 6. FINAL STUDENT LIST:")
        print("─" * 70)
        request = Request(method='GET')
        response = student_list.get(request)
        print(f"  📊 Final List:")
        print(f"  {json.dumps(response.data, indent=2)}")
        
        print("\n" + "═" * 70)
        print("🎯 REST API Demo Complete!")
        print("═" * 70)

demo = RESTAPIDemo()
demo.run_demo()
```
