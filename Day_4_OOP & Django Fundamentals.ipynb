<div style="background: #f8f9fa; padding: 25px; border-radius: 15px; border-left: 6px solid #28a745; margin: 20px 0;">
    <h2 style="color: #28a745; margin-top: 0;">📦 1. Classes and Objects</h2>
    <p>A class is a blueprint for creating objects. Objects are instances of classes.</p>
</div>

```python
class OOPClassesDemo:
    """Demonstrate classes and objects"""
    
    def display(self):
        print("\n" + "╔" + "═" * 70 + "╗")
        print("║" + " " * 22 + "📦 CLASSES AND OBJECTS" + " " * 23 + "║")
        print("╚" + "═" * 70 + "╝")
        
        # Define the Student class
        class Student:
            # Class attribute (shared by all instances)
            school = "Kathmandu University"
            
            # Constructor (initializes instance attributes)
            def __init__(self, name, age, course):
                self.name = name
                self.age = age
                self.course = course
            
            # Method
            def display_info(self):
                return f"{self.name} - {self.age} - {self.course}"
            
            def get_school(self):
                return self.school
        
        print("\n🔹 Creating a Class:")
        print("─" * 72)
        print('''  class Student:
      # Class attribute (shared by all instances)
      school = "Kathmandu University"
      
      # Constructor (initializes instance attributes)
      def __init__(self, name, age, course):
          self.name = name      # Instance attribute
          self.age = age        # Instance attribute
          self.course = course  # Instance attribute
      
      # Method
      def display_info(self):
          return f"{self.name} - {self.age} - {self.course}"''')
        
        # Create objects
        print("\n🔹 Creating Objects (Instances):")
        print("─" * 72)
        student1 = Student("Ram", 21, "CSIT")
        student2 = Student("Sita", 22, "BBA")
        student3 = Student("Hari", 20, "BSc IT")
        
        print(f"  📚 Student 1: {student1.display_info()}")
        print(f"  📚 Student 2: {student2.display_info()}")
        print(f"  📚 Student 3: {student3.display_info()}")
        
        print("\n🔹 Accessing Attributes:")
        print("─" * 72)
        print(f"  Ram's name: {student1.name}")
        print(f"  Ram's age: {student1.age}")
        print(f"  Ram's course: {student1.course}")
        print(f"  School: {student1.school}")
        
        print("\n🔹 Instance vs Class Attributes:")
        print("─" * 72)
        print(f"  Name: {student1.name} (Instance specific)")
        print(f"  Age: {student1.age} (Instance specific)")
        print(f"  School: {student1.school} (Shared by all instances)")
        
        print("\n💡 Classes in Django:")
        print("─" * 72)
        print("  • Models are classes (inherits from models.Model)")
        print("  • Forms are classes (inherits from forms.Form)")
        print("  • Views can be classes (inherits from View)")
        
        print()
        print("═" * 70)

classes_demo = OOPClassesDemo()
classes_demo.display()
```

    <h2 style="color: #fd7e14; margin-top: 0;">🧬 2. Inheritance</h2>
    <p>Inheritance allows a class to inherit attributes and methods from another class.</p>
</div>

```python
class InheritanceDemo:
    """Demonstrate inheritance"""
    
    def display(self):
        print("\n" + "╔" + "═" * 70 + "╗")
        print("║" + " " * 28 + "🧬 INHERITANCE" + " " * 29 + "║")
        print("╚" + "═" * 70 + "╝")
        
        # Base class
        class Person:
            def __init__(self, name, age):
                self.name = name
                self.age = age
            
            def display_info(self):
                return f"Name: {self.name}, Age: {self.age}"
        
        # Derived class
        class Student(Person):
            def __init__(self, name, age, course):
                super().__init__(name, age)
                self.course = course
            
            def display_info(self):
                return f"{super().display_info()}, Course: {self.course}"
            
            def study(self):
                return f"{self.name} is studying..."
        
        print("\n🔹 Base Class (Parent):")
        print("─" * 72)
        print('''  class Person:
      def __init__(self, name, age):
          self.name = name
          self.age = age
      
      def display_info(self):
          return f"Name: {self.name}, Age: {self.age}"''')
        
        print("\n🔹 Derived Class (Child) - Single Inheritance:")
        print("─" * 72)
        print('''  class Student(Person):
      def __init__(self, name, age, course):
          super().__init__(name, age)
          self.course = course
      
      def display_info(self):
          return f"{super().display_info()}, Course: {self.course}"''')
        
        # Using inheritance
        print("\n🔹 Using Inheritance:")
        print("─" * 72)
        person = Person("Ram", 25)
        student = Student("Sita", 22, "MBA")
        print(f"  📚 Person: {person.display_info()}")
        print(f"  📚 Student: {student.display_info()}")
        
        print("\n🔹 Method Overriding:")
        print("─" * 72)
        print(f"  📚 Person display: {person.display_info()}")
        print(f"  📚 Student display: {student.display_info()}")
        
        # Multiple inheritance
        print("\n🔹 Multiple Inheritance:")
        print("─" * 72)
        class Teacher:
            def teach(self):
                return "Teaching..."
        
        class Researcher:
            def research(self):
                return "Researching..."
        
        class Professor(Teacher, Researcher):
            def work(self):
                return f"{self.teach()} {self.research()}"
        
        print('''  class Teacher:
      def teach(self):
          return "Teaching..."
  
  class Researcher:
      def research(self):
          return "Researching..."
  
  class Professor(Teacher, Researcher):
      def work(self):
          return f"{self.teach()} {self.research()}"''')
        
        professor = Professor()
        print(f"\n  📚 Professor: {professor.work()}")
        
        print("\n💡 Inheritance in Django:")
        print("─" * 72)
        print("  • All Django models inherit from models.Model")
        print("  • All Django forms inherit from forms.Form or forms.ModelForm")
        print("  • Class-based views inherit from View")
        
        print()
        print("═" * 70)

inheritance_demo = InheritanceDemo()
inheritance_demo.display()
```

<div style="background: #f8f9fa; padding: 25px; border-radius: 15px; border-left: 6px solid #17a2b8; margin: 20px 0;">
    <h2 style="color: #17a2b8; margin-top: 0;">🔒 3. Encapsulation</h2>
    <p>Encapsulation bundles data and methods within a class, hiding internal details.</p>
</div>

```python
class EncapsulationDemo:
    """Demonstrate encapsulation"""
    
    def display(self):
        print("\n" + "╔" + "═" * 70 + "╗")
        print("║" + " " * 26 + "🔒 ENCAPSULATION" + " " * 27 + "║")
        print("╚" + "═" * 70 + "╝")
        
        # Encapsulation example
        print("\n🔹 Public vs Private Attributes:")
        print("─" * 72)
        print('''  class BankAccount:
      def __init__(self, owner, balance):
          self.owner = owner              # Public
          self.__balance = balance        # Private (name mangling)
      
      def get_balance(self):              # Getter
          return self.__balance
      
      def deposit(self, amount):          # Setter with validation
          if amount > 0:
              self.__balance += amount
              return f"Deposited {amount}. New balance: {self.__balance}"
          return "Invalid deposit amount"''')
        
        # Define the class
        class BankAccount:
            def __init__(self, owner, balance):
                self.owner = owner
                self.__balance = balance
            
            def get_balance(self):
                return self.__balance
            
            def deposit(self, amount):
                if amount > 0:
                    self.__balance += amount
                    return f"Deposited {amount}. New balance: {self.__balance}"
                return "Invalid deposit amount"
        
        print("\n🔹 Using Encapsulation:")
        print("─" * 72)
        account = BankAccount("Ram", 1000)
        print(f"  📊 Account created for {account.owner}")
        print(f"  💰 Initial balance: {account.get_balance()}")
        print(f"  ✅ {account.deposit(500)}")
        
        try:
            print(f"  ❌ Cannot access private attribute: {account.__balance}")
        except AttributeError as e:
            print(f"  ❌ Cannot access private attribute: {e}")
        
        print(f"  💰 Balance via getter: {account.get_balance()}")
        
        # Property decorator
        print("\n🔹 Property Decorator (Pythonic Encapsulation):")
        print("─" * 72)
        print('''  class Person:
      def __init__(self, name, age):
          self._name = name
          self._age = age
      
      @property
      def age(self):
          return self._age
      
      @age.setter
      def age(self, value):
          if value > 0:
              self._age = value
          else:
              raise ValueError("Age must be positive")''')
        
        class Person:
            def __init__(self, name, age):
                self._name = name
                self._age = age
            
            @property
            def age(self):
                return self._age
            
            @age.setter
            def age(self, value):
                if value > 0:
                    self._age = value
                else:
                    raise ValueError("Age must be positive")
            
            @property
            def name(self):
                return self._name
        
        print("\n📊 Using Property:")
        print("─" * 72)
        person = Person("Ram", 25)
        print(f"  📚 Person: {person.name}, Age: {person.age}")
        print(f"  ✅ Age: {person.age}")
        person.age = 30
        print(f"  🔄 Setting age to 30...")
        print(f"  ✅ Age updated to: {person.age}")
        
        print("\n💡 Encapsulation in Django:")
        print("─" * 72)
        print("  • Model fields are encapsulated")
        print("  • Form validation encapsulates data checking")
        print("  • Views encapsulate request handling logic")
        
        print()
        print("═" * 70)

encapsulation_demo = EncapsulationDemo()
encapsulation_demo.display()
```

    <h2 style="color: #6f42c1; margin-top: 0;">🔄 4. Polymorphism</h2>
    <p>Polymorphism allows objects of different classes to respond to the same method.</p>
</div>

```python
class PolymorphismDemo:
    """Demonstrate polymorphism"""
    
    def display(self):
        print("\n" + "╔" + "═" * 70 + "╗")
        print("║" + " " * 26 + "🔄 POLYMORPHISM" + " " * 27 + "║")
        print("╚" + "═" * 70 + "╝")
        
        print("\n🔹 Polymorphism with Inheritance:")
        print("─" * 72)
        print('''  class Animal:
      def make_sound(self):
          return "Some sound"
  
  class Dog(Animal):
      def make_sound(self):
          return "Woof!"
  
  class Cat(Animal):
      def make_sound(self):
          return "Meow!"''')
        
        class Animal:
            def make_sound(self):
                return "Some sound"
        
        class Dog(Animal):
            def make_sound(self):
                return "Woof!"
        
        class Cat(Animal):
            def make_sound(self):
                return "Meow!"
        
        class Bird(Animal):
            def make_sound(self):
                return "Chirp!"
        
        print("\n🔹 Using Polymorphism:")
        print("─" * 72)
        dog = Dog()
        cat = Cat()
        bird = Bird()
        
        print(f"  🐕 Dog says: {dog.make_sound()}")
        print(f"  🐈 Cat says: {cat.make_sound()}")
        print(f"  🐦 Bird says: {bird.make_sound()}")
        
        print("\n🔹 Polymorphism with Functions:")
        print("─" * 72)
        def animal_sound(animal):
            return f"{animal.make_sound()}"
        
        print(f"  🐕 Dog says: {animal_sound(dog)}")
        print(f"  🐈 Cat says: {animal_sound(cat)}")
        print(f"  🐦 Bird says: {animal_sound(bird)}")
        
        print("\n🔹 Polymorphism with Different Classes:")
        print("─" * 72)
        class Student:
            def __init__(self, name, course):
                self.name = name
                self.course = course
            
            def display(self):
                return f"Student: {self.name} - {self.course}"
        
        class Teacher:
            def __init__(self, name, subject):
                self.name = name
                self.subject = subject
            
            def display(self):
                return f"Teacher: {self.name} - {self.subject}"
        
        class Staff:
            def __init__(self, name, department):
                self.name = name
                self.department = department
            
            def display(self):
                return f"Staff: {self.name} - {self.department}"
        
        student = Student("Ram", "CSIT")
        teacher = Teacher("Prof. Sharma", "Mathematics")
        staff = Staff("Ms. Singh", "Administration")
        
        print(f"  📚 {student.display()}")
        print(f"  👨‍🏫 {teacher.display()}")
        print(f"  🏢 {staff.display()}")
        
        print("\n🔹 Duck Typing:")
        print("─" * 72)
        class Duck:
            def make_sound(self):
                return "Quack!"
        
        class Car:
            def make_sound(self):
                return "Beep Beep!"
        
        def sound_maker(obj):
            return f"{obj.make_sound()}"
        
        duck = Duck()
        car = Car()
        print(f"  🦆 Duck says: {sound_maker(duck)}")
        print(f"  🚗 Car says: {sound_maker(car)}")
        
        print("\n💡 Polymorphism in Django:")
        print("─" * 72)
        print("  • Different model classes can implement the same methods")
        print("  • Class-based views with different implementations")
        print("  • Template rendering with different contexts")
        
        print()
        print("═" * 70)

polymorphism_demo = PolymorphismDemo()
polymorphism_demo.display()
```

    <p>URLs route requests to views. Views handle logic and return responses.</p>
</div>

```python
class URLsViewsDemo:
    """Demonstrate URLs and views in Django"""
    
    def display(self):
        print("\n" + "╔" + "═" * 70 + "╗")
        print("║" + " " * 25 + "🔗 URLS AND VIEWS" + " " * 27 + "║")
        print("╚" + "═" * 70 + "╝")
        
        print("\n🔹 URL Routing in Django:")
        print("─" * 72)
        print('''  # Project-level urls.py
  from django.contrib import admin
  from django.urls import path, include

  urlpatterns = [
      path('admin/', admin.site.urls),
      path('', include('myapp.urls')),  # Include app URLs
  ]

  # App-level urls.py
  from django.urls import path
  from . import views

  urlpatterns = [
      path('', views.home, name='home'),
      path('about/', views.about, name='about'),
      path('student/<int:id>/', views.student_detail, name='student_detail'),
  ]''')
        
        print("\n🔹 Function-Based Views:")
        print("─" * 72)
        print('''  # views.py
  from django.shortcuts import render
  from django.http import HttpResponse

  def home(request):
      return HttpResponse("Welcome to my website!")

  def about(request):
      return HttpResponse("About us page.")

  def student_detail(request, id):
      return HttpResponse(f"Student details for ID: {id}")''')
        
        print("\n🔹 URL Patterns and Views:")
        print("─" * 72)
        print("  🔗 URL Pattern          →  👁️ View")
        print("  ─" * 47)
        print('  /                       →  home()')
        print('  /about/                 →  about()')
        print('  /student/1/             →  student_detail(id=1)')
        print('  /student/2/             →  student_detail(id=2)')
        
        print("\n🔹 Using Templates with Views:")
        print("─" * 72)
        print('''  def home(request):
      context = {
          'title': 'Home Page',
          'message': 'Welcome to Django!'
      }
      return render(request, 'home.html', context)''')
        
        print("\n💡 URLs and Views in Django:")
        print("─" * 72)
        print("  • URLs define the routing (what URL goes where)")
        print("  • Views contain the logic (what to do with the request)")
        print("  • Templates render the HTML (what to show the user)")
        print("  • Follow the MVT pattern: URL → View → Template")
        
        print()
        print("═" * 70)

urls_views_demo = URLsViewsDemo()
urls_views_demo.display()
```

```python
class PracticalProject:
    """Complete Django project example"""
    
    def display(self):
        print("\n" + "╔" + "═" * 70 + "╗")
        print("║" + " " * 20 + "💻 PRACTICAL: DJANGO PROJECT" + " " * 21 + "║")
        print("╚" + "═" * 70 + "╝")
        
        print("\n📋 Complete Django Project Example:")
        print("─" * 72)
        
        print("\n📁 File Structure:")
        print("─" * 72)
        print('''  myproject/
  ├── manage.py
  ├── myproject/
  │   ├── __init__.py
  │   ├── settings.py
  │   └── urls.py
  └── myapp/
      ├── __init__.py
      ├── admin.py
      ├── apps.py
      ├── models.py
      ├── views.py
      ├── urls.py
      └── tests.py''')
        
        print("\n📄 myproject/urls.py:")
        print("─" * 72)
        print('''  from django.contrib import admin
  from django.urls import path, include

  urlpatterns = [
      path('admin/', admin.site.urls),
      path('', include('myapp.urls')),
  ]''')
        
        print("\n📄 myapp/urls.py:")
        print("─" * 72)
        print('''  from django.urls import path
  from . import views

  urlpatterns = [
      path('', views.home, name='home'),
      path('about/', views.about, name='about'),
      path('student/<int:student_id>/', views.student_detail, name='student_detail'),
  ]''')
        
        print("\n📄 myapp/views.py:")
        print("─" * 72)
        print('''  from django.shortcuts import render
  from django.http import HttpResponse

  # Simulated database
  students = {
      1: {'name': 'Ram', 'age': 21, 'course': 'CSIT'},
      2: {'name': 'Sita', 'age': 22, 'course': 'BBA'},
      3: {'name': 'Hari', 'age': 20, 'course': 'BSc IT'},
  }

  def home(request):
      return HttpResponse("Welcome to the Student Management System!")

  def about(request):
      return HttpResponse("This is a simple Django app for managing students.")

  def student_detail(request, student_id):
      student = students.get(student_id)
      if student:
          return HttpResponse(f"{student['name']} - {student['age']} - {student['course']}")
      else:
          return HttpResponse("Student not found", status=404)''')
        
        print("\n🔹 Test the URLs:")
        print("─" * 72)
        print("  🌐 http://127.0.0.1:8000/             →  Welcome to the Student Management System!")
        print("  🌐 http://127.0.0.1:8000/about/       →  This is a simple Django app for managing students.")
        print("  🌐 http://127.0.0.1:8000/student/1/   →  Ram - 21 - CSIT")
        print("  🌐 http://127.0.0.1:8000/student/2/   →  Sita - 22 - BBA")
        
        print()
        print("═" * 70)
        print("💡 You've just built your first Django application! 🎉")
        print("═" * 70)

practical = PracticalProject()
practical.display()
```
