```python
class ListViewDemo:
    """Demonstrate ListView"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 28 + "🔍 LISTVIEW" + " " * 28)
        print("═" * 70)
        print()
        
        print("🔹 Basic ListView:")
        print("─" * 70)
        print('''
  from django.views.generic import ListView
  from .models import Student

  class StudentListView(ListView):
      model = Student
      template_name = 'students/student_list.html'
      context_object_name = 'students'
      ordering = ['name']
      paginate_by = 10''')
        
        print("\n🔹 Customizing ListView:")
        print("─" * 70)
        print('''
  class StudentListView(ListView):
      model = Student
      template_name = 'students/student_list.html'
      context_object_name = 'students'
      
      def get_queryset(self):
          # Custom queryset
          return Student.objects.filter(is_active=True).order_by('name')
      
      def get_context_data(self, **kwargs):
          # Add extra context
          context = super().get_context_data(**kwargs)
          context['total_students'] = Student.objects.count()
          context['active_students'] = Student.objects.filter(is_active=True).count()
          return context
      
      def get_template_names(self):
          # Dynamic template selection
          if self.request.user.is_staff:
              return ['students/admin_student_list.html']
          return ['students/student_list.html']''')
        
        print("\n🔹 Template Usage:")
        print("─" * 70)
        print('''
  <!-- student_list.html -->
  {% extends 'base.html' %}
  
  {% block content %}
      <h1>Students</h1>
      
      <!-- Display students -->
      {% for student in students %}
          <div class="student-card">
              <h3>{{ student.name }}</h3>
              <p>Age: {{ student.age }}</p>
              <p>Course: {{ student.course }}</p>
              <a href="{% url 'student_detail' student.id %}">View Details</a>
          </div>
      {% empty %}
          <p>No students found.</p>
      {% endfor %}
      
      <!-- Pagination -->
      {% if is_paginated %}
          <div class="pagination">
              {% if page_obj.has_previous %}
                  <a href="?page={{ page_obj.previous_page_number }}">Previous</a>
              {% endif %}
              
              {% for num in page_obj.paginator.page_range %}
                  <a href="?page={{ num }}">{{ num }}</a>
              {% endfor %}
              
              {% if page_obj.has_next %}
                  <a href="?page={{ page_obj.next_page_number }}">Next</a>
              {% endif %}
          </div>
      {% endif %}
  {% endblock %}''')
        
        print()
        print("═" * 70)

listview = ListViewDemo()
listview.display()
```

```python
class CreateViewDemo:
    """Demonstrate CreateView"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 28 + "➕ CREATEVIEW" + " " * 28)
        print("═" * 70)
        print()
        
        print("🔹 Basic CreateView:")
        print("─" * 70)
        print('''
  from django.views.generic import CreateView
  from django.urls import reverse_lazy
  from .models import Student
  from .forms import StudentForm

  class StudentCreateView(CreateView):
      model = Student
      form_class = StudentForm
      template_name = 'students/student_form.html'
      success_url = reverse_lazy('student_list')''')
        
        print("\n🔹 Customizing CreateView:")
        print("─" * 70)
        print('''
  class StudentCreateView(CreateView):
      model = Student
      fields = ['name', 'age', 'email', 'course']
      template_name = 'students/student_form.html'
      success_url = reverse_lazy('student_list')
      
      def get_initial(self):
          # Pre-fill form fields
          initial = super().get_initial()
          initial['course'] = 'CSIT'  # Default course
          return initial
      
      def form_valid(self, form):
          # Custom save logic
          form.instance.created_by = self.request.user
          messages.success(self.request, 'Student created successfully!')
          return super().form_valid(form)
      
      def form_invalid(self, form):
          messages.error(self.request, 'Please correct the errors below')
          return super().form_invalid(form)
      
      def get_context_data(self, **kwargs):
          context = super().get_context_data(**kwargs)
          context['title'] = 'Create New Student'
          context['button_text'] = 'Create Student'
          return context''')
        
        print("\n🔹 Template Usage:")
        print("─" * 70)
        print('''
  <!-- student_form.html -->
  {% extends 'base.html' %}
  
  {% block content %}
      <h1>{{ title }}</h1>
      
      <form method="POST">
          {% csrf_token %}
          
          <!-- Display form errors -->
          {% if form.errors %}
              <div class="errors">
                  <ul>
                      {% for field, errors in form.errors.items %}
                          {% for error in errors %}
                              <li>{{ field }}: {{ error }}</li>
                          {% endfor %}
                      {% endfor %}
                  </ul>
              </div>
          {% endif %}
          
          <!-- Render form fields -->
          {% for field in form %}
              <div class="form-group">
                  {{ field.label_tag }}
                  {{ field }}
                  {% if field.help_text %}
                      <small>{{ field.help_text }}</small>
                  {% endif %}
              </div>
          {% endfor %}
          
          <button type="submit">{{ button_text }}</button>
          <a href="{% url 'student_list' %}">Cancel</a>
      </form>
  {% endblock %}''')
        
        print()
        print("═" * 70)

createview = CreateViewDemo()
createview.display()
```

```python
class UpdateViewDemo:
    """Demonstrate UpdateView"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 28 + "✏️ UPDATEVIEW" + " " * 28)
        print("═" * 70)
        print()
        
        print("🔹 Basic UpdateView:")
        print("─" * 70)
        print('''
  from django.views.generic import UpdateView
  from django.urls import reverse_lazy
  from .models import Student
  from .forms import StudentForm

  class StudentUpdateView(UpdateView):
      model = Student
      form_class = StudentForm
      template_name = 'students/student_form.html'
      success_url = reverse_lazy('student_list')''')
        
        print("\n🔹 Customizing UpdateView:")
        print("─" * 70)
        print('''
  from django.contrib.auth.mixins import LoginRequiredMixin
  from django.core.exceptions import PermissionDenied

  class StudentUpdateView(LoginRequiredMixin, UpdateView):
      model = Student
      fields = ['name', 'age', 'email', 'course']
      template_name = 'students/student_form.html'
      success_url = reverse_lazy('student_list')
      
      def get_queryset(self):
          # Only allow editing of own students or staff
          if self.request.user.is_staff:
              return Student.objects.all()
          return Student.objects.filter(created_by=self.request.user)
      
      def form_valid(self, form):
          messages.success(self.request, 'Student updated successfully!')
          return super().form_valid(form)
      
      def get_context_data(self, **kwargs):
          context = super().get_context_data(**kwargs)
          context['title'] = 'Update Student'
          context['button_text'] = 'Update Student'
          return context
      
      def get_success_url(self):
          # Redirect to detail view after update
          return reverse_lazy('student_detail', kwargs={'pk': self.object.pk})''')
        
        print("\n🔹 Using ModelForm with UpdateView:")
        print("─" * 70)
        print('''
  # forms.py
  from django import forms
  from .models import Student

  class StudentForm(forms.ModelForm):
      class Meta:
          model = Student
          fields = ['name', 'age', 'email', 'course']
          widgets = {
              'name': forms.TextInput(attrs={'class': 'form-control'}),
              'age': forms.NumberInput(attrs={'class': 'form-control'}),
              'email': forms.EmailInput(attrs={'class': 'form-control'}),
              'course': forms.Select(attrs={'class': 'form-control'}),
          }
      
      def clean_email(self):
          email = self.cleaned_data.get('email')
          # Check if email exists for other students
          if Student.objects.exclude(pk=self.instance.pk).filter(email=email).exists():
              raise forms.ValidationError('This email is already in use')
          return email''')
        
        print()
        print("═" * 70)

updateview = UpdateViewDemo()
updateview.display()
```

    <h2 style="color: #ffc107;">⚖️ FBV vs CBV</h2>
    <p><strong>Function-Based Views vs Class-Based Views</strong></p>
</div>

```python
class FBVvsCBV:
    """Compare Function-Based Views and Class-Based Views"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 20 + "⚖️ FBV VS CBV COMPARISON" + " " * 20)
        print("═" * 70)
        print()
        
        print("🔹 Function-Based View (FBV):")
        print("─" * 70)
        print('''
  # views.py
  from django.shortcuts import render, get_object_or_404, redirect
  from .models import Student
  from .forms import StudentForm

  def student_list(request):
      students = Student.objects.all()
      return render(request, 'students/list.html', {'students': students})
  
  def student_detail(request, pk):
      student = get_object_or_404(Student, pk=pk)
      return render(request, 'students/detail.html', {'student': student})
  
  def student_create(request):
      if request.method == 'POST':
          form = StudentForm(request.POST)
          if form.is_valid():
              form.save()
              return redirect('student_list')
      else:
          form = StudentForm()
      return render(request, 'students/form.html', {'form': form})''')
        
        print("\n🔹 Class-Based View (CBV):")
        print("─" * 70)
        print('''
  # views.py
  from django.views.generic import ListView, DetailView, CreateView
  from django.urls import reverse_lazy
  from .models import Student
  from .forms import StudentForm

  class StudentListView(ListView):
      model = Student
      template_name = 'students/list.html'
      context_object_name = 'students'
  
  class StudentDetailView(DetailView):
      model = Student
      template_name = 'students/detail.html'
      context_object_name = 'student'
  
  class StudentCreateView(CreateView):
      model = Student
      form_class = StudentForm
      template_name = 'students/form.html'
      success_url = reverse_lazy('student_list')''')
        
        print("\n🔹 Comparison:")
        print("─" * 70)
        print("""
  ┌─────────────────────────────────────────────────────────────────────┐
  │  🔹 Function-Based Views         🔹 Class-Based Views              │
  │  ───────────────────────         ──────────────────────            │
  │                                                                   │
  │  ✅ Simple and direct            ✅ Reusable and DRY               │
  │  ✅ Easy to understand           ✅ Built-in features              │
  │  ✅ Explicit control             ✅ Less code for common tasks     │
  │  ❌ Code repetition              ❌ Steeper learning curve         │
  │  ❌ Harder to extend             ✅ Easy to extend via inheritance │
  │  ❌ No built-in mixins           ✅ Mixin support                  │
  │  ❌ Manual method handling       ✅ Automatic method dispatch      │
  └─────────────────────────────────────────────────────────────────────┘""")
        
        print("\n🔹 When to Use Which:")
        print("─" * 70)
        print('''
  📌 Use FBVs when:
  • You're building a simple view with unique logic
  • You need fine-grained control
  • Your team is new to Django
  • The view doesn't follow a common pattern
  
  📌 Use CBVs when:
  • You're implementing CRUD operations
  • You need reusable view patterns
  • You're using mixins
  • You want built-in features (pagination, forms, etc.)
  • You're building a large application''')
        
        print()
        print("═" * 70)

comparison = FBVvsCBV()
comparison.display()
```

    <h2 style="color: #20c997;">🔧 Practical: Complete CRUD Application</h2>
    <p><strong>Building a complete CRUD app using Generic Views</strong></p>
</div>

```python
import datetime
from typing import List, Dict, Optional, Any

# Simulated Django Generic Views
class Model:
    """Base model class"""
    pass

class Student(Model):
    """Student model"""
    
    def __init__(self, name, age, email, course):
        self.id = None
        self.name = name
        self.age = age
        self.email = email
        self.course = course
        self.created_at = datetime.datetime.now()
        self.updated_at = datetime.datetime.now()
    
    def __str__(self):
        return self.name

class StudentManager:
    """Simple manager for Student objects"""
    
    def __init__(self):
        self.objects = []
        self.next_id = 1
    
    def all(self):
        return self.objects
    
    def get(self, pk):
        for obj in self.objects:
            if obj.id == pk:
                return obj
        return None
    
    def create(self, **kwargs):
        student = Student(
            kwargs.get('name'),
            kwargs.get('age'),
            kwargs.get('email'),
            kwargs.get('course')
        )
        student.id = self.next_id
        self.next_id += 1
        self.objects.append(student)
        return student
    
    def update(self, pk, **kwargs):
        student = self.get(pk)
        if student:
            for key, value in kwargs.items():
                setattr(student, key, value)
            student.updated_at = datetime.datetime.now()
        return student
    
    def delete(self, pk):
        student = self.get(pk)
        if student:
            self.objects.remove(student)
            return True
        return False

class ListView:
    """Simulated ListView"""
    
    def __init__(self, manager, model, template_name='list.html', context_object_name='objects'):
        self.manager = manager
        self.model = model
        self.template_name = template_name
        self.context_object_name = context_object_name
    
    def get_queryset(self):
        return self.manager.all()
    
    def render(self):
        context = {self.context_object_name: self.get_queryset()}
        return context

class DetailView:
    """Simulated DetailView"""
    
    def __init__(self, manager, model, pk, template_name='detail.html', context_object_name='object'):
        self.manager = manager
        self.model = model
        self.pk = pk
        self.template_name = template_name
        self.context_object_name = context_object_name
    
    def get_object(self):
        return self.manager.get(self.pk)
    
    def render(self):
        context = {self.context_object_name: self.get_object()}
        return context

class CreateView:
    """Simulated CreateView"""
    
    def __init__(self, manager, model, data=None, template_name='form.html', success_url='/'):
        self.manager = manager
        self.model = model
        self.data = data
        self.template_name = template_name
        self.success_url = success_url
    
    def is_valid(self):
        return self.data is not None and all(self.data.values())
    
    def save(self):
        if self.is_valid():
            return self.manager.create(**self.data)
        return None
    
    def render(self):
        if self.is_valid():
            return {'success': True, 'object': self.save(), 'redirect': self.success_url}
        return {'success': False, 'form_data': self.data or {}}

class UpdateView:
    """Simulated UpdateView"""
    
    def __init__(self, manager, model, pk, data=None, template_name='form.html', success_url='/'):
        self.manager = manager
        self.model = model
        self.pk = pk
        self.data = data
        self.template_name = template_name
        self.success_url = success_url
    
    def get_object(self):
        return self.manager.get(self.pk)
    
    def is_valid(self):
        return self.data is not None and all(self.data.values())
    
    def save(self):
        if self.is_valid():
            return self.manager.update(self.pk, **self.data)
        return None
    
    def render(self):
        if self.is_valid():
            return {'success': True, 'object': self.save(), 'redirect': self.success_url}
        return {'success': False, 'object': self.get_object(), 'form_data': self.data or {}}

class DeleteView:
    """Simulated DeleteView"""
    
    def __init__(self, manager, model, pk, template_name='confirm_delete.html', success_url='/'):
        self.manager = manager
        self.model = model
        self.pk = pk
        self.template_name = template_name
        self.success_url = success_url
    
    def get_object(self):
        return self.manager.get(self.pk)
    
    def delete(self):
        return self.manager.delete(self.pk)
    
    def render(self):
        return {'object': self.get_object(), 'success_url': self.success_url}

class CRUDAppDemo:
    """Complete CRUD application demonstration"""
    
    def run_demo(self):
        print("\n" + "═" * 70)
        print(" " * 15 + "🔧 COMPLETE CRUD APPLICATION DEMO" + " " * 15)
        print("═" * 70)
        print()
        
        manager = StudentManager()
        
        # 1. CREATE
        print("🔹 1. CREATE (CreateView):")
        print("─" * 70)
        
        students_data = [
            {'name': 'Ram Sharma', 'age': 21, 'email': 'ram@example.com', 'course': 'CSIT'},
            {'name': 'Sita Patel', 'age': 22, 'email': 'sita@example.com', 'course': 'BBA'},
            {'name': 'Hari Gurung', 'age': 23, 'email': 'hari@example.com', 'course': 'CSIT'},
        ]
        
        created = []
        for data in students_data:
            create_view = CreateView(manager, Student, data=data, success_url='/students/')
            result = create_view.render()
            if result['success']:
                student = result['object']
                created.append(student)
                print(f"  ✅ Created: {student.name} (ID: {student.id})")
        
        # 2. LIST
        print("\n🔹 2. LIST (ListView):")
        print("─" * 70)
        list_view = ListView(manager, Student, context_object_name='students')
        context = list_view.render()
        for student in context['students']:
            print(f"  #{student.id}: {student.name} | {student.age} | {student.course}")
        
        # 3. DETAIL
        print("\n🔹 3. DETAIL (DetailView):")
        print("─" * 70)
        if created:
            student = created[0]
            detail_view = DetailView(manager, Student, student.id, context_object_name='student')
            context = detail_view.render()
            s = context['student']
            print(f"  📄 Student Details:")
            print(f"     Name: {s.name}")
            print(f"     Age: {s.age}")
            print(f"     Email: {s.email}")
            print(f"     Course: {s.course}")
        
        # 4. UPDATE
        print("\n🔹 4. UPDATE (UpdateView):")
        print("─" * 70)
        if created:
            student = created[1]
            update_data = {'name': student.name, 'age': 23, 'email': student.email, 'course': 'MBA'}
            update_view = UpdateView(manager, Student, student.id, data=update_data, success_url='/students/')
            result = update_view.render()
            if result['success']:
                updated = result['object']
                print(f"  ✅ Updated: {updated.name} | New Course: {updated.course}")
        
        # 5. DELETE
        print("\n🔹 5. DELETE (DeleteView):")
        print("─" * 70)
        if len(created) >= 3:
            student = created[2]
            print(f"  🗑️ Deleting: {student.name} (ID: {student.id})")
            delete_view = DeleteView(manager, Student, student.id, success_url='/students/')
            context = delete_view.render()
            if delete_view.delete():
                print(f"  ✅ Deleted successfully!")
        
        # 6. Final List
        print("\n🔹 6. FINAL STUDENT LIST:")
        print("─" * 70)
        for student in manager.all():
            print(f"  #{student.id}: {student.name} | {student.age} | {student.course}")
        
        print("\n" + "═" * 70)
        print("🎯 Complete CRUD Application Demo Complete!")
        print("═" * 70)

demo = CRUDAppDemo()
demo.run_demo()
```

```python
class MixinsDemo:
    """Demonstrate mixins for CBVs"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 25 + "🔧 MIXINS AND CUSTOMIZATION" + " " * 25)
        print("═" * 70)
        print()
        
        print("🔹 Common Mixins:")
        print("─" * 70)
        print('''
  from django.contrib.auth.mixins import (
      LoginRequiredMixin,
      PermissionRequiredMixin,
      UserPassesTestMixin,
  )
  from django.views.generic import ListView, DetailView
  from django.contrib import messages

  # 1. Login Required Mixin
  class StudentListView(LoginRequiredMixin, ListView):
      model = Student
      template_name = 'students/list.html'
      login_url = '/login/'
      redirect_field_name = 'redirect_to'
  
  # 2. Permission Required Mixin
  class StudentCreateView(LoginRequiredMixin, PermissionRequiredMixin, CreateView):
      model = Student
      fields = '__all__'
      permission_required = 'app.add_student'
      raise_exception = True
      
  # 3. User Passes Test Mixin
  class StudentUpdateView(LoginRequiredMixin, UserPassesTestMixin, UpdateView):
      model = Student
      fields = '__all__'
      
      def test_func(self):
          student = self.get_object()
          return self.request.user == student.created_by or self.request.user.is_staff''')
        
        print("\n🔹 Custom Mixin Example:")
        print("─" * 70)
        print('''
  from django.contrib import messages

  class MessageMixin:
      """Add success messages to views"""
      success_message = ''
      
      def form_valid(self, form):
          response = super().form_valid(form)
          if self.success_message:
              messages.success(self.request, self.success_message)
          return response
  
  class TitleMixin:
      """Add page titles to views"""
      page_title = ''
      
      def get_context_data(self, **kwargs):
          context = super().get_context_data(**kwargs)
          context['page_title'] = self.page_title
          return context
  
  # Using custom mixins
  class StudentCreateView(LoginRequiredMixin, MessageMixin, TitleMixin, CreateView):
      model = Student
      fields = ['name', 'age', 'email', 'course']
      success_message = 'Student created successfully!'
      page_title = 'Create New Student'
      
      def form_valid(self, form):
          form.instance.created_by = self.request.user
          return super().form_valid(form)''')
        
        print()
        print("═" * 70)

mixins = MixinsDemo()
mixins.display()
```
