<div style="background: #f8f9fa; padding: 25px; border-radius: 15px; border-left: 6px solid #28a745; margin: 20px 0;">
    <h2 style="color: #28a745; margin-top: 0;">📄 1. Django Template Language (DTL)</h2>
    <p>DTL is Django's built-in template system for generating HTML dynamically.</p>
</div>

```python
class TemplateLanguageDemo:
    """Demonstrate Django Template Language"""
    
    def display(self):
        print("\n" + "╔" + "═" * 70 + "╗")
        print("║" + " " * 20 + "📄 DJANGO TEMPLATE LANGUAGE" + " " * 21 + "║")
        print("╚" + "═" * 70 + "╝")
        
        print("\n🔹 Template Tags (Logic):")
        print("─" * 72)
        print('''  {% tag %}  →  Control flow tags (if, for, etc.)
  {% if user.is_authenticated %}
      Welcome, {{ user.username }}!
  {% else %}
      Please login.
  {% endif %}

  {% for student in students %}
      <li>{{ student.name }} - {{ student.age }}</li>
  {% empty %}
      <li>No students found.</li>
  {% endfor %}''')
        
        print("\n🔹 Template Variables (Data):")
        print("─" * 72)
        print('''  {{ variable }}  →  Display variable value
  {{ student.name }}  →  Display student's name
  {{ student.age|add:"1" }}  →  Filters modify values''')
        
        print("\n🔹 Template Filters:")
        print("─" * 72)
        print('''  {{ name|upper }}         →  RAM
  {{ name|lower }}         →  ram
  {{ name|title }}         →  Ram
  {{ name|length }}        →  3
  {{ text|truncatechars:10 }}  →  Shortened text...
  {{ date|date:"Y-m-d" }}  →  2024-01-01
  {{ value|default:"N/A" }}  →  N/A if value is empty''')
        
        print("\n🔹 Template Comments:")
        print("─" * 72)
        print('''  {# This is a single-line comment #}
  
  {% comment %}
      This is a multi-line comment
      It won't appear in the HTML output
  {% endcomment %}''')
        
        print("\n💡 Template Tags in Django:")
        print("─" * 72)
        print("  • {% for %}     →  Loop through data")
        print("  • {% if %}      →  Conditional logic")
        print("  • {% block %}   →  Template inheritance")
        print("  • {% extends %} →  Extend a base template")
        print("  • {% include %} →  Include another template")
        print("  • {% load %}    →  Load template tags")
        
        print()
        print("═" * 70)

template_lang = TemplateLanguageDemo()
template_lang.display()
```

    <h2 style="color: #6f42c1; margin-top: 0;">🧬 2. Template Inheritance</h2>
    <p>Template inheritance allows you to create a base layout and extend it in child templates.</p>
</div>

```python
class InheritanceDemo:
    """Demonstrate template inheritance"""
    
    def display(self):
        print("\n" + "╔" + "═" * 70 + "╗")
        print("║" + " " * 22 + "🧬 TEMPLATE INHERITANCE" + " " * 23 + "║")
        print("╚" + "═" * 70 + "╝")
        
        print("\n🔹 Base Template (base.html):")
        print("─" * 72)
        print('''  <!DOCTYPE html>
  <html>
  <head>
      <title>{% block title %}My Site{% endblock %}</title>
      <link rel="stylesheet" href="{% static 'css/style.css' %}">
  </head>
  <body>
      <header>
          <h1>My Website</h1>
          <nav>
              <a href="/">Home</a>
              <a href="/about/">About</a>
          </nav>
      </header>
  
      <main>
          {% block content %}
              <!-- Default content -->
          {% endblock %}
      </main>
  
      <footer>
          <p>&copy; 2024 My Website</p>
      </footer>
  </body>
  </html>''')
        
        print("\n🔹 Child Template (home.html):")
        print("─" * 72)
        print('''  {% extends 'base.html' %}
  
  {% block title %}Home Page{% endblock %}
  
  {% block content %}
      <h2>Welcome to the Home Page</h2>
      <p>This content replaces the base content block.</p>
  {% endblock %}''')
        
        print("\n🔹 Child Template (about.html):")
        print("─" * 72)
        print('''  {% extends 'base.html' %}
  
  {% block title %}About Us{% endblock %}
  
  {% block content %}
      <h2>About Our Company</h2>
      <p>We are a Django development company.</p>
  {% endblock %}''')
        
        print("\n💡 Template Inheritance Flow:")
        print("─" * 72)
        print("              base.html (Layout)")
        print("                    │")
        print("              extends / blocks")
        print("                    │")
        print("        ┌───────────┴───────────┐")
        print("        │                       │")
        print("   home.html              about.html")
        print("   (page content)         (page content)")
        
        print()
        print("═" * 70)

inherit_demo = InheritanceDemo()
inherit_demo.display()
```

    <h2 style="color: #17a2b8; margin-top: 0;">📤 3. Passing Data to Templates</h2>
    <p>Views pass data to templates using context dictionaries.</p>
</div>

```python
class PassingDataDemo:
    """Demonstrate passing data to templates"""
    
    def display(self):
        print("\n" + "╔" + "═" * 70 + "╗")
        print("║" + " " * 20 + "📤 PASSING DATA TO TEMPLATES" + " " * 21 + "║")
        print("╚" + "═" * 70 + "╝")
        
        print("\n🔹 View with Context (views.py):")
        print("─" * 72)
        print('''  from django.shortcuts import render
  from datetime import datetime

  def student_list(request):
      students = [
          {'name': 'Ram', 'age': 21, 'course': 'CSIT'},
          {'name': 'Sita', 'age': 22, 'course': 'BBA'},
          {'name': 'Hari', 'age': 20, 'course': 'BSc IT'},
      ]
      
      context = {
          'page_title': 'Student List',
          'students': students,
          'total_students': len(students),
          'current_date': datetime.now(),
          'user': request.user,
      }
      
      return render(request, 'student_list.html', context)''')
        
        print("\n🔹 Template Receiving Data (student_list.html):")
        print("─" * 72)
        print('''  <!DOCTYPE html>
  <html>
  <head>
      <title>{{ page_title }}</title>
  </head>
  <body>
      <h1>{{ page_title }}</h1>
      <p>Total Students: {{ total_students }}</p>
      <p>Current Date: {{ current_date|date:"Y-m-d" }}</p>
      
      <h2>Student List</h2>
      <ul>
          {% for student in students %}
              <li>
                  {{ student.name }} - Age: {{ student.age }}
                  - Course: {{ student.course }}
              </li>
          {% empty %}
              <li>No students found.</li>
          {% endfor %}
      </ul>
  </body>
  </html>''')
        
        # Simulated data
        students = [
            {'name': 'Ram', 'age': 21, 'course': 'CSIT'},
            {'name': 'Sita', 'age': 22, 'course': 'BBA'},
            {'name': 'Hari', 'age': 20, 'course': 'BSc IT'},
        ]
        
        print("\n🔹 Simulated Output:")
        print("─" * 72)
        print(f"  Page Title: Student List")
        print(f"  Total Students: {len(students)}")
        print(f"  Current Date: 2024-01-01")
        print()
        for student in students:
            print(f"  {student['name']} - Age: {student['age']} - Course: {student['course']}")
        
        print("\n🔹 Passing Multiple Data Types:")
        print("─" * 72)
        print('''  context = {
      'string': 'Hello World',
      'number': 42,
      'list': [1, 2, 3],
      'dict': {'key': 'value'},
      'object': student_object,
      'boolean': True,
      'none_value': None,
  }''')
        
        print("\n💡 Context Data in Django:")
        print("─" * 72)
        print("  • Context is a dictionary passed to render()")
        print("  • Keys become variable names in templates")
        print("  • Values can be any Python object")
        print("  • Context processors add global data")
        
        print()
        print("═" * 70)

data_demo = PassingDataDemo()
data_demo.display()
```

    <p>Static files are CSS, JavaScript, images, and other files served directly to the client.</p>
</div>

```python
class StaticFilesDemo:
    """Demonstrate static files in Django"""
    
    def display(self):
        print("\n" + "╔" + "═" * 70 + "╗")
        print("║" + " " * 25 + "🎨 STATIC FILES" + " " * 27 + "║")
        print("╚" + "═" * 70 + "╝")
        
        print("\n🔹 Static Files Directory Structure:")
        print("─" * 72)
        print('''  myapp/
  ├── static/
  │   ├── css/
  │   │   └── style.css
  │   ├── js/
  │   │   └── script.js
  │   └── images/
  │       └── logo.png
  └── templates/
      └── index.html''')
        
        print("\n🔹 settings.py Configuration:")
        print("─" * 72)
        print('''  # Static files settings
  STATIC_URL = '/static/'
  
  # Additional static files directories
  STATICFILES_DIRS = [
      BASE_DIR / "static",
  ]
  
  # Production static files
  STATIC_ROOT = BASE_DIR / "staticfiles"''')
        
        print("\n🔹 Loading Static Files in Templates:")
        print("─" * 72)
        print('''  {% load static %}
  
  <!DOCTYPE html>
  <html>
  <head>
      <!-- CSS -->
      <link rel="stylesheet" href="{% static 'css/style.css' %}">
      
      <!-- JavaScript -->
      <script src="{% static 'js/script.js' %}"></script>
  </head>
  <body>
      <!-- Image -->
      <img src="{% static 'images/logo.png' %}" alt="Logo">
  </body>
  </html>''')
        
        print("\n🔹 Example: style.css:")
        print("─" * 72)
        print('''  body {
      font-family: Arial, sans-serif;
      margin: 0;
      padding: 20px;
      background-color: #f4f4f4;
  }
  
  h1 {
      color: #333;
      text-align: center;
  }
  
  .student-card {
      background: white;
      padding: 15px;
      margin: 10px 0;
      border-radius: 5px;
      box-shadow: 0 2px 5px rgba(0,0,0,0.1);
  }''')
        
        print("\n🔹 Example: script.js:")
        print("─" * 72)
        print('''  document.addEventListener('DOMContentLoaded', function() {
      console.log('Page loaded!');
      
      // Alert when a student card is clicked
      document.querySelectorAll('.student-card').forEach(card => {
          card.addEventListener('click', function() {
              alert('Student: ' + this.dataset.name);
          });
      });
  });''')
        
        print("\n💡 Static Files in Django:")
        print("─" * 72)
        print("  • Static files are served directly to the client")
        print("  • {% load static %} must be used in templates")
        print("  • STATIC_URL is the URL prefix")
        print("  • STATICFILES_DIRS for additional directories")
        print("  • STATIC_ROOT for production (collectstatic)")
        
        print()
        print("═" * 70)

static_demo = StaticFilesDemo()
static_demo.display()
```

```python
class ReusableTemplatesDemo:
    """Demonstrate reusable templates"""
    
    def display(self):
        print("\n" + "╔" + "═" * 70 + "╗")
        print("║" + " " * 24 + "🔁 REUSABLE TEMPLATES" + " " * 25 + "║")
        print("╚" + "═" * 70 + "╝")
        
        print("\n🔹 1. Base Template (base.html):")
        print("─" * 72)
        print('''  {% load static %}
  <!DOCTYPE html>
  <html>
  <head>
      <title>{% block title %}My Site{% endblock %}</title>
      <link rel="stylesheet" href="{% static 'css/style.css' %}">
  </head>
  <body>
      <!-- Navigation Bar -->
      <nav class="navbar">
          <div class="nav-container">
              <a href="/" class="logo">MySite</a>
              <ul class="nav-menu">
                  <li><a href="/">Home</a></li>
                  <li><a href="/students/">Students</a></li>
                  <li><a href="/about/">About</a></li>
              </ul>
          </div>
      </nav>
  
      <!-- Main Content -->
      <main class="main-content">
          {% block content %}{% endblock %}
      </main>
  
      <!-- Footer -->
      <footer class="footer">
          <p>&copy; 2024 MySite. All rights reserved.</p>
      </footer>
  </body>
  </html>''')
        
        print("\n🔹 2. Reusable Component - student_card.html:")
        print("─" * 72)
        print('''  <div class="student-card">
      <h3>{{ student.name }}</h3>
      <p><strong>Age:</strong> {{ student.age }}</p>
      <p><strong>Course:</strong> {{ student.course }}</p>
      <p><strong>Email:</strong> {{ student.email }}</p>
      <button class="btn btn-primary">View Details</button>
  </div>''')
        
        print("\n🔹 3. Using Include in Templates:")
        print("─" * 72)
        print('''  {% extends 'base.html' %}
  
  {% block content %}
      <h2>Student List</h2>
      <div class="student-grid">
          {% for student in students %}
              {% include 'components/student_card.html' %}
          {% endfor %}
      </div>
  {% endblock %}''')
        
        print("\n🔹 4. Reusable Pagination Component:")
        print("─" * 72)
        print('''  <!-- components/pagination.html -->
  <div class="pagination">
      {% if page.has_previous %}
          <a href="?page=1">&laquo; First</a>
          <a href="?page={{ page.previous_page_number }}">Previous</a>
      {% endif %}
  
      <span class="current">
          Page {{ page.number }} of {{ page.paginator.num_pages }}
      </span>
  
      {% if page.has_next %}
          <a href="?page={{ page.next_page_number }}">Next</a>
          <a href="?page={{ page.paginator.num_pages }}">Last &raquo;</a>
      {% endif %}
  </div>''')
        
        print("\n🔹 5. Reusable Template Tags (custom_tags.py):")
        print("─" * 72)
        print('''  from django import template
  
  register = template.Library()
  
  @register.filter
  def format_price(value):
      return f"Rs. {value:,.2f}"
  
  @register.simple_tag
  def current_year():
      import datetime
      return datetime.datetime.now().year''')
        
        print("\n💡 Template Reusability Best Practices:")
        print("─" * 72)
        print("  • Use {% include %} for small, reusable components")
        print("  • Use {% extends %} for page-level inheritance")
        print("  • Create custom template tags for complex logic")
        print("  • Keep templates DRY (Don't Repeat Yourself)")
        print("  • Use block names consistently")
        
        print()
        print("═" * 70)

reusable_demo = ReusableTemplatesDemo()
reusable_demo.display()
```

```python
class PracticalDemo:
    """Complete student management system"""
    
    def display(self):
        print("\n" + "╔" + "═" * 70 + "╗")
        print("║" + " " * 18 + "💻 COMPLETE STUDENT MANAGEMENT SYSTEM" + " " * 19 + "║")
        print("╚" + "═" * 70 + "╝")
        
        print("\n📋 Project Structure:")
        print("─" * 72)
        print('''  student_management/
  ├── manage.py
  ├── student_management/
  │   ├── settings.py
  │   └── urls.py
  └── students/
      ├── static/
      │   ├── css/
      │   │   └── style.css
      │   └── js/
      │       └── script.js
      ├── templates/
      │   ├── base.html
      │   ├── home.html
      │   ├── student_list.html
      │   └── student_detail.html
      ├── views.py
      └── urls.py''')
        
        print("\n📄 views.py:")
        print("─" * 72)
        print('''  from django.shortcuts import render, get_object_or_404
  
  # Sample data
  students = {
      1: {'name': 'Ram', 'age': 21, 'course': 'CSIT', 'email': 'ram@example.com'},
      2: {'name': 'Sita', 'age': 22, 'course': 'BBA', 'email': 'sita@example.com'},
      3: {'name': 'Hari', 'age': 20, 'course': 'BSc IT', 'email': 'hari@example.com'},
      4: {'name': 'Gita', 'age': 23, 'course': 'MBA', 'email': 'gita@example.com'},
  }
  
  def home(request):
      context = {
          'page_title': 'Student Management System',
          'total_students': len(students),
      }
      return render(request, 'home.html', context)
  
  def student_list(request):
      context = {
          'page_title': 'Student List',
          'students': students,
          'total_students': len(students),
      }
      return render(request, 'student_list.html', context)
  
  def student_detail(request, student_id):
      student = get_object_or_404(students, id=student_id)
      context = {
          'page_title': f'Student: {student["name"]}',
          'student': student,
      }
      return render(request, 'student_detail.html', context)''')
        
        print("\n📄 urls.py:")
        print("─" * 72)
        print('''  from django.urls import path
  from . import views

  urlpatterns = [
      path('', views.home, name='home'),
      path('students/', views.student_list, name='student_list'),
      path('student/<int:student_id>/', views.student_detail, name='student_detail'),
  ]''')
        
        print("\n📄 base.html (Layout):")
        print("─" * 72)
        print('''  {% load static %}
  <!DOCTYPE html>
  <html>
  <head>
      <title>{% block title %}SMS{% endblock %}</title>
      <link rel="stylesheet" href="{% static 'css/style.css' %}">
  </head>
  <body>
      <nav>
          <div class="nav-container">
              <a href="/" class="brand">📚 SMS</a>
              <ul>
                  <li><a href="/">Home</a></li>
                  <li><a href="/students/">Students</a></li>
              </ul>
          </div>
      </nav>
      <main>{% block content %}{% endblock %}</main>
      <footer>
          <p>&copy; 2024 Student Management System</p>
      </footer>
      <script src="{% static 'js/script.js' %}"></script>
  </body>
  </html>''')
        
        print()
        print("═" * 70)
        print("💡 This is a complete, working Django application with templates and styling!")
        print("═" * 70)

practical = PracticalDemo()
practical.display()
```
