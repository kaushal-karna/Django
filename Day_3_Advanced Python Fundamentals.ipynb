    <h2 style="color: #28a745; margin-top: 0;">📋 1. Lists</h2>
    <p>Lists are ordered, mutable collections that can hold different data types.</p>
</div>

```python
class ListsDemo:
    """Demonstrate lists in Python"""
    
    def display(self):
        print("\n" + "╔" + "═" * 70 + "╗")
        print("║" + " " * 28 + "📋 LISTS" + " " * 29 + "║")
        print("╚" + "═" * 70 + "╝")
        
        print("\n🔹 Creating Lists:")
        print("─" * 72)
        numbers = [1, 2, 3, 4, 5]
        fruits = ['apple', 'banana', 'cherry']
        mixed = [1, 'hello', 3.14, True]
        print(f"  numbers = {numbers}")
        print(f"  fruits = {fruits}")
        print(f"  mixed = {mixed}")
        
        print("\n🔹 List Operations:")
        print("─" * 72)
        fruits = ['apple', 'banana', 'cherry']
        print(f"  📊 fruits = {fruits}")
        print(f"  fruits[0] = {fruits[0]}")
        print(f"  fruits[-1] = {fruits[-1]}")
        print(f"  fruits[1:3] = {fruits[1:3]}")
        
        print("\n🔹 List Methods:")
        print("─" * 72)
        fruits = ['apple', 'banana', 'cherry']
        fruits.append('date')
        print(f"  fruits.append('date')     → {fruits}")
        
        fruits = ['apple', 'banana', 'cherry']
        fruits.insert(1, 'peach')
        print(f"  fruits.insert(1, 'peach') → {fruits}")
        
        fruits = ['apple', 'banana', 'cherry']
        fruits.remove('banana')
        print(f"  fruits.remove('banana')   → {fruits}")
        
        fruits = ['apple', 'banana', 'cherry']
        popped = fruits.pop()
        print(f"  popped = fruits.pop()     → {fruits}")
        
        fruits = ['apple', 'banana', 'cherry']
        fruits.reverse()
        print(f"  fruits.reverse()          → {fruits}")
        
        fruits = ['cherry', 'apple', 'peach']
        fruits.sort()
        print(f"  fruits.sort()             → {fruits}")
        
        print("\n🔹 List Comprehension:")
        print("─" * 72)
        squares = [x**2 for x in range(1, 6)]
        print(f"  squares = [x**2 for x in range(1, 6)]")
        print(f"  → {squares}")
        
        even_numbers = [x for x in range(1, 11) if x % 2 == 0]
        print(f"\n  even_numbers = [x for x in range(1, 11) if x % 2 == 0]")
        print(f"  → {even_numbers}")
        
        print("\n💡 Lists in Django:")
        print("─" * 72)
        print("  • QuerySets are list-like objects")
        print("  • Form data is often stored in lists")
        print("  • Used for iterating in templates")
        
        print()
        print("═" * 70)

lists_demo = ListsDemo()
lists_demo.display()
```

    <h2 style="color: #fd7e14; margin-top: 0;">🧩 2. Tuples</h2>
    <p>Tuples are ordered, immutable collections. They cannot be changed after creation.</p>
</div>

```python
class TuplesDemo:
    """Demonstrate tuples in Python"""
    
    def display(self):
        print("\n" + "╔" + "═" * 70 + "╗")
        print("║" + " " * 28 + "🧩 TUPLES" + " " * 29 + "║")
        print("╚" + "═" * 70 + "╝")
        
        print("\n🔹 Creating Tuples:")
        print("─" * 72)
        coordinates = (10, 20)
        colors = ('red', 'green', 'blue')
        single = (5,)  # Note the comma
        print(f"  coordinates = {coordinates}")
        print(f"  colors = {colors}")
        print(f"  single = {single}  # Note the comma")
        
        print("\n🔹 Tuple Operations:")
        print("─" * 72)
        colors = ('red', 'green', 'blue')
        print(f"  📊 colors = {colors}")
        print(f"  colors[0] = {colors[0]}")
        print(f"  colors[-1] = {colors[-1]}")
        print(f"  colors[1:3] = {colors[1:3]}")
        
        print("\n🔹 Tuple Methods:")
        print("─" * 72)
        colors = ('red', 'green', 'blue')
        print(f"  colors.index('green') = {colors.index('green')}")
        print(f"  colors.count('red') = {colors.count('red')}")
        print(f"  len(colors) = {len(colors)}")
        
        print("\n🔹 Tuple vs List:")
        print("─" * 72)
        print("  Tuple:  (1, 2, 3)  →  Immutable, faster, uses less memory")
        print("  List:   [1, 2, 3]  →  Mutable, slower, uses more memory")
        
        print("\n🔹 Tuple Unpacking:")
        print("─" * 72)
        x, y = (10, 20)
        print(f"  x, y = (10, 20)")
        print(f"  x = {x}, y = {y}")
        
        name, age, course = ('Ram', 21, 'CSIT')
        print(f"\n  name, age, course = ('Ram', 21, 'CSIT')")
        print(f"  name = {name}, age = {age}, course = {course}")
        
        print("\n💡 Tuples in Django:")
        print("─" * 72)
        print("  • Settings.py uses tuples for constants")
        print("  • Database field choices use tuples")
        print("  • URL patterns use tuples")
        
        print()
        print("═" * 70)

tuples_demo = TuplesDemo()
tuples_demo.display()
```

    <h2 style="color: #17a2b8; margin-top: 0;">🔷 3. Sets</h2>
    <p>Sets are unordered collections of unique elements.</p>
</div>

```python
class SetsDemo:
    """Demonstrate sets in Python"""
    
    def display(self):
        print("\n" + "╔" + "═" * 70 + "╗")
        print("║" + " " * 28 + "🔷 SETS" + " " * 29 + "║")
        print("╚" + "═" * 70 + "╝")
        
        print("\n🔹 Creating Sets:")
        print("─" * 72)
        numbers = {1, 2, 3, 4, 5}
        fruits = {'apple', 'banana', 'cherry'}
        print(f"  numbers = {numbers}")
        print(f"  fruits = {fruits}")
        
        print("\n🔹 Set Operations:")
        print("─" * 72)
        set1 = {1, 2, 3, 4, 5}
        set2 = {4, 5, 6, 7, 8}
        print(f"  📊 set1 = {set1}")
        print(f"  📊 set2 = {set2}")
        print(f"\n  Union:          {set1.union(set2)}")
        print(f"  Intersection:   {set1.intersection(set2)}")
        print(f"  Difference:     {set1.difference(set2)}")
        print(f"  Symmetric Diff: {set1.symmetric_difference(set2)}")
        
        print("\n🔹 Set Methods:")
        print("─" * 72)
        fruits = {'apple', 'banana', 'cherry'}
        fruits.add('date')
        print(f"  fruits.add('date')        → {fruits}")
        
        fruits = {'apple', 'banana', 'cherry', 'date'}
        fruits.remove('banana')
        print(f"  fruits.remove('banana')    → {fruits}")
        
        fruits = {'apple', 'banana', 'cherry'}
        fruits.discard('orange')
        print(f"  fruits.discard('orange')   → No error if not found")
        
        print("\n🔹 Removing Duplicates:")
        print("─" * 72)
        duplicates = [1, 2, 2, 3, 3, 3, 4]
        unique = set(duplicates)
        print(f"  List with duplicates: {duplicates}")
        print(f"  Set conversion: {unique}")
        
        print("\n💡 Sets in Django:")
        print("─" * 72)
        print("  • Used for efficient membership testing")
        print("  • Removing duplicates from data")
        print("  • User permissions and roles")
        
        print()
        print("═" * 70)

sets_demo = SetsDemo()
sets_demo.display()
```

    <h2 style="color: #e83e8c; margin-top: 0;">📚 4. Dictionaries</h2>
    <p>Dictionaries store key-value pairs. They are mutable and ordered.</p>
</div>

```python
class DictsDemo:
    """Demonstrate dictionaries in Python"""
    
    def display(self):
        print("\n" + "╔" + "═" * 70 + "╗")
        print("║" + " " * 24 + "📚 DICTIONARIES" + " " * 26 + "║")
        print("╚" + "═" * 70 + "╝")
        
        print("\n🔹 Creating Dictionaries:")
        print("─" * 72)
        student = {'name': 'Ram', 'age': 21, 'course': 'CSIT'}
        empty = {}
        print(f"  student = {student}")
        print(f"  empty = {empty}")
        
        print("\n🔹 Accessing Values:")
        print("─" * 72)
        student = {'name': 'Ram', 'age': 21, 'course': 'CSIT'}
        print(f"  student['name'] = {student['name']}")
        print(f"  student.get('age') = {student.get('age')}")
        print(f"  student.get('grade', 'Not Found') = {student.get('grade', 'Not Found')}")
        
        print("\n🔹 Dictionary Methods:")
        print("─" * 72)
        student = {'name': 'Ram', 'age': 21, 'course': 'CSIT'}
        print(f"  student.keys()   → {student.keys()}")
        print(f"  student.values() → {student.values()}")
        print(f"  student.items()  → {student.items()}")
        
        print("\n🔹 Adding/Updating Items:")
        print("─" * 72)
        student = {'name': 'Ram', 'age': 21, 'course': 'CSIT'}
        student['email'] = 'ram@example.com'
        print(f"  student['email'] = 'ram@example.com'")
        print(f"  → {student}")
        
        student.update({'grade': 'A', 'year': 2})
        print(f"\n  student.update({{'grade': 'A', 'year': 2}})")
        print(f"  → {student}")
        
        print("\n🔹 Removing Items:")
        print("─" * 72)
        student = {'name': 'Ram', 'age': 21, 'course': 'CSIT'}
        popped = student.pop('age')
        print(f"  student.pop('age') = {popped}")
        
        student = {'name': 'Ram', 'age': 21, 'course': 'CSIT'}
        del student['course']
        print(f"  del student['course']")
        print(f"  → {student}")
        
        print("\n🔹 Dictionary Comprehension:")
        print("─" * 72)
        squares = {x: x**2 for x in range(1, 6)}
        print(f"  squares = {{x: x**2 for x in range(1, 6)}}")
        print(f"  → {squares}")
        
        print("\n💡 Dictionaries in Django:")
        print("─" * 72)
        print("  • Template context is a dictionary")
        print("  • Form data is stored in dictionaries")
        print("  • JSON responses are dictionary-like")
        
        print()
        print("═" * 70)

dicts_demo = DictsDemo()
dicts_demo.display()
```

    <h2 style="color: #dc3545; margin-top: 0;">⚠️ 5. Exception Handling</h2>
    <p>Handle errors gracefully using try-except blocks.</p>
</div>

```python
class ExceptionDemo:
    """Demonstrate exception handling in Python"""
    
    def display(self):
        print("\n" + "╔" + "═" * 70 + "╗")
        print("║" + " " * 22 + "⚠️ EXCEPTION HANDLING" + " " * 24 + "║")
        print("╚" + "═" * 70 + "╝")
        
        print("\n🔹 Basic Try-Except:")
        print("─" * 72)
        try:
            result = 10 / 5
            print(f"  ✅ Division successful: {result}")
        except:
            print("  ❌ An error occurred!")
        
        print("\n🔹 Specific Exceptions:")
        print("─" * 72)
        # Test 1: Valid division
        try:
            result = 10 / 2
            print(f"  ✅ 10 / 2 = {result}")
        except ZeroDivisionError:
            print("  ❌ Error: Division by zero!")
        
        # Test 2: Division by zero
        try:
            result = 10 / 0
            print(f"  ✅ 10 / 0 = {result}")
        except ZeroDivisionError:
            print("\n  ❌ Error: Division by zero!")
        
        # Test 3: Value error
        try:
            num = int("abc")
            print(f"  ✅ Converted: {num}")
        except ValueError:
            print("\n  ❌ Error: Invalid input! Please enter numbers.")
        
        print("\n🔹 Try-Except-Else-Finally:")
        print("─" * 72)
        try:
            result = 10 / 2
            print(f"  ✅ 10 / 2 = {result}")
        except ZeroDivisionError:
            print("  ❌ Error: Division by zero!")
        else:
            print("  ✅ No error occurred.")
        finally:
            print("  🔧 Cleaning up resources...")
        
        print("\n🔹 Raising Exceptions:")
        print("─" * 72)
        def validate_age(age):
            if age < 18 or age > 60:
                raise ValueError("Age must be between 18 and 60")
            return f"Age {age} is valid."
        
        print("  Testing age validation:")
        try:
            result = validate_age(25)
            print(f"  ✅ {result}")
        except ValueError as e:
            print(f"  ❌ {e}")
        
        try:
            result = validate_age(16)
            print(f"  ✅ {result}")
        except ValueError as e:
            print(f"  ❌ {e}")
        
        print("\n🔹 Common Exceptions:")
        print("─" * 72)
        print("  • ValueError: Invalid literal for int()")
        print("  • TypeError: Unsupported operand type(s)")
        print("  • KeyError: Key not found in dictionary")
        print("  • IndexError: List index out of range")
        print("  • FileNotFoundError: File not found")
        
        print("\n💡 Exception Handling in Django:")
        print("─" * 72)
        print("  • Database errors (IntegrityError)")
        print("  • Authentication errors")
        print("  • Validation errors")
        print("  • 404 Page Not Found")
        
        print()
        print("═" * 70)

exception_demo = ExceptionDemo()
exception_demo.display()
```

    <h2 style="color: #ffc107; margin-top: 0;">📁 7. File Handling</h2>
    <p>Read from and write to files for data persistence.</p>
</div>

```python
import os
import csv

class FileHandlingDemo:
    """Demonstrate file handling in Python"""
    
    def display(self):
        print("\n" + "╔" + "═" * 70 + "╗")
        print("║" + " " * 25 + "📁 FILE HANDLING" + " " * 27 + "║")
        print("╚" + "═" * 70 + "╝")
        
        # Writing to a file
        print("\n🔹 Writing to a File:")
        print("─" * 72)
        try:
            with open('test.txt', 'w') as file:
                file.write("Hello, Python!\n")
                file.write("This is a test file.\n")
                file.write("Writing and reading files is easy!\n")
            print("  ✅ File written successfully!")
        except Exception as e:
            print(f"  ❌ Error: {e}")
        
        # Reading from a file
        print("\n🔹 Reading from a File:")
        print("─" * 72)
        try:
            with open('test.txt', 'r') as file:
                content = file.read()
                print("  📖 File content:")
                for line in content.split('\n'):
                    if line:
                        print(f"  {line}")
        except FileNotFoundError:
            print("  ❌ File not found!")
        
        # Reading line by line
        print("\n🔹 Reading Line by Line:")
        print("─" * 72)
        try:
            with open('test.txt', 'r') as file:
                for i, line in enumerate(file, 1):
                    print(f"  Line {i}: {line.strip()}")
        except FileNotFoundError:
            print("  ❌ File not found!")
        
        # Appending to a file
        print("\n🔹 Appending to a File:")
        print("─" * 72)
        try:
            with open('test.txt', 'a') as file:
                file.write("\nThis line was appended!\n")
            print("  ✅ Data appended successfully!")
        except Exception as e:
            print(f"  ❌ Error: {e}")
        
        # Working with CSV files
        print("\n🔹 Working with CSV Files:")
        print("─" * 72)
        try:
            with open('students.csv', 'w', newline='') as file:
                writer = csv.writer(file)
                writer.writerow(['Name', 'Age', 'Course'])
                writer.writerow(['Ram', '21', 'CSIT'])
                writer.writerow(['Sita', '22', 'BBA'])
            
            with open('students.csv', 'r') as file:
                reader = csv.reader(file)
                print("  📊 CSV Data:")
                for row in reader:
                    print(f"  {row}")
        except Exception as e:
            print(f"  ❌ Error: {e}")
        
        # Cleanup
        try:
            os.remove('test.txt')
            os.remove('students.csv')
        except:
            pass
        
        print("\n🔹 File Handling Best Practices:")
        print("─" * 72)
        print("  ✅ Use 'with' statement for automatic file closing")
        print("  ✅ Handle file not found errors")
        print("  ✅ Use appropriate file modes (r, w, a, r+, w+)")
        print("  ✅ Always close files or use context manager")
        
        print("\n💡 File Handling in Django:")
        print("─" * 72)
        print("  • Reading static files (CSS, JS)")
        print("  • Uploading user files")
        print("  • Writing log files")
        print("  • Reading configuration files")
        
        print()
        print("═" * 70)

file_demo = FileHandlingDemo()
file_demo.display()
```

    <p>OOP is a programming paradigm that uses objects and classes.</p>

```python
class OOPDemo:
    """Demonstrate Object-Oriented Programming concepts"""
    
    def display(self):
        print("\n" + "╔" + "═" * 70 + "╗")
        print("║" + " " * 18 + "🏗️ OBJECT-ORIENTED PROGRAMMING" + " " * 19 + "║")
        print("╚" + "═" * 70 + "╝")
        
        print("\n🔹 Creating a Class:")
        print("─" * 72)
        print('''  class Student:
      def __init__(self, name, age, course):
          self.name = name
          self.age = age
          self.course = course''')
        
        # Define the Student class
        class Student:
            def __init__(self, name, age, course):
                self.name = name
                self.age = age
                self.course = course
            
            def display_info(self):
                return f"{self.name}, {self.age}, {self.course}"
            
            def update_age(self, new_age):
                self.age = new_age
                return f"{self.name} updated age to: {self.age}"
        
        # Creating objects
        print("\n🔹 Creating Objects:")
        print("─" * 72)
        student1 = Student("Ram", 21, "CSIT")
        student2 = Student("Sita", 22, "BBA")
        print(f"  Student 1: {student1.display_info()}")
        print(f"  Student 2: {student2.display_info()}")
        
        # Class methods
        print("\n🔹 Class Methods:")
        print("─" * 72)
        print(f"  Ram's age: {student1.age}")
        print(f"  {student2.update_age(23)}")
        
        # Inheritance example
        print("\n🔹 Inheritance:")
        print("─" * 72)
        
        class GraduateStudent(Student):
            def __init__(self, name, age, course, thesis):
                super().__init__(name, age, course)
                self.thesis = thesis
            
            def display_info(self):
                return f"{super().display_info()}, Thesis: '{self.thesis}'"
        
        grad_student = GraduateStudent("Sita", 24, "MBA", "Marketing Strategies")
        print("  📚 Basic Student:")
        print(f"  Name: {student1.name}, Age: {student1.age}, Course: {student1.course}")
        print("\n  🎓 Graduate Student:")
        print(f"  {grad_student.display_info()}")
        
        # Polymorphism demonstration
        print("\n🔹 Inheritance Demonstration:")
        print("─" * 72)
        print("  📚 Student display method...")
        print(f"  {student1.display_info()}")
        print("\n  🎓 GraduateStudent display method...")
        print(f"  Name: {grad_student.name}, Age: {grad_student.age}, Course: {grad_student.course}")
        print(f"  Thesis: '{grad_student.thesis}'")
        
        print("\n🔹 Key OOP Concepts:")
        print("─" * 72)
        print("  📦 Encapsulation: Bundle data and methods together")
        print("  🧬 Inheritance: Create new classes from existing ones")
        print("  🔄 Polymorphism: Use objects of different types through same interface")
        print("  🔒 Abstraction: Hide complexity and show only essential features")
        
        print("\n💡 OOP in Django:")
        print("─" * 72)
        print("  • Models are classes (inherits from models.Model)")
        print("  • Forms are classes")
        print("  • Views can be class-based")
        print("  • Custom classes for business logic")
        
        print()
        print("═" * 70)

oop_demo = OOPDemo()
oop_demo.display()
```
