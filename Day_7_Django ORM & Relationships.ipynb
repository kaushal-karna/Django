    <h2 style="color: #6f42c1;">📚 Understanding QuerySets</h2>
    <p>A <strong>QuerySet</strong> is a list of objects from the database with filtering and ordering capabilities.</p>
</div>

```python
from IPython.display import clear_output

class QuerySetDemo:
    """Visual representation of QuerySets"""
    
    def display(self):
        print("\033[1m\033[36m" + "═" * 70 + "\033[0m")
        print("\033[1m\033[35m" + " " * 22 + "📚 UNDERSTANDING QUERYSETS" + " " * 22 + "\033[0m")
        print("\033[1m\033[36m" + "═" * 70 + "\033[0m")
        print()
        
        print("🔹 What is a QuerySet?")
        print("─" * 70)
        print("""
  • A collection of database records
  • Lazy-loaded (doesn't hit database until evaluated)
  • Chainable methods for filtering, ordering, and more
  • Represents a SQL query""")
        
        print("\n🔹 QuerySet Characteristics:")
        print("─" * 70)
        print("""
  ┌─────────────────────────────────────────────────────┐
  │  🐍 Django QuerySet                                │
  │  ─────────────────────────────────────────────────  │
  │  • Lazy Evaluation: Executes only when needed      │
  │  • Cached: Results are cached for efficiency       │
  │  • Chainable: Methods return new QuerySets         │
  │  • Iterable: Can be used in loops                  │
  │  • Convertible: Can be converted to lists/values   │
  └─────────────────────────────────────────────────────┘""")
        
        print("\n🔹 When QuerySets are Evaluated:")
        print("─" * 70)
        print("""
  🟢 Lazy (Not Evaluated):
  ─────────────────────────
  students = Student.objects.all()
  students.filter(course='CSIT')
  students.order_by('name')
  
  🔴 Evaluated (Hits Database):
  ─────────────────────────────
  • Iteration (for student in students:)
  • Slicing (students[0:5])
  • len(students)
  • list(students)
  • bool(students)
  • print(students)""")
        
        print("\n\033[1m\033[36m" + "═" * 70 + "\033[0m")

demo = QuerySetDemo()
demo.display()
```

    <p><strong>Master Django's powerful filtering capabilities</strong></p>

```python
class FilteringDemo:
    """Demonstrate advanced filtering"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 20 + "🔍 ADVANCED FILTERING & SEARCHING" + " " * 20)
        print("═" * 70)
        print()
        
        print("🔹 Field Lookups:")
        print("─" * 70)
        lookups = [
            ("exact", "Exact match", "name__exact='Ram'"),
            ("iexact", "Case-insensitive match", "name__iexact='ram'"),
            ("contains", "Contains string", "name__contains='Sharma'"),
            ("icontains", "Case-insensitive contains", "name__icontains='sharma'"),
            ("startswith", "Starts with", "name__startswith='S'"),
            ("endswith", "Ends with", "name__endswith='a'"),
            ("gt", "Greater than", "age__gt=21"),
            ("gte", "Greater than or equal", "age__gte=18"),
            ("lt", "Less than", "age__lt=25"),
            ("lte", "Less than or equal", "age__lte=30"),
            ("in", "In a list", "course__in=['CSIT', 'BBA']"),
            ("range", "Between range", "age__range=(18, 25)"),
            ("isnull", "Is NULL", "email__isnull=True"),
        ]
        for lookup, description, example in lookups:
            print(f"  {lookup:10s} → {description:25s} → {example}")
        
        print("\n🔹 Combining Conditions:")
        print("─" * 70)
        print('''
  📝 AND Condition (default):
  Student.objects.filter(course='CSIT', age__gte=18)
  # Equivalent: WHERE course='CSIT' AND age>=18
  
  📝 OR Condition:
  from django.db.models import Q
  Student.objects.filter(Q(course='CSIT') | Q(course='BBA'))
  # Equivalent: WHERE course='CSIT' OR course='BBA'
  
  📝 NOT Condition:
  Student.objects.exclude(course='CSIT')
  # Equivalent: WHERE NOT course='CSIT'
  
  📝 Complex Query:
  Student.objects.filter(
      Q(course='CSIT') | Q(age__gte=22),
      is_active=True
  )
  # Equivalent: (course='CSIT' OR age>=22) AND is_active=True"'')
        
        print()
        print("═" * 70)''')

filtering = FilteringDemo()
filtering.display()
```

    <h2 style="color: #20c997;">🔗 One-to-One Relationships</h2>
    <p><strong>Each record in one table corresponds to exactly one record in another</strong></p>

```python
class OneToOneDemo:
    """Demonstrate One-to-One relationships"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 20 + "🔗 ONE-TO-ONE RELATIONSHIPS" + " " * 20)
        print("═" * 70)
        print()
        
        print("🔹 Model Definition:")
        print("─" * 70)
        print('''
  from django.db import models
  from django.contrib.auth.models import User

  class Profile(models.Model):
      # One-to-One relationship with User
      user = models.OneToOneField(
          User,
          on_delete=models.CASCADE,
          related_name='profile'
      )
      bio = models.TextField(blank=True)
      location = models.CharField(max_length=100, blank=True)
      birth_date = models.DateField(null=True, blank=True)
      phone = models.CharField(max_length=15, blank=True)
      profile_picture = models.ImageField(upload_to='profiles/', null=True, blank=True)
      
      def __str__(self):
          return f"{self.user.username}'s Profile"''')
        
        print("\n🔹 Operations:")
        print("─" * 70)
        print('''
  📝 CREATE:
  user = User.objects.create(username='ram')
  profile = Profile.objects.create(
      user=user,
      bio='Django developer',
      location='Kathmandu'
  )
  
  📝 READ (Forward):
  user = User.objects.get(username='ram')
  bio = user.profile.bio  # Access via related_name
  
  📝 READ (Reverse):
  profile = Profile.objects.get(user__username='ram')
  username = profile.user.username  # Access the user
  
  📝 UPDATE:
  profile = Profile.objects.get(user__username='ram')
  profile.location = 'Pokhara'
  profile.save()
  
  📝 DELETE:
  user = User.objects.get(username='ram')
  user.profile.delete()  # Delete profile only
  # OR
  user.delete()  # Deletes both user and profile (CASCADE)''')
        
        print("\n🔹 Use Cases:")
        print("─" * 70)
        print("""
  ✅ User Profiles (extending User model)
  ✅ Configuration settings
  ✅ Credit card details (one user, one card)
  ✅ Passport details (one person, one passport)""")
        
        print()
        print("═" * 70)

onetoone = OneToOneDemo()
onetoone.display()
```

    <h2 style="color: #fd7e14;">🔗 One-to-Many Relationships</h2>
    <p><strong>One record in a table can relate to many records in another</strong></p>

```python
class OneToManyDemo:
    """Demonstrate One-to-Many relationships"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 20 + "🔗 ONE-TO-MANY RELATIONSHIPS" + " " * 20)
        print("═" * 70)
        print()
        
        print("🔹 Model Definition:")
        print("─" * 70)
        print('''
  from django.db import models

  class Department(models.Model):
      name = models.CharField(max_length=100)
      code = models.CharField(max_length=10, unique=True)
      location = models.CharField(max_length=100)
      
      def __str__(self):
          return self.name

  class Employee(models.Model):
      # Many employees can belong to ONE department
      department = models.ForeignKey(
          Department,
          on_delete=models.SET_NULL,  # Or CASCADE, PROTECT, etc.
          null=True,
          blank=True,
          related_name='employees'
      )
      name = models.CharField(max_length=100)
      position = models.CharField(max_length=50)
      salary = models.DecimalField(max_digits=10, decimal_places=2)
      hire_date = models.DateField(auto_now_add=True)
      
      def __str__(self):
          return self.name''')
        
        print("\n🔹 Operations:")
        print("─" * 70)
        print('''
  📝 CREATE:
  dept = Department.objects.create(name='IT', code='IT001', location='Floor 3')
  emp1 = Employee.objects.create(
      department=dept,
      name='Ram Sharma',
      position='Developer',
      salary=75000
  )
  emp2 = Employee.objects.create(
      department=dept,
      name='Sita Patel',
      position='Designer',
      salary=65000
  )
  
  📝 READ (Forward - Many to One):
  emp = Employee.objects.get(name='Ram Sharma')
  dept_name = emp.department.name  # Get department
  
  📝 READ (Reverse - One to Many):
  dept = Department.objects.get(code='IT001')
  employees = dept.employees.all()  # All employees in department
  
  📝 FILTER by related field:
  it_employees = Employee.objects.filter(department__code='IT001')
  high_salary = Employee.objects.filter(salary__gt=70000)
  
  📝 UPDATE:
  emp = Employee.objects.get(name='Ram Sharma')
  emp.department = Department.objects.get(code='HR001')
  emp.save()''')
        
        print("\n🔹 Use Cases:")
        print("─" * 70)
        print("""
  ✅ Department → Employees
  ✅ Blog Post → Comments
  ✅ Course → Students
  ✅ Category → Products
  ✅ Author → Books""")
        
        print()
        print("═" * 70)

onetomany = OneToManyDemo()
onetomany.display()
```

<div style="background: #f8f9fa; padding: 20px; border-radius: 10px; border-left: 5px solid #e83e8c;">
    <h2 style="color: #e83e8c;">🔗 Many-to-Many Relationships</h2>
    <p><strong>Many records in one table relate to many records in another</strong></p>
</div>

```python
class ManyToManyDemo:
    """Demonstrate Many-to-Many relationships"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 20 + "🔗 MANY-TO-MANY RELATIONSHIPS" + " " * 20)
        print("═" * 70)
        print()
        
        print("🔹 Model Definition:")
        print("─" * 70)
        print('''
  from django.db import models

  class Student(models.Model):
      name = models.CharField(max_length=100)
      email = models.EmailField(unique=True)
      enrolled_date = models.DateField(auto_now_add=True)
      
      def __str__(self):
          return self.name

  class Course(models.Model):
      name = models.CharField(max_length=100)
      code = models.CharField(max_length=10, unique=True)
      credits = models.IntegerField()
      instructor = models.CharField(max_length=100)
      # Many students can take many courses
      students = models.ManyToManyField(
          Student,
          related_name='courses',
          through='Enrollment'  # Custom through model
      )
      
      def __str__(self):
          return self.name

  class Enrollment(models.Model):
      # Custom through model with extra fields
      student = models.ForeignKey(Student, on_delete=models.CASCADE)
      course = models.ForeignKey(Course, on_delete=models.CASCADE)
      enrolled_date = models.DateField(auto_now_add=True)
      grade = models.CharField(max_length=2, blank=True, null=True)
      
      class Meta:
          unique_together = [['student', 'course']]''')
        
        print("\n🔹 Operations:")
        print("─" * 70)
        print('''
  📝 CREATE:
  student1 = Student.objects.create(name='Ram Sharma', email='ram@example.com')
  student2 = Student.objects.create(name='Sita Patel', email='sita@example.com')
  course = Course.objects.create(
      name='Python Programming',
      code='PY101',
      credits=3,
      instructor='Mr. Sharma'
  )
  
  📝 ADD Students to Course:
  course.students.add(student1, student2)
  # OR
  student1.courses.add(course)
  
  📝 READ (Forward):
  course = Course.objects.get(code='PY101')
  students = course.students.all()
  
  📝 READ (Reverse):
  student = Student.objects.get(email='ram@example.com')
  courses = student.courses.all()
  
  📝 FILTER by related:
  python_students = Student.objects.filter(courses__code='PY101')
  
  📝 REMOVE Students from Course:
  course.students.remove(student1)
  
  📝 CLEAR all Students:
  course.students.clear()
  
  📝 SET specific Students:
  course.students.set([student1, student2])''')
        
        print("\n🔹 Use Cases:")
        print("─" * 70)
        print("""
  ✅ Students ↔ Courses
  ✅ Products ↔ Categories
  ✅ Articles ↔ Tags
  ✅ Users ↔ Groups
  ✅ Books ↔ Authors""")
        
        print()
        print("═" * 70)

manytomany = ManyToManyDemo()
manytomany.display()
```

```python
class AdvancedOperations:
    """Demonstrate advanced operations"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 20 + "⚡ ADVANCED QUERY OPERATIONS" + " " * 20)
        print("═" * 70)
        print()
        
        print("🔹 Aggregation:")
        print("─" * 70)
        print('''
  from django.db.models import Count, Sum, Avg, Max, Min
  
  # Count
  total_students = Student.objects.count()
  csit_count = Student.objects.filter(course='CSIT').count()
  
  # Sum
  total_salary = Employee.objects.aggregate(total=Sum('salary'))
  
  # Average
  avg_age = Student.objects.aggregate(avg_age=Avg('age'))
  
  # Max/Min
  max_salary = Employee.objects.aggregate(max=Max('salary'))
  
  # Group by with annotations
  from django.db.models import Count
  courses = Course.objects.annotate(
      student_count=Count('students')
  ).filter(student_count__gt=10)''')
        
        print("\n🔹 Ordering:")
        print("─" * 70)
        print('''
  # Ascending
  students = Student.objects.order_by('name')
  
  # Descending
  students = Student.objects.order_by('-name')
  
  # Multiple fields
  students = Student.objects.order_by('course', '-age')
  
  # Random order
  random_student = Student.objects.order_by('?').first()''')
        
        print("\n🔹 Slicing and Limiting:")
        print("─" * 70)
        print('''
  # Limit results
  first_five = Student.objects.all()[:5]
  
  # Skip and limit (pagination)
  page = Student.objects.all()[10:20]  # Records 11-20
  
  # First/Last
  first_student = Student.objects.first()
  last_student = Student.objects.last()''')
        
        print("\n🔹 Bulk Operations:")
        print("─" * 70)
        print('''
  # Bulk Create
  students = [
      Student(name='Ram', age=20),
      Student(name='Sita', age=22),
  ]
  Student.objects.bulk_create(students)
  
  # Bulk Update
  Student.objects.filter(course='CSIT').update(is_active=True)
  
  # Bulk Delete
  Student.objects.filter(is_active=False).delete()''')
        
        print()
        print("═" * 70)

advanced = AdvancedOperations()
advanced.display()
```

    <h2 style="color: #ffc107;">🔧 Practical: Building a Complete System</h2>
    <p><strong>University Management System with all relationship types</strong></p>
</div>

```python
import datetime
from typing import List, Optional

# Simulated Django Models for demonstration
class User:
    def __init__(self, username, email):
        self.username = username
        self.email = email
        self.id = None

class Profile:
    def __init__(self, user, bio='', location=''):
        self.user = user
        self.bio = bio
        self.location = location
        self.id = None

class Department:
    def __init__(self, name, code):
        self.name = name
        self.code = code
        self.id = None

class Student:
    def __init__(self, name, email, department):
        self.name = name
        self.email = email
        self.department = department
        self.id = None
        self.courses = []

class Course:
    def __init__(self, name, code, credits, instructor):
        self.name = name
        self.code = code
        self.credits = credits
        self.instructor = instructor
        self.id = None
        self.students = []

class UniversitySystem:
    """Complete university management system"""
    
    def __init__(self):
        self.students = []
        self.courses = []
        self.departments = []
        self.users = []
        self.profiles = []
        self.next_id = 1
    
    def create_department(self, name, code):
        dept = Department(name, code)
        dept.id = self.next_id
        self.next_id += 1
        self.departments.append(dept)
        return dept
    
    def create_student(self, name, email, department):
        student = Student(name, email, department)
        student.id = self.next_id
        self.next_id += 1
        self.students.append(student)
        return student
    
    def create_course(self, name, code, credits, instructor):
        course = Course(name, code, credits, instructor)
        course.id = self.next_id
        self.next_id += 1
        self.courses.append(course)
        return course
    
    def enroll_student(self, student, course):
        if student not in course.students:
            course.students.append(student)
            student.courses.append(course)
            return True
        return False
    
    def display_students(self):
        print("📚 ALL STUDENTS:")
        print("─" * 70)
        for student in self.students:
            dept_name = student.department.name if student.department else 'No Department'
            print(f"  #{student.id}: {student.name} | {student.email} | {dept_name}")
            if student.courses:
                courses = ', '.join([c.name for c in student.courses])
                print(f"    📖 Courses: {courses}")
    
    def display_courses(self):
        print("📚 ALL COURSES:")
        print("─" * 70)
        for course in self.courses:
            print(f"  #{course.id}: {course.name} ({course.code}) | Credits: {course.credits}")
            if course.students:
                students = ', '.join([s.name for s in course.students])
                print(f"    👨‍🎓 Students: {students}")
    
    def display_departments(self):
        print("📚 ALL DEPARTMENTS:")
        print("─" * 70)
        for dept in self.departments:
            dept_students = [s for s in self.students if s.department == dept]
            print(f"  #{dept.id}: {dept.name} ({dept.code}) | Students: {len(dept_students)}")

class UniversityDemo:
    """Run university system demonstration"""
    
    def run_demo(self):
        print("\n" + "═" * 70)
        print(" " * 15 + "🎓 UNIVERSITY MANAGEMENT SYSTEM DEMO" + " " * 15)
        print("═" * 70)
        print()
        
        system = UniversitySystem()
        
        print("🔹 CREATING DEPARTMENTS:")
        print("─" * 70)
        dept1 = system.create_department("Computer Science", "CS")
        dept2 = system.create_department("Business Administration", "BA")
        dept3 = system.create_department("Engineering", "ENG")
        print(f"  ✅ Created: {dept1.name} ({dept1.code})")
        print(f"  ✅ Created: {dept2.name} ({dept2.code})")
        print(f"  ✅ Created: {dept3.name} ({dept3.code})")
        
        print("\n🔹 CREATING STUDENTS (One-to-Many with Department):")
        print("─" * 70)
        students_data = [
            ("Ram Sharma", "ram@cs.com", dept1),
            ("Sita Patel", "sita@cs.com", dept1),
            ("Hari Gurung", "hari@ba.com", dept2),
            ("Gita Adhikari", "gita@eng.com", dept3),
        ]
        for name, email, dept in students_data:
            student = system.create_student(name, email, dept)
            print(f"  ✅ {student.name} → {dept.name}")
        
        print("\n🔹 CREATING COURSES:")
        print("─" * 70)
        courses_data = [
            ("Python Programming", "PY101", 3, "Prof. Sharma"),
            ("Data Structures", "DS101", 4, "Prof. Thapa"),
            ("Business Management", "BM101", 3, "Prof. Gurung"),
            ("Database Systems", "DB101", 3, "Prof. Poudel"),
        ]
        created_courses = []
        for name, code, credits, instructor in courses_data:
            course = system.create_course(name, code, credits, instructor)
            created_courses.append(course)
            print(f"  ✅ {course.name} ({course.code})")
        
        print("\n🔹 ENROLLING STUDENTS (Many-to-Many):")
        print("─" * 70)
        # Enroll students in courses
        enrollments = [
            (system.students[0], created_courses[0]),  # Ram → Python
            (system.students[0], created_courses[1]),  # Ram → Data Structures
            (system.students[1], created_courses[0]),  # Sita → Python
            (system.students[1], created_courses[3]),  # Sita → Database
            (system.students[2], created_courses[2]),  # Hari → Business
            (system.students[3], created_courses[1]),  # Gita → Data Structures
            (system.students[3], created_courses[3]),  # Gita → Database
        ]
        for student, course in enrollments:
            if system.enroll_student(student, course):
                print(f"  ✅ {student.name} enrolled in {course.name}")
        
        print("\n" + "─" * 70)
        system.display_departments()
        
        print("\n" + "─" * 70)
        system.display_students()
        
        print("\n" + "─" * 70)
        system.display_courses()
        
        print("\n🔹 RELATIONSHIP SUMMARY:")
        print("─" * 70)
        print("""
  🏛️ One-to-Many: Department → Students
     • One department has many students
     • Each student belongs to one department
  
  🔗 Many-to-Many: Students ↔ Courses
     • Students can take many courses
     • Courses can have many students
  
  📋 One-to-One: User → Profile (extendable)
     • Each user has one profile
     • Each profile belongs to one user""")
        
        print()
        print("═" * 70)
        print("🎯 All relationship types implemented successfully!")
        print("═" * 70)

demo = UniversityDemo()
demo.run_demo()
```

    <p><strong>Optimize your database queries for better performance</strong></p>

```python
class PerformanceOptimization:
    """Display performance optimization techniques"""
    
    def display(self):
        print("\n" + "═" * 70)
        print(" " * 20 + "⚡ PERFORMANCE OPTIMIZATION" + " " * 20)
        print("═" * 70)
        print()
        
        print("🔹 select_related() - For ForeignKey relationships:")
        print("─" * 70)
        print('''
  # ❌ Without select_related (N+1 queries)
  for student in Student.objects.all():
      print(student.department.name)  # Hits database each time
  
  # ✅ With select_related (1 query)
  for student in Student.objects.select_related('department').all():
      print(student.department.name)  # Uses cached data''')
        
        print("\n🔹 prefetch_related() - For ManyToMany relationships:")
        print("─" * 70)
        print('''
  # ❌ Without prefetch_related (N+1 queries)
  for student in Student.objects.all():
      print(student.courses.all())  # Hits database each time
  
  # ✅ With prefetch_related (2 queries)
  for student in Student.objects.prefetch_related('courses').all():
      print(student.courses.all())  # Uses cached data''')
        
        print("\n🔹 only() and defer() - Limit fields:")
        print("─" * 70)
        print('''
  # Only load specific fields
  students = Student.objects.only('name', 'email')
  
  # Defer loading specific fields
  students = Student.objects.defer('bio', 'profile_picture')''')
        
        print("\n🔹 exists() vs count():")
        print("─" * 70)
        print('''
  # ❌ Inefficient
  if len(Student.objects.filter(course='CSIT')) > 0:
      # Does full query
  
  # ✅ Efficient
  if Student.objects.filter(course='CSIT').exists():
      # Only checks existence''')
        
        print("\n🔹 Values and ValuesList:")
        print("─" * 70)
        print('''
  # Get dictionary of values
  data = Student.objects.values('name', 'age')
  
  # Get list of values
  names = Student.objects.values_list('name', flat=True)''')
        
        print()
        print("═" * 70)

optimization = PerformanceOptimization()
optimization.display()
```
